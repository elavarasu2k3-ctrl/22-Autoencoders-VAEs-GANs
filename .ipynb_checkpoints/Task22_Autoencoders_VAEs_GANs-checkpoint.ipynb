# Task 22 · Representation Learning & Generative Models
### Autoencoders, VAEs and GANs – eight phases, 14 exercises, three capstones

**What this notebook does.** It follows the Task 22 brief phase by phase. First *autoencoders*: a network trained to copy its input discovers a compact coding (basic, stacked, tied-weight, convolutional, recurrent, denoising, sparse), which is used for dimensionality reduction, visualization and unsupervised pretraining. Then *generation*: variational autoencoders that sample new images from a Gaussian latent space, and generative adversarial networks (simple GAN, DCGAN, conditional GAN, progressive-growing layers, a toy StyleGAN-style generator) – together with an honest look at why GANs are hard to train.

| Phase | Question | Main data |
|---|---|---|
| 1 | What is a coding? Does a linear autoencoder really do PCA? | synthetic 3D points near a plane |
| 2 | How good is a stacked autoencoder, and how do I look at its codings? | Fashion MNIST |
| 3 | Does unsupervised pretraining help with 500 labels? What do tied weights save? | MNIST, Fashion MNIST |
| 4 | Convolutional and recurrent autoencoders | Fashion MNIST, synthetic time series |
| 5 | Denoising and sparse autoencoders (ℓ1 and KL) | Fashion MNIST |
| 6 | Variational autoencoders: sampling, interpolation, latent loss | Fashion MNIST |
| 7 | A first GAN, two training loops, and the ways GAN training fails | Fashion MNIST |
| 8 | DCGAN, conditional GAN, progressive-growing layers, AdaIN / style mixing, the three capstones | Fashion MNIST, MNIST |

**Honesty rules used throughout.** (1) Anything stochastic is repeated with 3 seeds and shown as *mean ± sample std*; a gap smaller than the seed spread is reported as "not distinguishable". (2) Every "proof" (a hand calculation, an identity, a test of a custom layer) is an `audit(...)` call; all of them are collected on a scoreboard in the wrap-up. (3) GAN runs are noisy by nature, so every sample grid and loss curve is shown – also the ones that went wrong. (4) Sizes are scaled for a 2-core CPU, so the GANs are trained for far fewer steps than in the book; this is stated wherever it matters.

**One switch.** `QUICK` (in Section 0) shrinks every training run to seconds so you can check that the code runs. The saved outputs of this notebook come from a run with `QUICK = False`.

### 0.2 Small helpers
* `audit(label, ok, detail)` – a visible proof, collected on the scoreboard. `audit_trained(...)` is the same for claims about *how well a model trained*; in `QUICK` mode it is skipped.
* `mean_std`, `pm` – mean and *sample* standard deviation (`ddof=1`) and the "m ± s" text; `noise_verdict` says whether a gap is bigger than the seed noise.
* `fit_quiet` – `model.fit` without progress bars; returns the history and the seconds per epoch.
* `download` – fetches a file once into `task22_data/`.

```python
class Scoreboard:
    def __init__(self):
        self.rows, self.phase = [], "setup"

    def audit(self, label, ok, detail=""):
        ok = bool(ok)
        self.rows.append(dict(phase=self.phase, proof=label, passed=ok, detail=str(detail)))
        print(("[ ok ] " if ok else "[FAIL] ") + label + (f"   ({detail})" if str(detail) != "" else ""))

    def table(self):
        return pd.DataFrame(self.rows)


SB = Scoreboard()
audit = SB.audit


def audit_trained(label, cond_fn, detail=""):
    if QUICK:
        SB.rows.append(dict(phase=SB.phase, proof=label, passed=True, detail="not evaluated in QUICK mode"))
        print("[skip] " + label + "   (QUICK mode)")
        return
    SB.audit(label, cond_fn(), detail() if callable(detail) else detail)


RESULTS, CHANGELOG, EX = {}, [], {}


def log_change(old, new, why):
    CHANGELOG.append(dict(old_api=old, used_here=new, reason=why))


def set_seed(s):
    random.seed(s); np.random.seed(s); tf.random.set_seed(s)
    try:
        keras.utils.set_random_seed(s)
    except Exception:
        pass


def fresh():
    keras.backend.clear_session(); gc.collect()


def mean_std(v):
    v = np.asarray(v, float)
    return float(v.mean()), (float(v.std(ddof=1)) if len(v) > 1 else float("nan"))


def pm(v, digits=4, pct=False):
    m, s = mean_std(v)
    if pct:
        m, s = 100 * m, 100 * s
        digits = min(digits, 2)
    suffix = " %" if pct else ""
    return f"{m:.{digits}f}{suffix} ± {s:.{digits}f}{suffix}" if not np.isnan(s) else f"{m:.{digits}f}{suffix} (1 seed)"


def noise_verdict(a, b):
    """gap between two lists of seed results (b - a), compared with the larger seed std"""
    ma, sa = mean_std(a); mb, sb = mean_std(b)
    gap = mb - ma
    noise = np.nanmax([sa, sb]) if not (np.isnan(sa) and np.isnan(sb)) else float("nan")
    if np.isnan(noise):
        return "based on one seed only, so the noise cannot be judged", gap, noise
    return ("larger than the seed noise" if abs(gap) > 2 * noise else "within the seed noise (not distinguishable)"), gap, noise


class EpochClock(keras.callbacks.Callback):
    def __init__(self):
        super().__init__(); self.times = []

    def on_epoch_begin(self, epoch, logs=None):
        self.t0 = time.time()

    def on_epoch_end(self, epoch, logs=None):
        self.times.append(time.time() - self.t0)

    @property
    def per_epoch(self):
        return float(np.mean(self.times)) if self.times else float("nan")


def fit_quiet(model, *args, **kwargs):
    clock = EpochClock()
    cbs = list(kwargs.pop("callbacks", [])) + [clock]
    h = model.fit(*args, verbose=0, callbacks=cbs, **kwargs)
    return h.history, clock.per_epoch


plt.rcParams.update({"figure.dpi": 100, "axes.grid": True, "grid.alpha": 0.25, "axes.spines.top": False, "axes.spines.right": False, "figure.figsize": (6.4, 3.4)})

DATA_DIR, OUT_DIR = Path("task22_data"), Path("task22_outputs")
DATA_DIR.mkdir(exist_ok=True); OUT_DIR.mkdir(exist_ok=True)
GH = "https://raw.githubusercontent.com"


def download(urls, dest, timeout=120):
    dest = Path(dest)
    if dest.exists() and dest.stat().st_size > 0:
        return dest
    problems = []
    for u in urls:
        try:
            req = urllib.request.Request(u, headers={"User-Agent": "Mozilla/5.0"})
            with urllib.request.urlopen(req, timeout=timeout) as r, open(str(dest) + ".part", "wb") as f:
                while chunk := r.read(1 << 20):
                    f.write(chunk)
            os.replace(str(dest) + ".part", dest)
            return dest
        except Exception as e:
            problems.append(f"{u.split('/')[2]} ({type(e).__name__})")
    raise RuntimeError(f"could not get {dest.name}: " + "; ".join(problems))


def show_images(imgs, titles=None, ncols=None, size=1.1, cmap="binary", suptitle=None):
    """a row (or grid) of small 28x28 pictures; values in [0, 1]"""
    imgs = np.asarray(imgs); n = len(imgs); ncols = ncols or n; nrows = int(math.ceil(n / ncols))
    fig, axes = plt.subplots(nrows, ncols, figsize=(size * ncols, size * nrows * 1.05), squeeze=False)
    for i, ax in enumerate(axes.ravel()):
        ax.axis("off")
        if i < n:
            ax.imshow(np.squeeze(imgs[i]), cmap=cmap, vmin=0, vmax=1)
            if titles is not None:
                ax.set_title(str(titles[i]), fontsize=7)
    if suptitle:
        fig.suptitle(suptitle, fontsize=9)
    plt.tight_layout(); plt.show()
```

### 0.3 Data

```python
def read_idx(path):
    with gzip.open(path, "rb") as f:
        raw = f.read()
    ndim = raw[3]
    dims = [int.from_bytes(raw[4 + 4 * i: 8 + 4 * i], "big") for i in range(ndim)]
    return np.frombuffer(raw, dtype=np.uint8, offset=4 + 4 * ndim).reshape(dims)


def get_fashion():
    names = ["train-images-idx3-ubyte.gz", "train-labels-idx1-ubyte.gz", "t10k-images-idx3-ubyte.gz", "t10k-labels-idx1-ubyte.gz"]
    p = [download([f"{GH}/zalandoresearch/fashion-mnist/master/data/fashion/{n}", f"https://storage.googleapis.com/tensorflow/tf-keras-datasets/{n}"], DATA_DIR / n) for n in names]
    return (read_idx(p[0]), read_idx(p[1]).astype(np.int64)), (read_idx(p[2]), read_idx(p[3]).astype(np.int64))


def get_mnist():
    p = download([f"{GH}/mnielsen/neural-networks-and-deep-learning/master/data/mnist.pkl.gz"], DATA_DIR / "mnist.pkl.gz")
    with gzip.open(p, "rb") as f:
        tr, va, te = pickle.load(f, encoding="latin1")
    to_img = lambda X: np.round(X.reshape(-1, 28, 28) * 255).astype(np.uint8)
    return (np.concatenate([to_img(tr[0]), to_img(va[0])]), np.concatenate([tr[1], va[1]]).astype(np.int64)), (to_img(te[0]), te[1].astype(np.int64))


(fx, fy), (fxt, fyt) = get_fashion()
(mx, my), (mxt, myt) = get_mnist()
FASHION_CLASSES = ["T-shirt", "Trouser", "Pullover", "Dress", "Coat", "Sandal", "Shirt", "Sneaker", "Bag", "Boot"]
f32 = lambda a: (a.astype("float32") / 255.0)
X_all_fashion, y_all_fashion = f32(fx), fy                                   # 60 000 training images (used when "all unlabeled data" is meant)
X_train, y_train = f32(fx[:55000])[:SZ["ae_train"]], fy[:55000][:SZ["ae_train"]]
X_valid, y_valid = f32(fx[55000:])[:SZ["ae_valid"]], fy[55000:][:SZ["ae_valid"]]
X_test, y_test = f32(fxt), fyt
M_train_all, m_y_all = f32(mx), my                                           # MNIST: 60 000 training images
M_test, m_y_test = f32(mxt), myt
print("Fashion MNIST train/valid/test:", X_train.shape, X_valid.shape, X_test.shape, "| pixel range", float(X_train.min()), "–", float(X_train.max()))
print("MNIST train/test:", M_train_all.shape, M_test.shape)
audit("Fashion MNIST has the expected sizes (60 000 train, 10 000 test, 10 classes) and pixels in [0, 1]", fx.shape == (60000, 28, 28) and fxt.shape == (10000, 28, 28) and len(set(fy)) == 10 and X_train.max() <= 1.0)
audit("MNIST has the expected sizes (60 000 train incl. the 10 000 validation images, 10 000 test)", mx.shape == (60000, 28, 28) and mxt.shape == (10000, 28, 28))
```

### 1.1 The idea of a representation (Tasks 1–4, 8)
**Number sequences.** Try to remember `40, 27, 25, 36, 81, 57, 10, 73, 19, 68` and then `50, 48, 46, 44, …, 14`. The second one is far easier: you do not store 19 numbers, you store the *pattern* ("go down by 2") plus the start and the end. A good *representation* of the data is exactly such a short description from which the data can be rebuilt.

**Chess experts (Chase and Simon, 1973).** Chess masters look at a real position for five seconds and place almost all pieces correctly afterwards; on a *random* arrangement they do no better than beginners. Their memory is not better – they see patterns (well-known formations) and remember those. An autoencoder does the same: it only has room for a few numbers, so it is forced to find the patterns.

**What an autoencoder is.** An **encoder** (recognition network) turns an input into a **latent representation / coding**. A **decoder** (generative network) turns the coding back into a **reconstruction** with the same shape as the input. The training cost is a **reconstruction loss** that punishes the difference between input and output. The architecture is an ordinary MLP, except that the output layer has as many neurons as the input layer. Since the target equals the input, this is **self-supervised learning**: supervised training with labels that are made automatically.

| | **Undercomplete** | **Overcomplete** |
|---|---|---|
| coding size | smaller than the input | as large as or larger than the input |
| what happens | cannot copy the input trivially, so it must keep only the most important features | can learn the identity function and copy the input without learning anything |
| what is needed | enough capacity, but not too much | an extra constraint: noise (Phase 5), sparsity (Phase 5), or a probabilistic coding (Phase 6) |

**The constraints built in this task:** (1) a size limit on the coding (Phases 1–4), (2) noise on the input (denoising), (3) a sparsity penalty (sparse), plus (4) the probabilistic variational autoencoder (Phase 6).

### 1.2 A linear autoencoder on 3D data near a plane (Tasks 5–6)
The data: 200 points near a **2D plane** inside 3D (an arc plus a little noise). The autoencoder is `Dense(2)` → `Dense(3)`, with **no activation functions**, MSE loss and SGD with `learning_rate=0.1`. Inputs and targets are the same array.

```python
def generate_3d_data(m, w1=0.1, w2=0.3, noise=0.1, seed=42):
    rng = np.random.default_rng(seed)
    angles = rng.random(m) * 3 * np.pi / 2 - 0.5
    X = np.empty((m, 3))
    X[:, 0] = np.cos(angles) + np.sin(angles) / 2 + noise * rng.standard_normal(m)
    X[:, 1] = np.sin(angles) * 0.7 + noise * rng.standard_normal(m)
    X[:, 2] = X[:, 0] * w1 + X[:, 1] * w2 + noise * rng.standard_normal(m)
    return X.astype("float32")


X3 = generate_3d_data(200)
X3 = X3 - X3.mean(axis=0, keepdims=True)                         # centre the data (PCA centres it too)
print("data:", X3.shape, "| mean after centring:", np.round(X3.mean(axis=0), 6))

set_seed(42); fresh()
encoder3 = keras.Sequential([keras.Input(shape=(3,)), L.Dense(2)])
decoder3 = keras.Sequential([keras.Input(shape=(2,)), L.Dense(3)])
autoencoder3 = keras.Sequential([encoder3, decoder3])
autoencoder3.compile(loss="mse", optimizer=keras.optimizers.SGD(learning_rate=0.1))
h20, _ = fit_quiet(autoencoder3, X3, X3, epochs=20)
mse_20 = float(autoencoder3.evaluate(X3, X3, verbose=0))
print(f"after the 20 epochs asked for in the brief: reconstruction MSE (per number) = {mse_20:.5f}")
h_more, _ = fit_quiet(autoencoder3, X3, X3, epochs=480)
mse_500 = float(autoencoder3.evaluate(X3, X3, verbose=0))
print(f"after 500 epochs in total:                  reconstruction MSE (per number) = {mse_500:.5f}")
fig, ax = plt.subplots(figsize=(5.2, 2.8)); ax.plot(h20["loss"] + h_more["loss"]); ax.set_yscale("log"); ax.axvline(20, ls=":", color="gray"); ax.set_xlabel("epoch"); ax.set_ylabel("MSE"); ax.set_title("linear autoencoder: training loss"); plt.show()
```

```python
codings3 = encoder3.predict(X3, verbose=0)
fig = plt.figure(figsize=(9, 3.6))
ax = fig.add_subplot(1, 2, 1, projection="3d"); ax.scatter(*X3.T, s=8, c=X3[:, 0], cmap="viridis"); ax.set_title("the 3D data (near a plane)"); ax.set_xlabel("x1"); ax.set_ylabel("x2"); ax.set_zlabel("x3")
ax = fig.add_subplot(1, 2, 2); ax.scatter(codings3[:, 0], codings3[:, 1], s=10, c=X3[:, 0], cmap="viridis"); ax.set_title("the 2D codings"); ax.set_xlabel("z1"); ax.set_ylabel("z2"); ax.set_aspect("equal")
plt.tight_layout(); plt.show()
```

### 1.3 It is PCA (Task 7)
Fit Task 10's `PCA(n_components=2)` on the same data and compare (a) the reconstruction error and (b) the plane that is spanned. A linear autoencoder may use *any* basis of the plane, so the comparison of the **subspaces** (principal angles) is the right one, not a comparison of the individual weights.

```python
from scipy.linalg import subspace_angles

pca3 = PCA(n_components=2).fit(X3)
mse_pca = float(np.mean((pca3.inverse_transform(pca3.transform(X3)) - X3) ** 2))
W_dec = decoder3.layers[0].get_weights()[0]            # shape (2, 3): the two rows span the plane the decoder writes into
b_dec = decoder3.layers[0].get_weights()[1]
angles_deg = np.degrees(subspace_angles(W_dec.T, pca3.components_.T))
# the autoencoder's rows are an invertible mixture of the PCA directions: W_dec = A @ components
A, *_ = np.linalg.lstsq(pca3.components_.T, W_dec.T, rcond=None)
resid = np.abs(pca3.components_.T @ A - W_dec.T).max()
tab = pd.DataFrame({"reconstruction MSE (per number)": [mse_20, mse_500, mse_pca]}, index=["linear autoencoder, 20 epochs", "linear autoencoder, 500 epochs", "PCA (2 components)"])
display(tab.round(5))
print("principal angles between the decoder's plane and the PCA plane (degrees):", np.round(angles_deg, 3))
print("decoder rows written as a mixture of the PCA directions, A =\n", np.round(A.T, 3), "\n  (A is not a rotation: A·Aᵀ =\n", np.round(A.T @ A, 3), ")")
audit("500-epoch linear autoencoder reaches the PCA reconstruction error (within 2 %)", abs(mse_500 - mse_pca) / mse_pca < 0.02, f"AE {mse_500:.5f} vs PCA {mse_pca:.5f}")
audit("the decoder spans the same plane as PCA (both principal angles < 2°)", angles_deg.max() < 2.0, f"largest angle {angles_deg.max():.3f}°")
audit("the decoder rows are (almost exactly) a mixture of the PCA directions (max residual < 0.05) – same plane, different basis", resid < 0.05, f"max residual {resid:.1e}")
audit("the individual decoder rows are NOT the PCA directions (the basis differs)", not np.allclose(np.abs(A.T), np.eye(2), atol=0.05), f"A = {np.round(A.T, 2).tolist()}")
RESULTS["lin_ae"] = dict(mse_20=mse_20, mse_500=mse_500, mse_pca=mse_pca, angle=float(angles_deg.max()))
```

**Why the bases differ.** The loss only asks that `decoder(encoder(x))` is close to `x`. If `E` is the encoder and `D` the decoder, then `E' = M·E` and `D' = D·M⁻¹` give *exactly* the same reconstruction for every invertible 2×2 matrix `M`. The autoencoder therefore ends up with a random mixture of the principal directions, while PCA picks the one orthogonal basis ordered by variance. Same plane, same error, different coordinates – as the audits above show.

### 1.4 Too powerful is a failure mode (Task 9)
An encoder that is allowed to be a huge, flexible network can map each training input to **one arbitrary number** and the decoder can learn the inverse look-up table. The training error then becomes tiny, but nothing useful has been learned. To see this, take only **30** points of the plane data, force a **1-number coding**, and compare a modest linear model (PCA with one component) with a big non-linear network. Both are tested on 2 000 *new* points from the same source. (The points lie on an arc, so a curved 1-number coding can in principle describe them well – the thing to look at is the **gap between training error and new-data error**, not only which model wins.)

```python
Xs_all = generate_3d_data(2230, seed=7); Xs_all -= Xs_all.mean(axis=0)
Xs_train, Xs_test = Xs_all[:30], Xs_all[30:]
pca1 = PCA(n_components=1).fit(Xs_train)
mse_pca1_tr = float(np.mean((pca1.inverse_transform(pca1.transform(Xs_train)) - Xs_train) ** 2))
mse_pca1_te = float(np.mean((pca1.inverse_transform(pca1.transform(Xs_test)) - Xs_test) ** 2))
set_seed(3); fresh()
big = keras.Sequential([keras.Input(shape=(3,)), L.Dense(256, activation="relu"), L.Dense(256, activation="tanh"), L.Dense(1), L.Dense(256, activation="tanh"), L.Dense(256, activation="relu"), L.Dense(3)])
big.compile(loss="mse", optimizer=keras.optimizers.Adam(learning_rate=2e-3))
fit_quiet(big, Xs_train, Xs_train, epochs=pick(1500, 200), batch_size=30)
mse_big_tr = float(big.evaluate(Xs_train, Xs_train, verbose=0)); mse_big_te = float(big.evaluate(Xs_test, Xs_test, verbose=0))
display(pd.DataFrame({"train MSE": [mse_pca1_tr, mse_big_tr], "new-data MSE": [mse_pca1_te, mse_big_te]}, index=["PCA, 1 component (small)", "huge non-linear AE, 1 number (30 training points)"]).round(5))
audit_trained("the huge autoencoder fits the 30 training points far better than the small model (train MSE at least 3× lower)", lambda: mse_big_tr * 3 < mse_pca1_tr, f"{mse_big_tr:.5f} vs {mse_pca1_tr:.5f}")
audit_trained("... but its error on new points is much higher than on its training points (at least 5×) – it memorised", lambda: mse_big_te > 5 * mse_big_tr, f"new {mse_big_te:.5f} vs train {mse_big_tr:.5f}")
print("Lesson: good training error alone proves nothing; judge an autoencoder on held-out data (Phase 2, task 8).")
```

**Phase 1 summary.** A coding is a short description from which the input can be rebuilt. A linear undercomplete autoencoder with MSE loss learns the same plane as PCA (reconstruction error equal, angle between planes ≈ 0°) but in its own basis. Giving the network too much power destroys the point of the exercise: it memorises.

# Phase 2 · Stacked autoencoders, reconstructions and visualization

```python
SB.phase = "Phase 2"
from sklearn.neighbors import NearestNeighbors
```

### 2.1 A stacked (deep) autoencoder
A **stacked** or **deep** autoencoder has several hidden layers. It is usually **symmetric** around the central coding layer – a "sandwich": `784 → 100 → 30 → 100 → 784`. More layers let it learn more complex codings, but too much depth/capacity brings back the memorisation problem of Phase 1. Hidden layers use **SELU**, the output layer uses **sigmoid** (pixels are in [0, 1]), and the loss is **binary cross-entropy**.

**Why binary cross-entropy and not MSE?** Every output pixel is read as *the probability that this pixel should be black*, so reconstructing an image is a multi-label binary classification with 784 labels. With a sigmoid output, BCE gives a simple gradient (`prediction − target`), while MSE multiplies it by the sigmoid slope and gets small gradients when the unit saturates – so BCE tends to converge faster. The next cells measure that instead of just claiming it.

```python
def build_stacked(hidden=(100,), coding=30, act="selu", name="stacked"):
    """784 -> hidden... -> coding -> reversed hidden -> 784  (returns encoder, decoder, autoencoder)"""
    enc = keras.Sequential([keras.Input(shape=(28, 28)), L.Flatten()] + [L.Dense(h, activation=act) for h in hidden] + [L.Dense(coding, activation=act)], name=name + "_encoder")
    dec = keras.Sequential([keras.Input(shape=(coding,))] + [L.Dense(h, activation=act) for h in reversed(hidden)] + [L.Dense(784, activation="sigmoid"), L.Reshape([28, 28])], name=name + "_decoder")
    return enc, dec, keras.Sequential([enc, dec], name=name)


def rounded_accuracy(y_true, y_pred):
    """pixel 'accuracy' after rounding targets AND predictions to 0 / 1"""
    return keras.metrics.binary_accuracy(tf.round(y_true), tf.round(y_pred))


def pixel_mse(model, X, batch=1000):
    pred = model.predict(X, batch_size=batch, verbose=0)
    return float(np.mean((pred - X) ** 2))


def show_reconstructions(model, images=None, n_images=5, title=None):
    images = X_valid[:n_images] if images is None else images[:n_images]
    recon = model.predict(images, verbose=0)
    fig, axes = plt.subplots(2, n_images, figsize=(1.15 * n_images, 2.5))
    for i in range(n_images):
        for r, im in enumerate((images[i], recon[i])):
            axes[r, i].imshow(np.squeeze(im), cmap="binary", vmin=0, vmax=1); axes[r, i].axis("off")
    axes[0, 0].set_title("input", fontsize=7, loc="left"); axes[1, 0].set_title("reconstruction", fontsize=7, loc="left")
    if title:
        fig.suptitle(title, fontsize=9)
    plt.tight_layout(); plt.show()


def train_ae(build, loss="binary_crossentropy", lr=1.5, epochs=None, seed=0, X=None, Xv=None, batch_size=None, optimizer=None, metrics=None, **bkw):
    X = X_train if X is None else X; Xv = X_valid if Xv is None else Xv
    set_seed(seed); fresh()
    enc, dec, ae = build(**bkw)
    ae.compile(loss=loss, optimizer=optimizer or keras.optimizers.SGD(learning_rate=lr), metrics=metrics or [])
    hist, sec = fit_quiet(ae, X, X, epochs=epochs or SZ["ae_epochs"], batch_size=batch_size or SZ["ae_batch"], validation_data=(Xv, Xv))
    return enc, dec, ae, hist, sec
```

### 2.2 Find a learning rate, then train (Tasks 2–3)
The book uses SGD with the large learning rate 1.5. A short sweep (2 epochs, one seed) finds the best value for **each** loss, so the comparison afterwards is fair.

```python
lr_grid = [0.3, 1.0, 3.0, 6.0, 12.0, 24.0, 48.0]
sweep_rows = []
for loss_name in ("binary_crossentropy", "mse"):
    for lr in lr_grid:
        try:
            *_, ae_, h_, _ = train_ae(build_stacked, loss=loss_name, lr=lr, epochs=pick(2, 1), seed=0)
            ae_.compile(loss="mse", optimizer="sgd")
            vm = h_["val_loss"][-1] if loss_name == "binary_crossentropy" else h_["val_loss"][-1]
            pm_ = pixel_mse(ae_, X_valid)
        except Exception as e:
            vm, pm_ = float("nan"), float("nan")
        sweep_rows.append(dict(loss=loss_name, lr=lr, val_loss=vm, val_pixel_mse=pm_))
sweep = pd.DataFrame(sweep_rows)
display(sweep.pivot(index="lr", columns="loss", values="val_pixel_mse").round(5).rename(columns=lambda c: f"validation pixel-MSE after short run ({c})"))
best_lr = {k: float(g.loc[g.val_pixel_mse.fillna(9).idxmin(), "lr"]) for k, g in sweep.groupby("loss")}
print("chosen learning rates:", best_lr)
```

**Train the stacked autoencoder** (3 seeds, same architecture), once with BCE and once with MSE, each at its own best learning rate. `rounded_accuracy` is the custom metric explained in 2.3.

```python
runs = {"binary_crossentropy": [], "mse": []}
models_bce = []
for loss_name in runs:
    for s in SZ["seeds"]:
        enc, dec, ae, h, sec = train_ae(build_stacked, loss=loss_name, lr=best_lr[loss_name], seed=s, metrics=[rounded_accuracy])
        runs[loss_name].append(dict(seed=s, hist=h, sec=sec, pix_mse=pixel_mse(ae, X_valid), acc=float(h["val_rounded_accuracy"][-1])))
        if loss_name == "binary_crossentropy":
            models_bce.append((enc, dec, ae))
        else:
            del enc, dec, ae
tab = pd.DataFrame({k: {"learning rate": best_lr[k], "validation pixel-MSE": pm([r["pix_mse"] for r in v], 5), "rounded accuracy": pm([r["acc"] for r in v], 4), "s / epoch": round(float(np.mean([r["sec"] for r in v])), 1)} for k, v in runs.items()}).T
display(tab)
mse_bce = [r["pix_mse"] for r in runs["binary_crossentropy"]]; mse_mse = [r["pix_mse"] for r in runs["mse"]]
verdict, gap, noise = noise_verdict(mse_bce, mse_mse)
print(f"pixel-MSE: MSE-trained minus BCE-trained = {gap:+.5f} – {verdict} (seed std ≈ {noise:.5f})")
RESULTS["bce_vs_mse"] = dict(bce=mse_bce, mse=mse_mse)
fig, ax = plt.subplots(1, 2, figsize=(9, 3))
for k, c in (("binary_crossentropy", "C0"), ("mse", "C1")):
    for i, r in enumerate(runs[k]):
        ax[0].plot(r["hist"]["val_loss"], color=c, alpha=0.7, label=k if i == 0 else None)
ax[0].set_title("validation loss (each in its own units)"); ax[0].set_xlabel("epoch"); ax[0].legend(fontsize=7)
ax[1].boxplot([mse_bce, mse_mse], tick_labels=["BCE", "MSE"]); ax[1].set_title("validation pixel-MSE after training (3 seeds)"); plt.tight_layout(); plt.show()
audit_trained("the stacked autoencoder reconstructs well: validation pixel-MSE below 0.02 (the 'predict the mean image' baseline is much higher)", lambda: max(mse_bce) < 0.02, f"BCE-trained: {pm(mse_bce, 5)}")
mean_img_mse = float(np.mean((X_train.mean(axis=0, keepdims=True) - X_valid) ** 2))
audit("the 'predict the mean training image' baseline has a larger pixel-MSE than the trained autoencoder", min(mse_bce) < mean_img_mse, f"baseline {mean_img_mse:.4f} vs AE {pm(mse_bce, 4)}")
```

### 2.3 Why `accuracy` fails and the rounded-accuracy metric (Task 4)
Keras' plain `"accuracy"` metric compares predicted and true **values** for exact equality (or class indices). A sigmoid output like `0.9731` is never exactly equal to a target like `1.0`, so the metric is almost always ~0 and tells you nothing. The suggested fix rounds both the targets and the predictions to 0/1 first (`rounded_accuracy`), so it asks "did we get the pixel's black/white decision right?".

```python
enc0, dec0, ae0 = models_bce[0]
pred_v = ae0.predict(X_valid, verbose=0)
m_plain = keras.metrics.Accuracy(); m_plain.update_state(X_valid, pred_v)
m_round = keras.metrics.Accuracy(); m_round.update_state(np.round(X_valid), np.round(pred_v))
print(f"exact-equality accuracy on the raw values : {float(m_plain.result()):.4f}")
print(f"rounded accuracy (black/white decisions)   : {float(m_round.result()):.4f}")
audit("exact-equality 'accuracy' on sigmoid outputs is meaningless (< 0.05) while rounded accuracy is high (> 0.8)", float(m_plain.result()) < 0.05 and float(m_round.result()) > 0.8, f"{float(m_plain.result()):.3f} vs {float(m_round.result()):.3f}")
```

`show_reconstructions` plots validation images (top) and their reconstructions (bottom).

**What is lost.** The outline and the kind of garment are clearly recognisable, but thin details are smoothed away: textures and patterns, labels, fine edges and the exact grey level inside large areas look washed out. ### 2.5 Fixes, one at a time (Task 6)
Start from the baseline (100 → 30, 10 epochs) and change **one** thing: train **longer** (2× epochs), make it **deeper** (300 → 100 → 30 → 100 → 300), or make the **coding larger** (30 → 64). Validation pixel-MSE, 3 seeds each.

```python
variants = {
    "baseline (100-30, 10 epochs)": None,
    "train longer (2× epochs)": dict(epochs=2 * SZ["ae_epochs"]),
    "deeper (300-100-30)": dict(hidden=(300, 100)),
    "larger coding (64)": dict(coding=64),
}
fix_rows, fix_vals = [], {}
for name, kw in variants.items():
    vals, params = [], None
    for i, s in enumerate(SZ["seeds"]):
        if kw is None:
            vals.append(runs["binary_crossentropy"][i]["pix_mse"]); params = models_bce[i][2].count_params(); continue
        if i == 0:                                        # each new architecture gets its own short learning-rate search (a deeper net needs a smaller step)
            bkw0 = {k: v for k, v in kw.items() if k != "epochs"}
            if bkw0:
                trial = {lr: pixel_mse(train_ae(build_stacked, lr=lr, epochs=pick(2, 1), seed=0, **bkw0)[2], X_valid) for lr in (0.3, 1.0, 3.0, 6.0)}
                lr_v = min(trial, key=lambda k: (trial[k] if trial[k] == trial[k] else 9)); print(f"{name}: learning-rate search {({k: round(v, 4) for k, v in trial.items()})} → {lr_v}")
            else:
                lr_v = best_lr["binary_crossentropy"]
        enc, dec, ae, h, _ = train_ae(build_stacked, lr=lr_v, seed=s, **kw)
        vals.append(pixel_mse(ae, X_valid)); params = ae.count_params()
    fix_vals[name] = vals
    fix_rows.append({"change": name, "parameters": params, "validation pixel-MSE": pm(vals, 5), "mean": np.mean(vals)})
fix_tab = pd.DataFrame(fix_rows).set_index("change")
display(fix_tab[["parameters", "validation pixel-MSE"]])
base = fix_vals["baseline (100-30, 10 epochs)"]
for name in list(variants)[1:]:
    v, g, n = noise_verdict(base, fix_vals[name]); print(f"{name:28s}: gap to baseline {g:+.5f} – {v}")
RESULTS["ae_fixes"] = fix_vals
audit_trained("a larger coding (64) reconstructs better than the 30-number baseline", lambda: np.mean(fix_vals["larger coding (64)"]) < np.mean(base), f"{np.mean(fix_vals['larger coding (64)']):.5f} vs {np.mean(base):.5f}")
```

**The trade-off.** Every fix improves the *reconstruction* because it gives the network more room (more parameters, more training, or a bigger bottleneck). But the autoencoder is only useful if it is **constrained enough to learn patterns** instead of copying. Powerful enough to reconstruct well, constrained enough to learn something – that balance is the design problem of this whole task. A bigger coding is the quickest way to a better picture and also the quickest way to learn nothing.

### 2.6 Visualization: autoencoder → t-SNE (Task 7)
t-SNE is excellent for 2D pictures but does not scale to many features or points. The standard recipe: let the autoencoder do the heavy lifting (784 → 30 numbers, fast, scales to huge data sets), then run t-SNE on the 30-dimensional codings.

```python
enc0 = models_bce[0][0]
n_t = SZ["tsne_points"]
codes_v = enc0.predict(X_valid[:n_t], verbose=0)
t0 = time.time()
tsne = TSNE(n_components=2, init="pca", learning_rate="auto", random_state=42).fit_transform(codes_v)
print(f"t-SNE on {n_t} codings of 30 numbers: {time.time() - t0:.0f} s")
yv = y_valid[:n_t]
fig, ax = plt.subplots(figsize=(7, 5.4))
sc = ax.scatter(tsne[:, 0], tsne[:, 1], c=yv, cmap="tab10", s=6)
for c in range(10):
    cx, cy = np.median(tsne[yv == c], axis=0); ax.text(cx, cy, FASHION_CLASSES[c], fontsize=8, weight="bold", ha="center", bbox=dict(boxstyle="round,pad=0.15", fc="white", alpha=0.7, lw=0))
ax.set_title("t-SNE of the 30-number codings (Fashion MNIST validation images)"); ax.axis("off"); plt.show()
# how well do the classes separate?  fraction of the 5 nearest neighbours (in the 2D picture) that share the class
nn = NearestNeighbors(n_neighbors=6).fit(tsne); idx = nn.kneighbors(tsne, return_distance=False)[:, 1:]
purity = np.array([(yv[idx[yv == c]] == c).mean() for c in range(10)])
pur_tab = pd.Series(purity, index=FASHION_CLASSES, name="share of 5 nearest neighbours with the same class").sort_values(ascending=False)
display(pur_tab.round(3).to_frame())
mix = pd.crosstab(pd.Series(yv, name="true class").map(dict(enumerate(FASHION_CLASSES))), pd.Series(yv[idx[:, 0]], name="nearest neighbour is").map(dict(enumerate(FASHION_CLASSES))))
pairs = [(a, b, int(mix.loc[a, b])) for a in mix.index for b in mix.columns if a != b]
pairs.sort(key=lambda t: -t[2])
print("most common confusions in the picture (class → class of its nearest neighbour):", [(a, b, n) for a, b, n in pairs[:4]])
print(f"best separated classes: {list(pur_tab.index[:3])};  most overlapping: {list(pur_tab.index[-3:])}")
audit("the 2D picture is much better than random: the average neighbour purity is above 0.5 (random would be about 0.1)", purity.mean() > 0.5, f"{purity.mean():.3f}")
RESULTS["tsne_purity"] = pur_tab.to_dict()
```

**Reading the picture.** The classes named under "best separated" form tidy islands; the classes listed under "most overlapping" – above all the upper-body garments (T-shirt, shirt, pullover, coat) – sit on top of each other, because in 28×28 pixels they really do look alike. That is not a failure of t-SNE: the same pairs are the ones that a classifier confuses (compare the printed most common confusions).

### 2.7 Is the autoencoder good? (Task 8, Exercise 3)
**If an autoencoder reconstructs its inputs perfectly, is it necessarily good? No.** An overcomplete or too-powerful autoencoder can simply learn the identity function or a look-up table (Phase 1.4). The extreme case costs two lines: an "autoencoder" that is just `Flatten` followed by `Reshape` has reconstruction error **exactly 0** and a coding of 784 numbers – it has learned nothing.

**Ways to evaluate an autoencoder** (all five measured below on the stacked autoencoder):
1. **Reconstruction error on held-out data** (not on the training set).
2. **Visual inspection** of reconstructions (2.4).
3. **Downstream task** on the codings: does a simple classifier do well on 30 numbers?
4. **Quality of a codings visualization** (2.6).
5. **Behaviour on corrupted inputs** (does the reconstruction stay sensible when the input is damaged?).

```python
identity = keras.Sequential([keras.Input(shape=(28, 28)), L.Flatten(), L.Reshape([28, 28])])
id_err = pixel_mse(identity, X_valid)
audit("a 'perfect' identity autoencoder has reconstruction error exactly 0 while learning no representation at all", id_err == 0.0, f"MSE = {id_err}")

ntr = min(len(X_train), 5000)
Z_tr, Z_te = enc0.predict(X_train[:ntr], verbose=0), enc0.predict(X_valid, verbose=0)
pca30 = PCA(n_components=30).fit(X_train[:ntr].reshape(ntr, -1))
P_tr, P_te = pca30.transform(X_train[:ntr].reshape(ntr, -1)), pca30.transform(X_valid.reshape(len(X_valid), -1))
def lin_acc(A_tr, A_te):
    mu, sd = A_tr.mean(0), A_tr.std(0) + 1e-6
    return float(LogisticRegression(max_iter=300).fit((A_tr - mu) / sd, y_train[:ntr]).score((A_te - mu) / sd, y_valid))
acc_ae, acc_pca, acc_raw = lin_acc(Z_tr, Z_te), lin_acc(P_tr, P_te), lin_acc(X_train[:ntr].reshape(ntr, -1), X_valid.reshape(len(X_valid), -1))
rng = np.random.default_rng(0)
noisy = np.clip(X_valid + rng.normal(0, 0.3, X_valid.shape), 0, 1).astype("float32")
mse_clean = pixel_mse(models_bce[0][2], X_valid)
pred_noisy = models_bce[0][2].predict(noisy, verbose=0)
mse_noisy_vs_clean = float(np.mean((pred_noisy - X_valid) ** 2)); mse_noisy_input = float(np.mean((noisy - X_valid) ** 2))
ev = pd.DataFrame({
    "1 held-out reconstruction pixel-MSE": [f"{mse_clean:.5f}"],
    "3 linear classifier on the 30-number coding": [f"{acc_ae:.3f}"],
    "3 ... on 30 PCA components": [f"{acc_pca:.3f}"],
    "3 ... on the 784 raw pixels": [f"{acc_raw:.3f}"],
    "5 noisy input → reconstruction vs. CLEAN image (MSE)": [f"{mse_noisy_vs_clean:.5f}"],
    "5 (for comparison: the noisy input itself vs. clean)": [f"{mse_noisy_input:.5f}"]}).T.rename(columns={0: "value"})
display(ev)
audit("a plain linear classifier on only 30 autoencoder numbers reaches above 70 % accuracy (it kept the information that matters)", acc_ae > 0.70, f"{acc_ae:.3f}")
audit_trained("on noisy input the reconstruction is closer to the clean image than the noisy input itself (the bottleneck filters noise)", lambda: mse_noisy_vs_clean < mse_noisy_input, f"{mse_noisy_vs_clean:.4f} < {mse_noisy_input:.4f}")
EX[3] = ("done", "section 2.7 (+ identity example in 1.4 and 2.7)")
RESULTS["eval_ae"] = dict(acc_ae=acc_ae, acc_pca=acc_pca, acc_raw=acc_raw)
```

**Phase 2 summary.** A 784-100-30-100-784 SELU/sigmoid autoencoder trained with BCE reconstructs the validation images recognisably; MSE vs BCE is compared with each loss at its own learning rate and judged against the seed spread; the fixes (longer, deeper, larger coding) trade reconstruction for freedom to memorise; t-SNE on 30 numbers is a quick, readable picture; and *held-out* error, a downstream classifier and noisy input tests are the real evidence – not a perfect reconstruction.

# Phase 3 · Unsupervised pretraining, tying weights and layer-wise training

```python
SB.phase = "Phase 3"
```

### 3.1 Unsupervised pretraining with a stacked autoencoder (Task 1)
Labels are expensive, images are cheap. Suppose you have 60 000 images but only 500 of them have labels. **Plan:** (1) train an autoencoder on *all* images, labelled or not – it needs no labels; (2) cut off its **encoder** and use it as the **lower layers** of a classifier; (3) add a `Dense(10, softmax)` on top and train on the 500 labels. The encoder already knows which pixel patterns matter, so the 500 labels only have to teach the last step.
**Freeze or fine-tune?** With very little labelled data, freezing at least the lowest pretrained layers is safer (the few labels cannot spoil them). With more labels, fine-tuning usually wins. The next experiment measures both.

### 3.2 Experiment: 500 labels on MNIST (Task 2 – part of Exercise 9)
* The autoencoder (784 → 200 → 64 → 200 → 784, SELU/sigmoid, BCE, Adam) is trained on **all 60 000 MNIST training images**, without labels.
* The classifier is the same encoder + `Dense(10, softmax)`, trained on **500 labelled images (50 per class)**, 40 epochs, Adam, in three configurations: **(a)** random initialisation, **(b)** pretrained encoder, frozen, **(c)** pretrained encoder, fine-tuned.
* Evaluation on the 10 000 MNIST test images; 3 seeds (the seed changes the pretraining, the choice of the 500 images and the classifier's initial weights).

```python
def balanced_subset(y, per_class, rng):
    return np.concatenate([rng.choice(np.flatnonzero(y == c), per_class, replace=False) for c in range(10)])


def make_classifier(encoder, trainable, lr=1e-3):
    clf = keras.Sequential([encoder, L.Dense(10, activation="softmax")])
    encoder.trainable = trainable
    clf.compile(loss="sparse_categorical_crossentropy", optimizer=keras.optimizers.Adam(learning_rate=lr), metrics=["accuracy"])
    return clf


def clone_encoder(enc):
    new = keras.models.clone_model(enc); new.set_weights(enc.get_weights()); return new


pre_rows, pre_models = [], {}
for s in SZ["seeds"]:
    set_seed(s); fresh()
    idx = balanced_subset(m_y_all, SZ["pre_labels"] // 10, np.random.default_rng(s))
    Xl, yl = M_train_all[idx], m_y_all[idx]
    enc, dec, ae = build_stacked(hidden=(200,), coding=64, name="pre")
    ae.compile(loss="binary_crossentropy", optimizer=keras.optimizers.Adam(learning_rate=1e-3))
    fit_quiet(ae, M_train_all, M_train_all, epochs=SZ["pre_ae_epochs"], batch_size=32)
    rec_mse = pixel_mse(ae, M_test)
    for name, (pretrained, trainable) in {"(a) random initialisation": (False, True), "(b) pretrained, frozen": (True, False), "(c) pretrained, fine-tuned": (True, True)}.items():
        set_seed(s)
        e = clone_encoder(enc) if pretrained else keras.models.clone_model(enc)           # clone_model without set_weights = fresh random weights
        if pretrained and not trainable:
            assert all(np.allclose(a, b) for a, b in zip(e.get_weights(), enc.get_weights()))
        clf = make_classifier(e, trainable)
        before = [w.copy() for w in e.get_weights()]
        fit_quiet(clf, Xl, yl, epochs=SZ["pre_clf_epochs"], batch_size=32)
        acc = float(clf.evaluate(M_test, m_y_test, verbose=0)[1])
        changed = max(float(np.abs(a - b).max()) for a, b in zip(before, e.get_weights()))
        pre_rows.append(dict(seed=s, config=name, acc=acc, enc_changed=changed, ae_mse=rec_mse))
pre = pd.DataFrame(pre_rows)
tab = pre.groupby("config", sort=False).agg(acc=("acc", lambda v: pm(v, pct=True)), mean=("acc", "mean"))
display(tab[["acc"]].rename(columns={"acc": f"MNIST test accuracy, {SZ['pre_labels']} labels, {len(SZ['seeds'])} seed(s)"}))
a_ = pre[pre.config.str.startswith("(a)")].acc.values; b_ = pre[pre.config.str.startswith("(b)")].acc.values; c_ = pre[pre.config.str.startswith("(c)")].acc.values
for nm, v in (("(b) frozen", b_), ("(c) fine-tuned", c_)):
    verdict, gap, noise = noise_verdict(a_, v); print(f"{nm:15s} minus random init: {100 * gap:+.2f} points – {verdict}")
RESULTS["pretrain500"] = dict(random=list(a_), frozen=list(b_), tuned=list(c_))
audit("the frozen encoder really stayed frozen (weights unchanged) and the fine-tuned and random ones really changed", pre[pre.config.str.startswith("(b)")].enc_changed.max() == 0.0 and pre[pre.config.str.startswith("(c)")].enc_changed.min() > 0 and pre[pre.config.str.startswith("(a)")].enc_changed.min() > 0, f"max change frozen = {pre[pre.config.str.startswith('(b)')].enc_changed.max()}")
audit_trained("pretraining helps: the fine-tuned pretrained encoder beats random initialisation with 500 labels (mean over seeds)", lambda: c_.mean() > a_.mean(), f"{100 * c_.mean():.2f} % vs {100 * a_.mean():.2f} %")
```

**Conclusion.** The table and the printed gaps above are the answer: compare each pretrained variant with the random start and see whether the gap is larger than the seed spread. Remember what this is *not*: a test of pretraining with a lot of labels – there the gap shrinks, because enough labels can teach the lower layers by themselves.

### 3.3 Tying weights (Tasks 3–6)
In a symmetric autoencoder with `N` layers (not counting the input), where layer `N/2` is the coding layer, the decoder weights can be defined as the **transpose** of the encoder weights: `W_{N−L+1} = W_Lᵀ` for `L = 1 … N/2`. Biases are *not* shared. **Why:** the weights are halved, training is faster, and there is less overfitting.

**`DenseTranspose`** is a custom layer that holds a reference to a `Dense` layer, creates its **own bias** in `build()` (its length is the Dense layer's **input** size, because the output of the transposed layer lives in the input space), and in `call()` computes `x · Wᵀ + b` followed by the activation. `tf.matmul(x, W, transpose_b=True)` is more efficient than `tf.matmul(x, tf.transpose(W))` because the transpose is folded into the matrix multiplication and no transposed copy of `W` is made.

```python
class DenseTranspose(L.Layer):
    def __init__(self, dense, activation=None, **kwargs):
        super().__init__(**kwargs)
        self.dense = dense
        self.activation = keras.activations.get(activation)

    def build(self, batch_input_shape):
        self.biases = self.add_weight(name="bias", shape=[self.dense.kernel.shape[0]], initializer="zeros")
        super().build(batch_input_shape)

    def call(self, inputs):
        z = tf.matmul(inputs, self.dense.kernel, transpose_b=True)
        return self.activation(z + self.biases)


def build_tied():
    d1, d2 = L.Dense(100, activation="selu"), L.Dense(30, activation="selu")
    enc = keras.Sequential([keras.Input(shape=(28, 28)), L.Flatten(), d1, d2], name="tied_encoder")
    dec = keras.Sequential([keras.Input(shape=(30,)), DenseTranspose(d2, activation="selu"), DenseTranspose(d1, activation="sigmoid"), L.Reshape([28, 28])], name="tied_decoder")
    return enc, dec, keras.Sequential([enc, dec], name="tied")


enc_u, dec_u, ae_u = build_stacked()
enc_t, dec_t, ae_t = build_tied()
n_u, n_t_ = ae_u.count_params(), ae_t.count_params()
print(f"untied: {n_u:,} parameters   (78,500 + 3,030 + 3,100 + 79,184 = {78500 + 3030 + 3100 + 79184:,})")
print(f"tied:   {n_t_:,} parameters   (78,500 + 3,030 + 100 + 784 = {78500 + 3030 + 100 + 784:,})")
audit("untied autoencoder has exactly 163,814 parameters (the brief's check value)", n_u == 163_814, n_u)
audit("tied autoencoder has exactly 82,414 parameters (the brief's check value)", n_t_ == 82_414, n_t_)
audit("tying nearly halves the parameter count", n_t_ < 0.51 * n_u, f"{n_t_ / n_u:.3f} of the untied size")
```

```python
# train both with the same optimizer and epochs, 3 seeds
tie_rows = []
for s in SZ["seeds"]:
    for name, builder in (("untied", build_stacked), ("tied", build_tied)):
        enc, dec, ae, h, sec = train_ae(builder, optimizer=keras.optimizers.Adam(learning_rate=1e-3), seed=s)
        tie_rows.append(dict(model=name, seed=s, pix_mse=pixel_mse(ae, X_valid), sec=sec, params=ae.count_params()))
        if name == "tied":
            last_tied = (enc, dec, ae)
tie = pd.DataFrame(tie_rows)
display(tie.groupby("model", sort=False).agg(parameters=("params", "first"), validation_pixel_MSE=("pix_mse", lambda v: pm(v, 5)), seconds_per_epoch=("sec", lambda v: round(float(np.mean(v)), 1))))
u_, t_ = tie[tie.model == "untied"].pix_mse.values, tie[tie.model == "tied"].pix_mse.values
verdict, gap, noise = noise_verdict(u_, t_); print(f"tied minus untied pixel-MSE: {gap:+.5f} – {verdict}")
RESULTS["tied"] = dict(untied=list(u_), tied=list(t_))
```

**Test: the weights are literally shared, not just equal.** (1) The decoder layer holds the *same object* as the encoder's `Dense`. (2) After training, the decoder's first layer computes exactly `x · W_encoderᵀ + b`. (3) If the encoder's kernel is overwritten, the decoder's output changes accordingly – which could not happen if it were a copy.

```python
enc, dec, ae = last_tied
d_enc2 = enc.layers[-1]; dt1 = dec.layers[0]
audit("the DenseTranspose layer references the encoder's own Dense layer object (not a copy)", dt1.dense is d_enc2)
x = np.random.default_rng(0).standard_normal((4, 30)).astype("float32")
manual = tf.keras.activations.selu(x @ d_enc2.kernel.numpy().T + dt1.biases.numpy())
audit("after training, the decoder layer equals selu(x · W_encoderᵀ + b) computed by hand", np.allclose(dt1(x).numpy(), manual.numpy(), atol=1e-5), f"max diff {np.abs(dt1(x).numpy() - manual.numpy()).max():.1e}")
old_kernel = d_enc2.kernel.numpy().copy()
new_kernel = np.random.default_rng(1).standard_normal(old_kernel.shape).astype("float32") * 0.1
d_enc2.kernel.assign(new_kernel)
changed_out = dt1(x).numpy(); expected = tf.keras.activations.selu(x @ new_kernel.T + dt1.biases.numpy()).numpy()
audit("overwriting the ENCODER kernel changes the DECODER output exactly as the transpose predicts (truly shared)", np.allclose(changed_out, expected, atol=1e-5) and not np.allclose(changed_out, manual.numpy(), atol=1e-3))
d_enc2.kernel.assign(old_kernel)
own_vars = [w for w in dec.weights if id(w) not in {id(v) for v in enc.weights}]
audit("the only decoder variables that are not encoder variables are the two bias vectors (100 + 784 numbers) – no decoder kernels exist", sum(int(np.prod(w.shape)) for w in own_vars) == 100 + 784 and len(own_vars) == 2, f"{len(own_vars)} variables, {sum(int(np.prod(w.shape)) for w in own_vars)} numbers")
EX[5] = ("done", "section 3.3 (rule, DenseTranspose, parameter counts 163,814 vs 82,414, sharing test)")
```

### 3.4 Greedy layer-wise training (Tasks 7–9)
**Procedure.** (1) Train a first, *shallow* autoencoder (784 → 100 → 784) on the inputs. (2) Use its encoder to turn the whole training set into 100-number codings. (3) Train a second autoencoder (100 → 30 → 100) on those codings. (4) Stack the hidden layers of both (784 → 100 → 30) and then the output layers in reverse order (30 → 100 → 784). The first autoencoder uses BCE (pixels), the second uses MSE (its inputs are unbounded SELU outputs, not probabilities).

**History.** Hinton et al. (2006) showed that deep nets can be pretrained layer by layer without labels (with restricted Boltzmann machines); Bengio et al. (2007) showed that autoencoders do the same job. For some years this was the only efficient way to train deep networks, until the tricks of Task 16 (initialisation, SELU/ReLU, Batch Norm) made one-shot training of deep nets possible. Today it is rarely used, but you meet it in papers. Its spirit lives on in **self-supervised pretraining**, e.g. the masked-word language models of Task 21: learn from unlabeled data first, specialise later.

```python
def greedy_ae(X, X_val, seed, epochs):
    set_seed(seed); fresh(); t0 = time.time()
    e1 = keras.Sequential([keras.Input(shape=(28, 28)), L.Flatten(), L.Dense(100, activation="selu")], name="g_enc1")
    d1 = keras.Sequential([keras.Input(shape=(100,)), L.Dense(784, activation="sigmoid"), L.Reshape([28, 28])], name="g_dec1")
    a1 = keras.Sequential([e1, d1]); a1.compile(loss="binary_crossentropy", optimizer=keras.optimizers.Adam(learning_rate=1e-3))
    fit_quiet(a1, X, X, epochs=epochs, batch_size=SZ["ae_batch"])
    H = e1.predict(X, batch_size=1000, verbose=0)
    e2 = keras.Sequential([keras.Input(shape=(100,)), L.Dense(30, activation="selu")], name="g_enc2")
    d2 = keras.Sequential([keras.Input(shape=(30,)), L.Dense(100, activation="selu")], name="g_dec2")
    a2 = keras.Sequential([e2, d2]); a2.compile(loss="mse", optimizer=keras.optimizers.Adam(learning_rate=1e-3))
    fit_quiet(a2, H, H, epochs=epochs, batch_size=SZ["ae_batch"])
    stacked = keras.Sequential([keras.Input(shape=(28, 28)), L.Flatten(), e1.layers[-1], e2.layers[-1], d2.layers[-1], d1.layers[0], L.Reshape([28, 28])], name="greedy")
    return stacked, time.time() - t0, a1, a2


half = max(1, SZ["ae_epochs"] // 2)
gr_rows = []
for s in SZ["seeds"]:
    g, sec_g, a1, a2 = greedy_ae(X_train, X_valid, s, half)
    mse_g = pixel_mse(g, X_valid)
    g.compile(loss="binary_crossentropy", optimizer=keras.optimizers.Adam(learning_rate=1e-3)); t0 = time.time()
    fit_quiet(g, X_train, X_train, epochs=max(1, SZ["ae_epochs"] // 5), batch_size=SZ["ae_batch"]); sec_f = time.time() - t0
    mse_gf = pixel_mse(g, X_valid)
    enc, dec, ae, h, sec_e = train_ae(build_stacked, optimizer=keras.optimizers.Adam(learning_rate=1e-3), seed=s, epochs=SZ["ae_epochs"])
    gr_rows += [dict(method="greedy layer-wise (2 × %d epochs)" % half, seed=s, pix_mse=mse_g, sec=sec_g), dict(method="greedy + short end-to-end fine-tuning", seed=s, pix_mse=mse_gf, sec=sec_g + sec_f),
                dict(method="end-to-end, same architecture (%d epochs)" % SZ["ae_epochs"], seed=s, pix_mse=pixel_mse(ae, X_valid), sec=sec_e * SZ["ae_epochs"])]
    if s == SZ["seeds"][0]:
        greedy_model = g
gr = pd.DataFrame(gr_rows)
display(gr.groupby("method", sort=False).agg(validation_pixel_MSE=("pix_mse", lambda v: pm(v, 5)), training_seconds=("sec", lambda v: round(float(np.mean(v)), 1))))
e2e = gr[gr.method.str.startswith("end-to-end")].pix_mse.values; grd = gr[gr.method.str.startswith("greedy layer")].pix_mse.values
verdict, gap, noise = noise_verdict(e2e, grd); print(f"greedy minus end-to-end pixel-MSE: {gap:+.5f} – {verdict}")
RESULTS["greedy"] = dict(greedy=list(grd), e2e=list(e2e))
audit("the greedily built stack has the same architecture as the end-to-end autoencoder (same parameter count)", greedy_model.count_params() == build_stacked()[2].count_params(), f"{greedy_model.count_params():,}")
audit_trained("the greedy stack reconstructs far better than the mean-image baseline (pixel-MSE below half the baseline)", lambda: grd.mean() < 0.5 * mean_img_mse, f"{grd.mean():.4f} vs baseline {mean_img_mse:.4f}")
```

**Reading the table.** Compare the three rows: greedy training needs two small training runs, and without a joint polishing step it is usually no better than end-to-end training of the same architecture; a few end-to-end epochs on top close most of the gap. That is why one-shot training took over once it became reliable.

**Phase 3 summary.** Unsupervised pretraining is measured against a random start with only 500 labels (frozen and fine-tuned, three seeds). Tying weights with `DenseTranspose` shrinks the autoencoder from 163,814 to 82,414 parameters and the weights are provably shared. Greedy layer-wise training is implemented and compared with end-to-end training.

### 4.1 Why convolutions for images (Task 1)
A dense autoencoder treats an image as a flat list of 784 numbers: it has to learn separately, for every pixel position, what an edge looks like, and its parameter count explodes with image size. For anything larger than tiny images a **convolutional autoencoder** is better (Task 20): the **encoder** is an ordinary CNN – convolutions plus pooling that shrink height and width while the depth (number of feature maps) grows – and the **decoder** does the reverse with **transposed convolutions** (or upsampling followed by a convolution).

```python
def build_conv_ae(name="conv"):
    enc = keras.Sequential([keras.Input(shape=(28, 28)), L.Reshape([28, 28, 1]),
                            L.Conv2D(16, 3, padding="same", activation="selu"), L.MaxPool2D(2),
                            L.Conv2D(32, 3, padding="same", activation="selu"), L.MaxPool2D(2),
                            L.Conv2D(64, 3, padding="same", activation="selu"), L.MaxPool2D(2)], name=name + "_encoder")
    dec = keras.Sequential([keras.Input(shape=(3, 3, 64)),
                            L.Conv2DTranspose(32, 3, strides=2, padding="valid", activation="selu"),
                            L.Conv2DTranspose(16, 3, strides=2, padding="same", activation="selu"),
                            L.Conv2DTranspose(1, 3, strides=2, padding="same", activation="sigmoid"), L.Reshape([28, 28])], name=name + "_decoder")
    return enc, dec, keras.Sequential([enc, dec], name=name)


def layer_shapes(seq, input_shape):
    x = tf.zeros((1,) + tuple(input_shape)); out = []
    for l in seq.layers:
        x = l(x); out.append((l.__class__.__name__, tuple(x.shape[1:])))
    return out


enc_c, dec_c, ae_c = build_conv_ae()
es, ds = layer_shapes(enc_c, (28, 28)), layer_shapes(dec_c, (3, 3, 64))
shape_tab = pd.DataFrame([("encoder",) + e for e in es] + [("decoder",) + d for d in ds], columns=["part", "layer", "output shape"])
display(shape_tab)
enc_spatial = [s[1][0] for s in es if s[0] in ("MaxPooling2D", "MaxPool2D")]
dec_spatial = [s[1][0] for s in ds if s[0] == "Conv2DTranspose"]
audit("encoder spatial size follows 28 → 14 → 7 → 3 (max pooling with 'valid' padding floors 7/2 = 3)", enc_spatial == [14, 7, 3], enc_spatial)
audit("decoder spatial size follows 3 → 7 → 14 → 28", dec_spatial == [7, 14, 28], dec_spatial)
audit("'valid' transposed convolution: (3 − 1)·2 + 3 = 7", (3 - 1) * 2 + 3 == 7 and ds[0][1][0] == 7)
audit("the coding is a 3 × 3 × 64 tensor (576 numbers)", enc_c.output_shape[1:] == (3, 3, 64))
```

**Shape audit, in words.** `Conv2D(…, "same")` keeps the size; each `MaxPool2D(2)` halves it: 28 → 14 → 7 → **3**, because with `"valid"` padding 7/2 is rounded down and the last row/column is lost. The decoder must get from 3 back to 28 with three stride-2 transposed convolutions. With `"same"` padding a stride-2 transposed convolution doubles the size, so 3 would become 6 → 12 → 24 – not 28. Therefore the first layer uses **`"valid"` padding**, for which the output size is `(3 − 1)·2 + 3 = 7`; after that "same" padding gives 7 → 14 → 28.

### 4.2 Train it and compare with a dense autoencoder (Tasks 3–4)
The convolutional autoencoder has a coding of **576 numbers** (3×3×64), so the fair dense partner is a dense autoencoder with a similar coding, `784 → 576 → 784` (`hidden=()`), not the 30-number one of Phase 2. Both use BCE and Adam; 3 seeds. The 30-number stacked autoencoder is shown as a reference.

```python
conv_models, rows4 = {}, []
Xc, Xcv = X_train[:SZ["conv_train"]], X_valid
for s in SZ["seeds"]:
    for name, builder, kw in (("convolutional (coding 3×3×64 = 576)", build_conv_ae, {}), ("dense, coding 576 (784-576-784)", build_stacked, dict(hidden=(), coding=576, act="selu")), ("dense stacked, coding 30 (Phase 2)", build_stacked, {})):
        enc, dec, ae, h, sec = train_ae(builder, optimizer=keras.optimizers.Adam(learning_rate=1e-3), seed=s, epochs=SZ["conv_epochs"], X=Xc, batch_size=64 if "conv" in name else 32, **kw)
        rows4.append(dict(model=name, seed=s, pix_mse=pixel_mse(ae, Xcv), params=ae.count_params(), sec=sec))
        if s == SZ["seeds"][0]:
            conv_models[name] = ae
t4 = pd.DataFrame(rows4)
display(t4.groupby("model", sort=False).agg(parameters=("params", "first"), validation_pixel_MSE=("pix_mse", lambda v: pm(v, 5)), seconds_per_epoch=("sec", lambda v: round(float(np.mean(v)), 1))))
RESULTS["conv_vs_dense"] = {k: list(g.pix_mse) for k, g in t4.groupby("model", sort=False)}
cm = t4[t4.model.str.startswith("conv")].pix_mse.values; dm = t4[t4.model.str.startswith("dense, coding 576")].pix_mse.values
verdict, gap, noise = noise_verdict(dm, cm); print(f"convolutional minus dense(576): {gap:+.5f} – {verdict}")
audit("the convolutional autoencoder has far fewer parameters than the dense one with a coding of similar size (at least 5× fewer)", t4[t4.model.str.startswith("dense, coding 576")].params.iloc[0] > 5 * t4[t4.model.str.startswith("conv")].params.iloc[0], f"{t4[t4.model.str.startswith('dense, coding 576')].params.iloc[0]:,} vs {t4[t4.model.str.startswith('conv')].params.iloc[0]:,}")
```

```python
imgs = X_valid[:8]
fig, axes = plt.subplots(4, 8, figsize=(9.2, 4.9))
rows_ = [("input", imgs)] + [(k, m.predict(imgs, verbose=0)) for k, m in conv_models.items()]
for r, (title, arr) in enumerate(rows_):
    for c in range(8):
        axes[r, c].imshow(arr[c], cmap="binary", vmin=0, vmax=1); axes[r, c].axis("off")
    axes[r, 0].set_title(title, fontsize=7, loc="left")
plt.tight_layout(); plt.show()
```

### 4.3 Upsampling + convolution instead of transposed convolution (Task 5)
Transposed convolutions with a kernel size that is not a multiple of the stride overlap unevenly and can leave a **checkerboard** pattern. The alternative decoder resizes first (`UpSampling2D`, nearest neighbour) and then smooths with an ordinary `Conv2D`. The coding (3×3×64) is padded to 7×7 with one row/column of zeros so that 7 → 14 → 28 works the same way.

```python
def build_conv_ae_up(name="conv_up"):
    enc, _, _ = build_conv_ae()
    dec = keras.Sequential([keras.Input(shape=(3, 3, 64)), L.UpSampling2D(2), L.ZeroPadding2D(((0, 1), (0, 1))), L.Conv2D(32, 3, padding="same", activation="selu"),
                            L.UpSampling2D(2), L.Conv2D(16, 3, padding="same", activation="selu"),
                            L.UpSampling2D(2), L.Conv2D(1, 3, padding="same", activation="sigmoid"), L.Reshape([28, 28])], name=name + "_decoder")
    return enc, dec, keras.Sequential([enc, dec], name=name)


def checker_score(recon, orig):
    """energy of the reconstruction error at the highest spatial frequency (period-2 'checkerboard'), relative to the energy over all frequencies"""
    F = np.abs(np.fft.fft2(recon - orig, axes=(1, 2)))
    return float(F[:, 14, 14].mean() / F.mean())


up_rows = []
for s in SZ["seeds"]:
    enc, dec, ae, h, sec = train_ae(build_conv_ae_up, optimizer=keras.optimizers.Adam(learning_rate=1e-3), seed=s, epochs=SZ["conv_epochs"], X=Xc, batch_size=64)
    up_rows.append(dict(seed=s, pix_mse=pixel_mse(ae, Xcv), checker=checker_score(ae.predict(Xcv[:1000], verbose=0), Xcv[:1000]), params=ae.count_params()))
    if s == SZ["seeds"][0]:
        ae_up = ae
up_ = pd.DataFrame(up_rows)
conv_ae_first = conv_models["convolutional (coding 3×3×64 = 576)"]
ck_conv = checker_score(conv_ae_first.predict(Xcv[:1000], verbose=0), Xcv[:1000])
display(pd.DataFrame({"parameters": [conv_ae_first.count_params(), ae_up.count_params()], "validation pixel-MSE": [pm(cm, 5), pm(up_.pix_mse, 5)],
                      "checkerboard score (lower = fewer artifacts)": [f"{ck_conv:.4f} (seed {SZ['seeds'][0]})", f"{up_.checker.iloc[0]:.4f} (seed {SZ['seeds'][0]})"]}, index=["transposed convolutions", "UpSampling2D + Conv2D"]))
verdict, gap, noise = noise_verdict(cm, up_.pix_mse.values); print(f"upsampling minus transposed: {gap:+.5f} – {verdict}")
RESULTS["conv_up"] = dict(transposed=list(cm), upsample=list(up_.pix_mse), checker=(ck_conv, float(up_.checker.iloc[0])))
show_images(np.concatenate([X_valid[:8], conv_ae_first.predict(X_valid[:8], verbose=0), ae_up.predict(X_valid[:8], verbose=0)]), ncols=8, size=1.0, suptitle="top: input · middle: transposed convolutions · bottom: UpSampling2D + Conv2D")
audit("the upsampling variant has the same coding shape and output shape as the transposed one", ae_up.output_shape == conv_ae_first.output_shape)
```

**How to read this.** Look at the middle and bottom rows: checkerboard artifacts show up as a faint regular grid on flat areas. The checkerboard score measures the same thing as a number (error energy at the highest spatial frequency); with small kernels and a few epochs of training the visible difference is modest, and the table above says whether it is bigger than the seed noise.

### 4.4 Recurrent autoencoder (Task 6)
For sequences, the **encoder is a sequence-to-vector RNN** (the last LSTM returns only its final output = the coding) and the **decoder is a vector-to-sequence RNN**. `RepeatVector` is the decoder's first layer because the decoder LSTM needs an input at *every* time step, and all it has is the one coding vector – so the same coding is fed at every step. A Fashion MNIST image is read as a **sequence of 28 rows of 28 pixels**.

```python
def build_rec_ae(coding=30, steps=28, feats=28, name="rec"):
    enc = keras.Sequential([keras.Input(shape=(None, feats)), L.LSTM(100, return_sequences=True), L.LSTM(coding)], name=name + "_encoder")
    dec = keras.Sequential([keras.Input(shape=(coding,)), L.RepeatVector(steps), L.LSTM(100, return_sequences=True), L.TimeDistributed(L.Dense(feats, activation="sigmoid"))], name=name + "_decoder")
    return enc, dec, keras.Sequential([enc, dec], name=name)


Xr = X_train[:SZ["rec_train"]]
rec_rows = []
for s in SZ["seeds"][:pick(2, 1)]:
    enc, dec, ae, h, sec = train_ae(build_rec_ae, optimizer=keras.optimizers.Adam(learning_rate=2e-3), seed=s, epochs=SZ["rec_epochs"], X=Xr, batch_size=64)
    rec_rows.append(dict(seed=s, pix_mse=pixel_mse(ae, X_valid), params=ae.count_params(), sec=sec))
    if s == SZ["seeds"][0]:
        rec_ae = ae
rec_t = pd.DataFrame(rec_rows)
dense30 = t4[t4.model.str.startswith("dense stacked")].pix_mse.values
display(pd.DataFrame({"parameters": [rec_ae.count_params(), 163814], "coding size": [30, 30], "validation pixel-MSE": [pm(rec_t.pix_mse, 5), pm(dense30, 5)], "seconds / epoch": [round(float(rec_t.sec.mean()), 1), "–"]}, index=["recurrent (LSTM 100 → LSTM 30)", "dense stacked 784-100-30"]))
show_reconstructions(rec_ae, n_images=8, title="recurrent autoencoder: each image read as 28 rows")
RESULTS["rec_ae"] = list(rec_t.pix_mse)
audit_trained("the recurrent autoencoder beats the 'mean image' baseline", lambda: rec_t.pix_mse.mean() < mean_img_mse, f"{rec_t.pix_mse.mean():.4f} vs {mean_img_mse:.4f}")
```

### 4.5 A recurrent autoencoder on time series (Task 7)
Synthetic series in the spirit of Task 20: 64 steps of a noisy sine wave with a random phase and amplitude, whose frequency falls into one of **three bands** (low, medium, high) – that is the "label". The recurrent autoencoder compresses each series to a **coding of 8 numbers**. We measure the reconstruction error and then use the codings as features to predict the frequency band, comparing with the raw 64 numbers and with 8 PCA components.

```python
def make_series(n, rng, steps=64):
    band = rng.integers(0, 3, n)
    lo, hi = np.array([0.02, 0.07, 0.16])[band], np.array([0.05, 0.12, 0.30])[band]
    f = rng.uniform(lo, hi); phase = rng.uniform(0, 2 * np.pi, n); amp = rng.uniform(0.5, 1.5, n)
    t = np.arange(steps)[None, :]
    x = amp[:, None] * np.sin(2 * np.pi * f[:, None] * t + phase[:, None]) + 0.15 * rng.standard_normal((n, steps))
    return x.astype("float32")[..., None], band


rng_ts = np.random.default_rng(5)
Xs_tr, ys_tr = make_series(SZ["ts_series"], rng_ts); Xs_te, ys_te = make_series(2000, rng_ts)
sc = float(Xs_tr.std()); Xs_tr, Xs_te = Xs_tr / sc, Xs_te / sc
print("series:", Xs_tr.shape, "| band frequencies (cycles per step): 0.02–0.05, 0.07–0.12, 0.16–0.30 | scale", round(sc, 3))
fig, ax = plt.subplots(1, 3, figsize=(10, 2.2), sharey=True)
for b in range(3):
    for i in np.flatnonzero(ys_tr == b)[:3]:
        ax[b].plot(Xs_tr[i, :, 0], lw=1)
    ax[b].set_title(["low band", "medium band", "high band"][b], fontsize=8)
plt.tight_layout(); plt.show()


def build_ts_ae(coding=8, steps=64):
    enc = keras.Sequential([keras.Input(shape=(steps, 1)), L.LSTM(32, return_sequences=True), L.LSTM(coding)], name="ts_encoder")
    dec = keras.Sequential([keras.Input(shape=(coding,)), L.RepeatVector(steps), L.LSTM(32, return_sequences=True), L.TimeDistributed(L.Dense(1))], name="ts_decoder")
    return enc, dec, keras.Sequential([enc, dec], name="ts_ae")


ts_rows = []
for s in SZ["seeds"]:
    set_seed(s); fresh()
    enc, dec, ae = build_ts_ae()
    ae.compile(loss="mse", optimizer=keras.optimizers.Adam(learning_rate=3e-3))
    fit_quiet(ae, Xs_tr, Xs_tr, epochs=SZ["ts_epochs"], batch_size=64)
    mse_ts = float(ae.evaluate(Xs_te, Xs_te, verbose=0))
    Ztr, Zte = enc.predict(Xs_tr, verbose=0), enc.predict(Xs_te, verbose=0)
    pca8 = PCA(8).fit(Xs_tr[:, :, 0]); pca_mse = float(np.mean((pca8.inverse_transform(pca8.transform(Xs_te[:, :, 0])) - Xs_te[:, :, 0]) ** 2))
    def clf_acc(A, B):
        mu, sd = A.mean(0), A.std(0) + 1e-6
        return float(LogisticRegression(max_iter=500).fit((A - mu) / sd, ys_tr).score((B - mu) / sd, ys_te))
    ts_rows.append(dict(seed=s, ae_mse=mse_ts, pca8_mse=pca_mse, acc_code=clf_acc(Ztr, Zte), acc_pca=clf_acc(pca8.transform(Xs_tr[:, :, 0]), pca8.transform(Xs_te[:, :, 0])), acc_raw=clf_acc(Xs_tr[:, :, 0], Xs_te[:, :, 0])))
ts = pd.DataFrame(ts_rows)
base_mse = float(np.mean((Xs_te - Xs_te.mean()) ** 2))
display(pd.DataFrame({"reconstruction MSE (test)": [pm(ts.ae_mse, 4), pm(ts.pca8_mse, 4), f"{base_mse:.4f}"],
                      "linear classifier → frequency band (test accuracy)": [pm(ts.acc_code, 3), pm(ts.acc_pca, 3), "chance = 0.333"]}, index=["recurrent AE, 8-number coding", "PCA, 8 components", "predict the mean / guess"]).assign(**{"…on the 64 raw numbers": ["–", "–", pm(ts.acc_raw, 3)]}))
RESULTS["ts_ae"] = dict(mse=list(ts.ae_mse), pca=list(ts.pca8_mse), acc_code=list(ts.acc_code), acc_raw=list(ts.acc_raw), acc_pca=list(ts.acc_pca))
audit_trained("the time-series autoencoder reconstructs far better than the 'predict the mean' baseline (error below 40 % of it)", lambda: ts.ae_mse.mean() < 0.4 * base_mse, f"{ts.ae_mse.mean():.4f} vs {base_mse:.4f}")
audit_trained("a plain linear classifier on the 8-number coding finds the frequency band clearly above chance (> 0.6)", lambda: ts.acc_code.mean() > 0.6, f"{ts.acc_code.mean():.3f}")
```

**Note on the failed proof above (kept on purpose).** The recurrent autoencoder does reconstruct better than guessing the mean (0.66 against 0.995), but my threshold "below 40 % of the baseline" was too strict: with an 8-number coding PCA reconstructs better still (0.50). What the recurrent coding does better is *content*: a linear classifier on its 8 numbers finds the frequency band with accuracy 0.95, against 0.51 for the 8 PCA components. So here the coding with the higher squared error is the more useful one – another case where reconstruction error and usefulness of a coding are different things.

### 4.6 Which autoencoder for which data? (Task 8)
| Flavour | Use it for | What the coding size controls |
|---|---|---|
| **Dense** (stacked, tied) | small or tabular inputs, flattened tiny images, features that have no neighbourhood structure | the number of numbers kept – a global compression ratio; too small → blurry, too large → copying |
| **Convolutional** | images (any size); fewer parameters, uses the 2D neighbourhood | the depth and the spatial size of the final feature maps (here 3×3×64), i.e. how much spatial detail survives |
| **Recurrent** | sequences: time series, text, rows of pixels; any length (one fixed `RepeatVector` length here) | the length of the summary vector for the *whole* sequence – the single number-list through which all temporal information must pass |
In this notebook the evidence is: 4.2 shows the parameter count and error of the convolutional model next to a dense model with a similar coding size, and 4.5 shows what an 8-number recurrent coding keeps of a 64-step series (its linear-classifier accuracy for the frequency band, compared with raw values and PCA).

**Phase 4 summary.** Shape arithmetic of the convolutional autoencoder is audited (28→14→7→3 and 3→7→14→28), it is compared with a dense autoencoder of similar coding size, the transposed-convolution decoder is compared with upsampling + convolution (with a checkerboard score), and a recurrent autoencoder is built for image rows and for synthetic time series, where its 8-number coding is tested as a feature set.

### 5.1 Denoising autoencoders (Tasks 1–2)
**Idea.** Another way to stop an autoencoder from simply copying: **add noise to the inputs and train it to recover the original, clean inputs.** The idea goes back to LeCun (1987); Vincent et al. (2008, 2010) used it for feature extraction and stacked denoising autoencoders. Two kinds of noise: pure **Gaussian noise** (`GaussianNoise` layer) or **randomly switched-off inputs** (`Dropout` on the inputs). Both layers are only active **during training**; at prediction time they pass the input through unchanged.

The first model: `Flatten → Dropout(0.5) → Dense(100, selu) → Dense(30, selu)` as the encoder, the usual decoder, and **clean** targets.

```python
def build_denoiser(kind="dropout", level=0.5, coding=30, name="dae"):
    layers_in = [keras.Input(shape=(28, 28)), L.Flatten()]
    if kind == "dropout":
        layers_in.append(L.Dropout(level))
    elif kind == "gaussian":
        layers_in.append(L.GaussianNoise(level))
    enc = keras.Sequential(layers_in + [L.Dense(100, activation="selu"), L.Dense(coding, activation="selu")], name=name + "_encoder")
    dec = keras.Sequential([keras.Input(shape=(coding,)), L.Dense(100, activation="selu"), L.Dense(784, activation="sigmoid"), L.Reshape([28, 28])], name=name + "_decoder")
    return enc, dec, keras.Sequential([enc, dec], name=name)


enc_d, dec_d, dae = train_ae(build_denoiser, optimizer=keras.optimizers.Adam(learning_rate=1e-3), seed=11, epochs=SZ["dn_epochs"], batch_size=64, kind="dropout", level=0.5)[:3]
drop_layer = enc_d.layers[1]
noisy_demo = drop_layer(tf.constant(X_valid[:8].reshape(8, 784)), training=True).numpy().reshape(8, 28, 28)       # dropout active, exactly as in training
clean_rec = dae.predict(X_valid[:8], verbose=0)
audit("Dropout is only active in training: the same layer returns the input unchanged at prediction time", np.allclose(drop_layer(tf.constant(X_valid[:8].reshape(8, 784)), training=False).numpy(), X_valid[:8].reshape(8, 784)))
audit("with training=True about half of the pixels are switched off (the kept ones are rescaled by 1/(1−0.5) = 2)", 0.4 < np.mean(noisy_demo == 0) < 0.9 and noisy_demo.max() > 1.5, f"share of exact zeros {np.mean(noisy_demo == 0):.2f} (the background is zero anyway), max value {noisy_demo.max():.2f}")
rec_noisy = dae.predict(noisy_demo, verbose=0)                       # the model sees the noisy image exactly as it did during training
fig, axes = plt.subplots(3, 8, figsize=(9.2, 3.7))
for r, (title, arr) in enumerate((("clean input", X_valid[:8]), ("noisy input (dropout 0.5, training=True)", np.clip(noisy_demo, 0, 1)), ("reconstruction from the NOISY input", rec_noisy))):
    for c in range(8):
        axes[r, c].imshow(arr[c], cmap="binary", vmin=0, vmax=1); axes[r, c].axis("off")
    axes[r, 0].set_title(title, fontsize=7, loc="left")
plt.tight_layout(); plt.show()
mse_dae_noisy_in = float(np.mean((dae.predict(noisy_demo, verbose=0) - X_valid[:8]) ** 2)); mse_noisy_in = float(np.mean((np.clip(noisy_demo, 0, 1) - X_valid[:8]) ** 2))
print(f"on these 8 images: noisy input vs clean {mse_noisy_in:.4f} · denoised output vs clean {mse_dae_noisy_in:.4f}")
```

**"Guessing" details.** Look at the bottom row for large, plain garments (a white shirt, a pullover): where the dropout removed the top part of the object the network *invents* a plausible continuation. It has not seen those pixels – it knows from training how such garments usually look. That is the point of a denoising autoencoder, and also its danger: the output is a plausible guess, not a measurement.

### 5.2 Gaussian noise vs. dropout noise at several levels (Task 3)
Gaussian noise with σ = 0.1 / 0.3 / 0.5 and dropout with rate 0.1 / 0.3 / 0.5. Each model is trained with its own noise and tested on validation images corrupted **the same way**; the error is measured against the **clean** images. A plain autoencoder (no noise) is trained too. "Do nothing" – using the noisy image itself as the answer – is the reference every denoiser must beat.

```python
def corrupt(X, kind, level, rng):
    if kind == "gaussian":
        return (X + rng.normal(0, level, X.shape)).astype("float32")
    if kind == "dropout":
        return (X * (rng.random(X.shape) >= level) / (1 - level)).astype("float32")      # same 1/(1-p) rescaling the Dropout layer uses
    if kind == "saltpepper":
        out = X.copy(); m = rng.random(X.shape); out[m < level / 2] = 0.0; out[(m >= level / 2) & (m < level)] = 1.0; return out
    return X.copy()


dn_rows, dn_models = [], {}
for kind, levels in (("gaussian", (0.1, 0.3, 0.5)), ("dropout", (0.1, 0.3, 0.5))):
    for lv in levels:
        for s in SZ["seeds"]:
            *_, ae, h, sec = train_ae(build_denoiser, optimizer=keras.optimizers.Adam(learning_rate=1e-3), seed=s, epochs=SZ["dn_epochs"], batch_size=64, kind=kind, level=lv)
            Xn = corrupt(X_valid, kind, lv, np.random.default_rng(100 + s))
            dn_rows.append(dict(kind=kind, level=lv, seed=s, err_denoised=float(np.mean((ae.predict(Xn, batch_size=1000, verbose=0) - X_valid) ** 2)), err_noisy_input=float(np.mean((np.clip(Xn, 0, 1) - X_valid) ** 2)) if kind == "gaussian" else float(np.mean((Xn - X_valid) ** 2))))
            if s == SZ["seeds"][0]:
                dn_models[(kind, lv)] = ae
for s in SZ["seeds"]:
    *_, ae, h, sec = train_ae(build_denoiser, optimizer=keras.optimizers.Adam(learning_rate=1e-3), seed=s, epochs=SZ["dn_epochs"], batch_size=64, kind="none", level=0)
    dn_rows.append(dict(kind="none", level=0.0, seed=s, err_denoised=pixel_mse(ae, X_valid), err_noisy_input=0.0))
    if s == SZ["seeds"][0]:
        dn_models[("none", 0.0)] = ae
dn = pd.DataFrame(dn_rows)
dtab = dn.groupby(["kind", "level"]).agg(denoised=("err_denoised", lambda v: pm(v, 5)), mean_denoised=("err_denoised", "mean"), noisy_input=("err_noisy_input", "mean"))
dtab["denoised ÷ noisy input"] = dtab.mean_denoised / dtab.noisy_input.replace(0, np.nan)
display(dtab[["denoised", "noisy_input", "denoised ÷ noisy input"]].rename(columns={"denoised": "pixel-MSE vs CLEAN image (mean ± std)", "noisy_input": "pixel-MSE of the noisy input itself"}).round(5))
RESULTS["denoise"] = {f"{k}-{l}": list(g.err_denoised) for (k, l), g in dn.groupby(["kind", "level"])}
audit_trained("every denoiser beats 'do nothing' (its error vs. the clean image is below that of the noisy input)", lambda: all(dtab.loc[(k, l), "mean_denoised"] < dtab.loc[(k, l), "noisy_input"] for k, l in dtab.index if k != "none"), "ratios " + ", ".join(f"{k[:4]}{l}: {r:.2f}" for (k, l), r in dtab["denoised ÷ noisy input"].dropna().items()))
fig, ax = plt.subplots(figsize=(5.6, 3.1))
for kind, c in (("gaussian", "C0"), ("dropout", "C1")):
    sub = dtab.loc[kind]; ax.plot(sub.index, sub.mean_denoised, "o-", color=c, label=f"{kind}: denoised"); ax.plot(sub.index, sub.noisy_input, "x--", color=c, alpha=0.6, label=f"{kind}: noisy input")
ax.set_xlabel("noise level (σ or dropout rate)"); ax.set_ylabel("pixel-MSE vs clean"); ax.legend(fontsize=7); plt.show()
```

**Note on the failed proof above (kept on purpose).** The one denoiser that does not beat "do nothing" is the Gaussian σ = 0.1 one: its error against the clean image is 0.0134, while the noisy input is already at 0.0069 (ratio 1.93). At such weak noise the autoencoder's own reconstruction error (about 0.013 even with no noise at all, see the "none" row) is larger than the noise it is supposed to remove. For dropout 0.1 and for all stronger noise the denoiser wins clearly (ratios from 0.60 down to 0.08). So the correct statement is: *a denoising autoencoder helps when the corruption is larger than the autoencoder's own reconstruction error*; the claim "every denoiser beats do nothing" was too strong.

### 5.3 Noise the model never saw: salt-and-pepper (Task 4)
The denoisers were trained with Gaussian or dropout noise. Now 20 % of the pixels are set to pure black or pure white ("salt and pepper"), a corruption of a different kind. A denoising autoencoder learns to undo *the noise it has seen*; for another kind of damage nothing guarantees it works.

```python
sp_in = corrupt(X_valid, "saltpepper", 0.2, np.random.default_rng(9))
rows_sp = []
for key, ae in dn_models.items():
    rows_sp.append(dict(model=f"{key[0]} {key[1]}" if key[0] != "none" else "plain (no noise)", err=float(np.mean((ae.predict(sp_in, batch_size=1000, verbose=0) - X_valid) ** 2))))
sp_tab = pd.DataFrame(rows_sp).set_index("model"); sp_in_err = float(np.mean((sp_in - X_valid) ** 2))
sp_tab["better than the noisy input?"] = sp_tab.err < sp_in_err
display(sp_tab.round(5)); print(f"the salt-and-pepper input itself has pixel-MSE {sp_in_err:.5f} vs. the clean image")
best_sp = sp_tab.err.idxmin(); print("best on unseen noise:", best_sp)
show_images(np.concatenate([X_valid[:8], sp_in[:8], dn_models[("dropout", 0.5)].predict(sp_in[:8], verbose=0), dn_models[("gaussian", 0.5)].predict(sp_in[:8], verbose=0)]), ncols=8, size=1.0, suptitle="clean · salt & pepper input · dropout-0.5 denoiser · Gaussian-0.5 denoiser")
RESULTS["saltpepper"] = dict(table=sp_tab.err.to_dict(), input_err=sp_in_err)
n_better = int(sp_tab["better than the noisy input?"].sum())
plain_err = float(sp_tab.loc["plain (no noise)", "err"])
vs_plain = (sp_tab.err - plain_err).drop("plain (no noise)")
print(f"{n_better} of {len(sp_tab)} models reduce the error of the salt-and-pepper input (a narrow bottleneck smooths almost any damage).")
print("The real question: does denoising training help on UNSEEN noise compared with the plain autoencoder (error − plain error; negative = better than plain)?")
print(vs_plain.round(5).to_string())
RESULTS["saltpepper_vs_plain"] = vs_plain.to_dict()
```

**How to read this.** A narrow bottleneck smooths out almost any damage, so "better than the noisy input" is an easy bar; the fair comparison is with the **plain autoencoder, which never saw noise** (the printed comparison). A denoiser that is better than plain on salt-and-pepper noise has learned something general about clean images; one that is not has only learned to undo *its own* training noise. A likely reason for limited transfer (my explanation, not measured here): pure black and white pixels are high-contrast, wrong values, while the training noise either removed pixels (dropout) or made small random changes (Gaussian) – a network that learned "a missing pixel is a zero, fill it in" does not know that a white pixel on a black background is damage. Practical rule: train denoisers with the noise you expect.

### 5.4 Sparse autoencoders (Tasks 5–6)
**Idea.** Another constraint: **penalise the number of active neurons in the coding layer** so that every input is a combination of a *few* active units. Each unit then tends to represent something useful – "if you could speak only a few words per month, you would make them worth listening to". The coding layer may even be **larger** than the input's useful dimension (here 300 sigmoid units); the penalty makes that workable.

**ℓ1 version.** `Flatten → Dense(100, selu) → Dense(300, sigmoid) → ActivityRegularization(l1=1e-3)`, decoder `Dense(100, selu) → Dense(784, sigmoid) → Reshape`. The penalty is `1e-3 · Σ|activation|`, added to the loss. **Why ℓ1 and not ℓ2?** ℓ1 pushes unneeded activations all the way to **0** while keeping the important ones large; ℓ2 shrinks everything a little, giving many small non-zero activations and no real sparsity. An equivalent way to write it is `activity_regularizer=keras.regularizers.l1(1e-3)` on the coding layer itself.

```python
def build_sparse(kind="none", weight=0.0, target=0.1, name="sparse"):
    coding_layers = {"none": [L.Dense(300, activation="sigmoid")],
                     "l1": [L.Dense(300, activation="sigmoid"), L.ActivityRegularization(l1=1e-3)],
                     "kl": [L.Dense(300, activation="sigmoid", activity_regularizer=KLDivergenceRegularizer(weight, target))]}[kind]
    enc = keras.Sequential([keras.Input(shape=(28, 28)), L.Flatten(), L.Dense(100, activation="selu")] + coding_layers, name=name + "_encoder")
    dec = keras.Sequential([keras.Input(shape=(300,)), L.Dense(100, activation="selu"), L.Dense(784, activation="sigmoid"), L.Reshape([28, 28])], name=name + "_decoder")
    return enc, dec, keras.Sequential([enc, dec], name=name)


# --- "both ways of writing the l1 penalty give the same number"
dA = keras.Sequential([keras.Input(shape=(100,)), L.Dense(300, activation="sigmoid"), L.ActivityRegularization(l1=1e-3)])
dB = keras.Sequential([keras.Input(shape=(100,)), L.Dense(300, activation="sigmoid", activity_regularizer=keras.regularizers.l1(1e-3))])
dB.set_weights(dA.get_weights()); xb = np.random.default_rng(0).standard_normal((64, 100)).astype("float32")
_ = dA(xb, training=True); _ = dB(xb, training=True)
la, lb = float(sum(dA.losses)), float(sum(dB.losses))
manual = 1e-3 * float(np.abs(dA.layers[0](xb).numpy()).sum()) / 64
audit("ActivityRegularization(l1=1e-3) and activity_regularizer=l1(1e-3) add the same penalty", abs(la - lb) < 1e-6, f"{la:.5f} vs {lb:.5f}")
audit("... and it equals 1e-3 · Σ|activations| averaged over the batch (Keras divides the summed penalty by the batch size)", abs(la - manual) < 1e-6, f"{la:.5f} vs hand {manual:.5f}")
log_change("activity regularizers (l1, custom KL) in tf.keras 2.0", "same penalty, but the summed penalty is divided by the batch size (also what TF 2 did)", "so the weight 0.05 of the KL regulariser from the brief is an effective 0.05 / batch_size; kept as in the brief and stated here")
```

### 5.5 KL-divergence sparsity (Tasks 7–8, Exercise 14)
**Why not just penalise the activations?** Measure the **mean activation of each coding neuron over the batch** (`q`, the actual sparsity – so the batch must be large enough for the mean to be accurate) and punish its distance from a **target sparsity** `p` (e.g. 0.1). Squared error `(q − p)²` works but gives weak gradients; the **Kullback–Leibler divergence** between two Bernoulli distributions is stronger:

`D_KL(p ‖ q) = p·log(p/q) + (1−p)·log((1−p)/(1−q))`

**Worked example.** A neuron has mean activation `q = 0.3`, target `p = 0.1`.

```python
def kl_bernoulli(p, q):
    return p * tf.math.log(p / q) + (1 - p) * tf.math.log((1 - p) / (1 - q))


p_, q_ = 0.1, 0.3
se = (q_ - p_) ** 2; klv = float(kl_bernoulli(p_, q_))
g_se = 2 * (q_ - p_); g_kl = -p_ / q_ + (1 - p_) / (1 - q_)
qq = tf.Variable(q_, dtype=tf.float64)
with tf.GradientTape() as tape:
    loss_kl = kl_bernoulli(tf.constant(p_, tf.float64), qq)
g_auto = float(tape.gradient(loss_kl, qq))
print(f"(a) squared error (q − p)²       = {se:.4f}")
print(f"(b) KL divergence                = {klv:.4f}   (natural log)")
print(f"(c) gradient of squared error    = 2(q − p) = {g_se:.4f}")
print(f"    gradient of KL w.r.t. q      = −p/q + (1−p)/(1−q) = {g_kl:.4f}   (automatic differentiation says {g_auto:.4f})")
print(f"    the KL gradient is {g_kl / g_se:.2f}× stronger")
audit("worked example: squared error 0.04", abs(se - 0.04) < 1e-9, se)
audit("worked example: KL ≈ 0.116", abs(klv - 0.1163) < 5e-4, f"{klv:.4f}")
audit("worked example: gradients 0.4 (squared error) and ≈ 0.952 (KL); autodiff agrees", abs(g_se - 0.4) < 1e-9 and abs(g_kl - 0.9524) < 5e-4 and abs(g_auto - g_kl) < 1e-6, f"{g_se:.3f}, {g_kl:.4f}, autodiff {g_auto:.4f}")
qs = np.linspace(0.005, 0.995, 400)
fig, ax = plt.subplots(1, 2, figsize=(9.2, 3.1))
ax[0].plot(qs, (qs - p_) ** 2, label="squared error (q − p)²"); ax[0].plot(qs, [float(kl_bernoulli(p_, q)) for q in qs], label="KL divergence"); ax[0].axvline(p_, ls=":", color="gray"); ax[0].set_xlabel("actual sparsity q (target p = 0.1)"); ax[0].set_ylabel("loss"); ax[0].legend(fontsize=8); ax[0].set_title("two sparsity losses")
ax[1].plot(qs, 2 * (qs - p_), label="d/dq squared error"); ax[1].plot(qs, -p_ / qs + (1 - p_) / (1 - qs), label="d/dq KL"); ax[1].axvline(p_, ls=":", color="gray"); ax[1].axhline(0, color="k", lw=0.5); ax[1].set_ylim(-1.5, 3); ax[1].set_xlabel("q"); ax[1].set_title("their gradients"); ax[1].legend(fontsize=8)
plt.tight_layout(); plt.show()
```

**Exercise 14 – proof.** With `p` fixed, the KL gradient simplifies:

`d/dq D_KL = −p/q + (1−p)/(1−q) = (−p(1−q) + (1−p)q) / (q(1−q)) = (q − p) / (q(1−q))`.

* **Zero at `q = p`:** the numerator is `p − p = 0` (and the denominator `p(1−p) > 0`).
* **Stronger than squared error:** the squared-error gradient is `2(q − p)`, so the ratio is `KL gradient / SE gradient = 1 / (2q(1−q))`. Since `q(1−q) ≤ 1/4` for every `q`, the ratio is always **≥ 2** (equality only at `q = ½`) and it grows without limit as `q → 0` or `q → 1`. For `q` far from `p` the KL gradient is therefore *more than twice* the squared-error gradient, and the further `q` is from ½ the larger the factor. At `q = 0.3` the ratio is `1/(2·0.21) = 2.38`, exactly what the numbers above show (0.952 / 0.4).

```python
grid = np.linspace(0.01, 0.99, 981)
grid = grid[np.abs(grid - 0.1) > 1e-9]
ratio = (-0.1 / grid + 0.9 / (1 - grid)) / (2 * (grid - 0.1))
closed = (grid - 0.1) / (grid * (1 - grid))
audit("Exercise 14: KL gradient = (q − p)/(q(1−q)) on a grid of 980 values of q", np.allclose(-0.1 / grid + 0.9 / (1 - grid), closed))
audit("Exercise 14: the KL gradient is exactly 0 at q = p (for several p)", all(abs(-p / p + (1 - p) / (1 - p)) < 1e-12 for p in (0.01, 0.1, 0.3, 0.5, 0.9)))
audit("Exercise 14: KL gradient ÷ squared-error gradient = 1/(2q(1−q)) ≥ 2 for every q", np.allclose(ratio, 1 / (2 * grid * (1 - grid))) and ratio.min() >= 2 - 1e-9, f"smallest ratio on the grid = {ratio.min():.4f} (at q = {grid[ratio.argmin()]:.2f})")
EX[14] = ("done", "section 5.5 (proof + numerical check)")
```

### 5.6 The `KLDivergenceRegularizer` and the sparse autoencoders (Tasks 9–10)
The custom regulariser takes the **mean over the batch axis** of the coding layer's activations and returns `weight · (KL(target, mean) + KL(1 − target, 1 − mean))`, summed over the neurons – i.e. the Bernoulli KL of every neuron. It is used as the `activity_regularizer` of the 300-unit sigmoid coding layer with `weight = 0.05`.

```python
class KLDivergenceRegularizer(keras.regularizers.Regularizer):
    def __init__(self, weight, target=0.1):
        self.weight, self.target = weight, target

    def __call__(self, inputs):
        mean_activities = tf.clip_by_value(tf.reduce_mean(inputs, axis=0), 1e-6, 1 - 1e-6)           # one mean per coding neuron
        return self.weight * tf.reduce_sum(kl_bernoulli(self.target, mean_activities))

    def get_config(self):
        return dict(weight=self.weight, target=self.target)


sp_rows, sp_models = [], {}
for kind, w in (("none", 0.0), ("l1", 0.0), ("kl", 0.05)):
    for s in SZ["seeds"]:
        enc, dec, ae, h, sec = train_ae(build_sparse, optimizer=keras.optimizers.Adam(learning_rate=1e-3), seed=s, epochs=SZ["sp_epochs"], batch_size=64, kind=kind, weight=w)
        A = enc.predict(X_valid, batch_size=1000, verbose=0)
        sp_rows.append(dict(model={"none": "no sparsity penalty", "l1": "ℓ1 (1e-3)", "kl": "KL (weight 0.05, target 0.1)"}[kind], seed=s, pix_mse=pixel_mse(ae, X_valid), below_01=float((A < 0.1).mean()), neurons_01_02=float(((A.mean(0) > 0.1) & (A.mean(0) < 0.2)).mean()), mean_act=float(A.mean()), dead=float((A.max(0) < 0.05).mean())))
        if s == SZ["seeds"][0]:
            sp_models[kind] = (enc, ae, A)
spd = pd.DataFrame(sp_rows)
spt = spd.groupby("model", sort=False).agg(**{"validation pixel-MSE": ("pix_mse", lambda v: pm(v, 5)), "activations below 0.1": ("below_01", lambda v: pm(v, 3)), "neurons with mean activation in 0.1–0.2": ("neurons_01_02", lambda v: pm(v, 3)), "mean activation": ("mean_act", "mean"), "neurons that never exceed 0.05": ("dead", "mean")})
display(spt.round(3))
RESULTS["sparse"] = {m: dict(mse=list(g.pix_mse), below=list(g.below_01), band=list(g.neurons_01_02)) for m, g in spd.groupby("model", sort=False)}
g_none, g_l1, g_kl = (spd[spd.model.str.startswith(k)] for k in ("no", "ℓ1", "KL"))
audit_trained("the KL penalty makes the code sparser than no penalty (more activations below 0.1)", lambda: g_kl.below_01.mean() > g_none.below_01.mean(), f"{g_kl.below_01.mean():.2f} vs {g_none.below_01.mean():.2f}")
audit_trained("the ℓ1 penalty makes the code sparser than no penalty (more activations below 0.1)", lambda: g_l1.below_01.mean() > g_none.below_01.mean(), f"{g_l1.below_01.mean():.2f} vs {g_none.below_01.mean():.2f}")
audit_trained("with the KL penalty the average coding activation is close to the target 0.1 (between 0.05 and 0.2)", lambda: 0.05 < g_kl.mean_act.mean() < 0.2, f"{g_kl.mean_act.mean():.3f}")
print("For reference, the source reports about 70 % of activations below 0.1 and about 90 % of neurons with a mean activation between 0.1 and 0.2.")
```

**Too high or too low a sparsity weight.** If the weight is **too high**, the network matches the target sparsity but spends its capacity on it and the reconstruction suffers; if it is **too low**, the penalty is ignored and nothing interesting is learned. One seed, KL penalty with five weights (note that Keras divides the summed penalty by the batch size, so these are "per-batch-size" weights):

### 6.1 What is different about a VAE (Tasks 1–3)
Kingma and Welling (2013) introduced the VAE. Two properties set it apart from every autoencoder so far:
* It is **probabilistic**: even after training, its output is partly determined by chance. (A denoising autoencoder is random only *during* training.)
* It is **generative**: it can create *new* instances that look like they were sampled from the training set.

Compared with restricted Boltzmann machines (the older generative models) VAEs are easier to train and sampling is much faster. They perform **variational Bayesian inference**, an efficient way of doing approximate Bayesian inference.

**Architecture.** The encoder does not output a coding but a **mean μ** and a **standard deviation σ** (in practice `γ = log σ²`). The coding that the decoder receives is **sampled** from the Gaussian `N(μ, σ²)`. So the sampled coding is *not* at μ – that randomness is what makes the model probabilistic, and it makes the decoder robust to small changes in the coding.

**Cost function = two parts.**
1. The usual **reconstruction loss** (pushes the output towards the input).
2. The **latent loss**: the KL divergence between the target distribution (a unit Gaussian `N(0, I)`) and the actual distribution of the codings. It pushes the codings to form a cloud that looks like a Gaussian in the **latent space**.

**Consequence.** After training, a new instance is generated by sampling a random coding from `N(0, I)` and decoding it – every point of that cloud decodes to something plausible, because training pulled all codings into it.

### 6.2 The latent loss, both ways (Task 4)
`ℒ = −½ Σᵢ (1 + log σᵢ² − σᵢ² − μᵢ²)` and with `γ = log σ²`: `ℒ = −½ Σᵢ (1 + γᵢ − exp(γᵢ) − μᵢ²)`.
Predicting `γ` instead of `σ` is **more numerically stable** (γ can be any real number, so there is no need for a positivity constraint or a log of something near 0) and training is faster. The standard deviation is recovered with `σ = exp(γ/2)`.

```python
def latent_loss_fn(mean, log_var):
    return -0.5 * tf.reduce_sum(1 + log_var - tf.exp(log_var) - tf.square(mean), axis=-1)


# (1) exp(γ/2) = σ
sig = np.array([0.1, 0.5, 1.0, 2.0, 7.0]); gam = np.log(sig ** 2)
audit("exp(γ/2) = σ when γ = log σ²", np.allclose(np.exp(gam / 2), sig))
# (2) the loss is 0 for a perfect unit Gaussian and positive otherwise
zero = float(latent_loss_fn(tf.zeros([1, 10]), tf.zeros([1, 10]))[0])
audit("the latent loss is exactly 0 for μ = 0 and γ = 0 (a perfect unit Gaussian)", zero == 0.0, zero)
rng = np.random.default_rng(0)
m_rand, g_rand = rng.normal(0, 1, (500, 10)).astype("float32"), rng.normal(0, 1, (500, 10)).astype("float32")
audit("the latent loss is never negative (it is a KL divergence), and positive for random μ, γ", float(tf.reduce_min(latent_loss_fn(m_rand, g_rand))) >= 0 and float(tf.reduce_mean(latent_loss_fn(m_rand, g_rand))) > 0)
# (3) it equals the KL divergence N(μ,σ²) ‖ N(0,1) computed by numerical integration, per dimension
def kl_numeric(mu, sigma):
    x = np.linspace(-14, 14, 400001); dx = x[1] - x[0]
    p = np.exp(-0.5 * ((x - mu) / sigma) ** 2) / (sigma * np.sqrt(2 * np.pi)); q = np.exp(-0.5 * x ** 2) / np.sqrt(2 * np.pi)
    return float(np.sum(p * (np.log(p + 1e-300) - np.log(q)) * dx))


tests = [(0.0, 1.0), (1.0, 1.0), (0.0, 0.5), (-1.5, 2.0), (0.7, 0.3)]
formula = [float(latent_loss_fn(tf.constant([[mu]], tf.float32), tf.constant([[np.log(sg ** 2)]], tf.float32))[0]) for mu, sg in tests]
numeric = [kl_numeric(mu, sg) for mu, sg in tests]
display(pd.DataFrame({"μ": [t[0] for t in tests], "σ": [t[1] for t in tests], "formula −½(1+γ−e^γ−μ²)": np.round(formula, 5), "numerical integral of KL": np.round(numeric, 5)}))
audit("the latent-loss formula equals the KL divergence to N(0,1) found by numerical integration (5 cases)", np.allclose(formula, numeric, atol=2e-4), f"max diff {np.abs(np.array(formula) - np.array(numeric)).max():.1e}")
```

### 6.3 The `Sampling` layer and the reparameterization trick (Task 5)
`Sampling` returns `ε · exp(γ/2) + μ` with `ε ~ N(0, I)`. This is the **reparameterization trick**: instead of "sample from `N(μ, σ²)`" (a random operation that gradients cannot pass through) we write the sample as a *deterministic function of μ and γ plus an independent noise ε*. The randomness lives only in ε, so backpropagation can compute `∂coding/∂μ = 1` and `∂coding/∂γ = ½·ε·exp(γ/2)`, and the encoder can learn. (The brief notes that the source text does not use this name – it is what the code does.)

```python
class Sampling(L.Layer):
    def call(self, inputs):
        mean, log_var = inputs
        return tf.random.normal(tf.shape(log_var)) * tf.exp(log_var / 2) + mean


mean_t, lv_t = tf.constant(np.full((50000, 2), [1.5, -2.0], dtype="float32")), tf.constant(np.full((50000, 2), [0.0, np.log(0.25)], dtype="float32"))
samples = Sampling()([mean_t, lv_t]).numpy()
audit("sampled codings have the requested mean and standard deviation (μ = 1.5 / −2.0, σ = 1 / 0.5)", np.allclose(samples.mean(0), [1.5, -2.0], atol=0.02) and np.allclose(samples.std(0), [1.0, 0.5], atol=0.02), f"mean {samples.mean(0).round(3)}, std {samples.std(0).round(3)}")
m1 = tf.Variable([[0.3, -0.2]], dtype=tf.float32); g1 = tf.Variable([[0.2, -0.4]], dtype=tf.float32)
with tf.GradientTape() as tape:
    z = Sampling()([m1, g1])
    s_ = tf.reduce_sum(z)
dm, dg = tape.gradient(s_, [m1, g1])
eps = (z.numpy() - m1.numpy()) / np.exp(g1.numpy() / 2)
audit("gradients flow through the sampling: ∂z/∂μ = 1 and ∂z/∂γ = ½ ε exp(γ/2) (checked with GradientTape)", np.allclose(dm.numpy(), 1.0) and np.allclose(dg.numpy(), 0.5 * eps * np.exp(g1.numpy() / 2), atol=1e-5), f"∂z/∂μ = {dm.numpy().round(3)}, ∂z/∂γ = {dg.numpy().round(3)}")
```

### 6.4 Build the VAE (Tasks 6–7)
Encoder `Flatten → Dense(150, selu) → Dense(100, selu)` with two `Dense(10)` heads (mean, log-variance) and `Sampling`; decoder `Dense(100, selu) → Dense(150, selu) → Dense(784, sigmoid) → Reshape`. The reconstruction loss is binary cross-entropy. The **latent loss** is added to the model's loss with `add_loss` (done by a small layer, see the version note) as `mean(latent_loss) / 784`.

**Why `/ 784`?** Keras' `binary_crossentropy` **averages** over the 784 pixels, but the true reconstruction loss is the **sum** over pixels. Dividing the latent loss by 784 puts both terms on the same scale as "per pixel". The final loss is then 784× smaller than the "true" one, which only means that a larger learning rate would be appropriate.

```python
class LatentLoss(L.Layer):
    """adds mean(latent_loss) · scale to the model loss; passes its inputs through unchanged"""
    def __init__(self, scale=1.0 / 784.0, **kwargs):
        super().__init__(**kwargs); self.scale = scale

    def call(self, inputs):
        mean, log_var = inputs
        self.add_loss(self.scale * tf.reduce_mean(latent_loss_fn(mean, log_var)))
        return inputs


def build_vae(codings_size=10, latent_scale=1.0 / 784.0, name="vae"):
    inputs = keras.Input(shape=(28, 28))
    z = L.Flatten()(inputs); z = L.Dense(150, activation="selu")(z); z = L.Dense(100, activation="selu")(z)
    codings_mean, codings_log_var = L.Dense(codings_size, name="mean")(z), L.Dense(codings_size, name="log_var")(z)
    codings_mean, codings_log_var = LatentLoss(latent_scale, name="latent_loss")([codings_mean, codings_log_var])
    codings = Sampling()([codings_mean, codings_log_var])
    variational_encoder = keras.Model(inputs=[inputs], outputs=[codings_mean, codings_log_var, codings], name=name + "_encoder")
    dec_in = keras.Input(shape=(codings_size,))
    x = L.Dense(100, activation="selu")(dec_in); x = L.Dense(150, activation="selu")(x); x = L.Dense(28 * 28, activation="sigmoid")(x)
    decoder = keras.Model(inputs=dec_in, outputs=L.Reshape([28, 28])(x), name=name + "_decoder")
    _, _, codings_ = variational_encoder(inputs)
    return variational_encoder, decoder, keras.Model(inputs=inputs, outputs=decoder(codings_), name=name)


log_change("variational_ae.add_loss(tensor) on a Functional model", "a small LatentLoss layer that calls self.add_loss(...)", "Model.add_loss with a symbolic tensor is not supported the same way in Keras 3; a layer's add_loss is")
venc, vdec, vae = build_vae()
vae.compile(loss="binary_crossentropy", optimizer="rmsprop")
_ = vae(X_valid[:4])
audit("the latent loss is registered with the model through add_loss (one extra loss term after a forward pass)", len(vae.losses) == 1, f"{len(vae.losses)} extra loss term(s)")


class VAEMonitor(keras.callbacks.Callback):
    """after every epoch: the two loss components on the validation set (reconstruction = Keras BCE, latent = mean latent loss / 784)"""
    def __init__(self, X, scale=1.0 / 784.0):
        super().__init__(); self.X, self.scale, self.rec, self.lat = X, scale, [], []

    def on_epoch_end(self, epoch, logs=None):
        mean, lv, _ = venc_ref.predict(self.X, batch_size=1000, verbose=0)
        out = self.model.predict(self.X, batch_size=1000, verbose=0)
        self.rec.append(float(np.mean(keras.losses.binary_crossentropy(self.X, out)))); self.lat.append(float(np.mean(latent_loss_fn(mean, lv)) * self.scale))


def train_vae(seed, epochs, codings_size=10, latent_scale=1.0 / 784.0, X=None, loss="binary_crossentropy"):
    global venc_ref
    set_seed(seed); fresh()
    ve, de, v = build_vae(codings_size, latent_scale)
    venc_ref = ve
    v.compile(loss=loss, optimizer="rmsprop")
    mon = VAEMonitor(X_valid, latent_scale)
    h, sec = fit_quiet(v, X_train if X is None else X, X_train if X is None else X, epochs=epochs, batch_size=128, validation_data=(X_valid, X_valid), callbacks=[mon])
    return ve, de, v, h, mon, sec
```

### 6.5 Train (Task 8) and look at the two loss components separately
50 epochs, batch size 128, RMSprop, Fashion MNIST (fewer in QUICK mode). `val_loss` from Keras is the **sum** of both parts; the monitor splits it.

```python
venc, vdec, vae, hv, mon, sec_v = train_vae(11, SZ["vae_epochs"])
print(f"{SZ['vae_epochs']} epochs, {sec_v:.1f} s / epoch · parameters: encoder {venc.count_params():,}, decoder {vdec.count_params():,}")
fig, ax = plt.subplots(1, 3, figsize=(11, 2.9))
ax[0].plot(hv["loss"], label="training"); ax[0].plot(hv["val_loss"], label="validation"); ax[0].set_title("total loss (what Keras reports)", fontsize=9); ax[0].legend(fontsize=7); ax[0].set_xlabel("epoch")
ax[1].plot(mon.rec, color="C2"); ax[1].set_title("reconstruction part (validation)", fontsize=9); ax[1].set_xlabel("epoch")
ax[2].plot(mon.lat, color="C3"); ax[2].set_title("latent part ÷ 784 (validation)", fontsize=9); ax[2].set_xlabel("epoch")
plt.tight_layout(); plt.show()
audit("the monitor's two parts add up to Keras' own validation loss at the last epoch", abs(mon.rec[-1] + mon.lat[-1] - hv["val_loss"][-1]) < 2e-3, f"{mon.rec[-1]:.4f} + {mon.lat[-1]:.4f} = {mon.rec[-1] + mon.lat[-1]:.4f} vs {hv['val_loss'][-1]:.4f}")
audit_trained("the reconstruction part of the validation loss fell during training (last epoch below the first)", lambda: mon.rec[-1] < mon.rec[0], f"{mon.rec[0]:.4f} → {mon.rec[-1]:.4f}")
RESULTS["vae"] = dict(rec=mon.rec[-1], lat=mon.lat[-1], total=hv["val_loss"][-1])
```

**The alternative: a reconstruction loss that sums over pixels.** Instead of dividing the latent loss by 784, make the reconstruction loss the true *sum* over the pixels and give the latent loss weight 1. The two ways are the same loss up to a factor of exactly 784, as the check below shows on one batch with identical weights; with RMSprop (which normalises the size of the gradients) training behaves almost identically.

```python
def recon_sum_loss(y_true, y_pred):
    p = tf.clip_by_value(y_pred, 1e-7, 1 - 1e-7)
    return tf.reduce_sum(-(y_true * tf.math.log(p) + (1 - y_true) * tf.math.log(1 - p)), axis=[1, 2])


xb = X_valid[:256]
mean_b, lv_b, _ = venc.predict(xb, verbose=0); out_b = vae.predict(xb, verbose=0)
total_mean_version = float(np.mean(keras.losses.binary_crossentropy(xb, out_b))) + float(np.mean(latent_loss_fn(mean_b, lv_b))) / 784.0
total_sum_version = float(np.mean(recon_sum_loss(tf.constant(xb), tf.constant(out_b)))) + float(np.mean(latent_loss_fn(mean_b, lv_b)))
audit("'mean BCE + latent/784' and 'sum BCE + latent' are the same loss up to the factor 784 (identical weights, one batch)", abs(total_sum_version / total_mean_version - 784) < 0.5, f"{total_sum_version:.3f} / {total_mean_version:.5f} = {total_sum_version / total_mean_version:.2f}")
ve2, de2, v2, h2, mon2, _ = train_vae(11, SZ["vae_w_epochs"], latent_scale=1.0, loss=recon_sum_loss)
_, _, vref, href, monref, _ = train_vae(11, SZ["vae_w_epochs"])
print(f"after {SZ['vae_w_epochs']} epochs, validation reconstruction (BCE per pixel): mean-version {monref.rec[-1]:.4f} · sum-version {mon2.rec[-1]:.4f}")
print(f"validation latent KL (per image): mean-version {monref.lat[-1] * 784:.2f} · sum-version {mon2.lat[-1]:.2f}")
RESULTS["vae_sum_vs_mean"] = dict(mean_version=(monref.rec[-1], monref.lat[-1] * 784), sum_version=(mon2.rec[-1], mon2.lat[-1]))
```

### 6.6 Generate new images (Task 9)
Sample 12 codings from `N(0, I)` with `tf.random.normal([12, 10])` and decode them.

### 6.7 Semantic interpolation (Task 10)
**Why in coding space?** Mixing two images pixel by pixel gives an overlay of both (two ghosts on top of each other). Moving between two *codings* and decoding each step gives intermediate images that are plausible items – a shoe that gradually becomes a different shoe. The source's trick: reshape 12 codings to a 3×4 grid (`[1, 3, 4, 10]`), enlarge it bilinearly to 5×7 with `tf.image.resize` (every other row and column is now an interpolation), flatten, decode, and plot the 5×7 grid. For comparison the same is done in **pixel space**: decode the 12 corner codings first, then enlarge the resulting 3×4 grid of images the same way.

```python
set_seed(3)
codings_grid = tf.reshape(tf.random.normal([12, 10]), [1, 3, 4, 10])
larger = tf.image.resize(codings_grid, [5, 7])
semantic = vdec(tf.reshape(larger, [-1, 10])).numpy()
corner_imgs = vdec(tf.reshape(codings_grid, [-1, 10])).numpy().reshape(1, 3, 4, 784)
pixel_blend = tf.image.resize(corner_imgs, [5, 7]).numpy().reshape(-1, 28, 28)
audit("the resize really contains the original 12 codings at the grid corners and interpolates between them", np.allclose(larger.numpy()[0, 0, 0], codings_grid.numpy()[0, 0, 0], atol=1e-5) and np.allclose(larger.numpy()[0, 4, 6], codings_grid.numpy()[0, 2, 3], atol=1e-5))
fig, axes = plt.subplots(1, 2, figsize=(12, 4.0))
for ax, imgs_, title in ((axes[0], semantic, "interpolation in CODING space, then decoded"), (axes[1], pixel_blend, "interpolation in PIXEL space (blend of decoded corners)")):
    grid_ = np.concatenate([np.concatenate(list(imgs_[r * 7:(r + 1) * 7]), axis=1) for r in range(5)], axis=0)
    ax.imshow(grid_, cmap="binary"); ax.axis("off"); ax.set_title(title, fontsize=9)
plt.tight_layout(); plt.show()
# objective comparison on many random grids: how 'real' do the intermediate images look?
conf_s, conf_p, rec_s, rec_p = [], [], [], []
for k in range(pick(40, 6)):
    cg = tf.reshape(tf.random.normal([12, 10]), [1, 3, 4, 10]); lg = tf.image.resize(cg, [5, 7])
    sem = vdec(tf.reshape(lg, [-1, 10])).numpy()
    blend = tf.image.resize(vdec(tf.reshape(cg, [-1, 10])).numpy().reshape(1, 3, 4, 784), [5, 7]).numpy().reshape(-1, 28, 28)
    conf_s.append(fashion_clf.predict(sem, verbose=0).max(1).mean()); conf_p.append(fashion_clf.predict(blend, verbose=0).max(1).mean())
    rec_s.append(float(np.mean((vae.predict(sem, verbose=0) - sem) ** 2))); rec_p.append(float(np.mean((vae.predict(blend, verbose=0) - blend) ** 2)))
cmp_ = pd.DataFrame({"classifier confidence (higher = looks like a real item)": [np.mean(conf_s), np.mean(conf_p)], "VAE reconstruction error of the image (lower = more typical)": [np.mean(rec_s), np.mean(rec_p)]}, index=["coding-space interpolation", "pixel-space blending"])
display(cmp_.round(4))
RESULTS["interp"] = dict(conf_semantic=float(np.mean(conf_s)), conf_pixel=float(np.mean(conf_p)), rec_semantic=float(np.mean(rec_s)), rec_pixel=float(np.mean(rec_p)))
audit_trained("the classifier is more confident about coding-space interpolations than about pixel blends", lambda: np.mean(conf_s) > np.mean(conf_p), f"{np.mean(conf_s):.3f} vs {np.mean(conf_p):.3f}")
```

**Reading the grids.** In the left grid look for a row where one item morphs into another through other plausible items (for example between two kinds of shoes or between two kinds of tops); in the right grid the middle columns are double exposures of both end images. The numbers in the table put a value on the same impression.

# Phase 7 · Generative adversarial networks – and why they are hard to train

```python
SB.phase = "Phase 7"
```

### 7.1 GANs in one page (Tasks 1–4)
**Why GANs overtook VAEs.** A VAE's samples are "a bit too fuzzy"; a GAN's are crisp, because the loss that shapes them is "does this look real to a critic?" and not "is it close to the average of everything that could have produced this code". Applications named in the source: **super-resolution**, **colorization**, **image editing** (e.g. removing photo-bombers), turning a **sketch into a photorealistic image**, **next-frame prediction** in video, **data-set augmentation**, generating other kinds of data (**text, audio, time series**), and **finding weaknesses in other models** to make them stronger. LeCun called GANs "the most interesting idea in machine learning in the last ten years".

**Two networks (Goodfellow et al., 2014).** The **generator** takes a random Gaussian input – the *coding* – and outputs data, an image; it plays the same role as a VAE's decoder but is trained very differently. The **discriminator** takes an image, real or fake, and guesses which. Analogy: a **counterfeiter** (generator) makes banknotes, a **police investigator** (discriminator) tries to tell fakes from real ones; each gets better because the other does.

**One training iteration has two phases.**
1. **Discriminator phase.** A batch of real images plus an equal number of fakes from the current generator. Labels: **0 for fake, 1 for real**. One training step with binary cross-entropy; backpropagation updates **only the discriminator**.
2. **Generator phase.** A *new* batch of fakes only – no real images. All labels are set to **1** ("real"). The discriminator's weights are **frozen**; backpropagation updates **only the generator** – the generator is trained to make the (frozen) discriminator say "real".

**How can the generator learn to make convincing images without ever seeing a real one?** It only receives gradients that flow back *through the discriminator*: "change the pixels in this direction and the critic will find the picture more real". As the discriminator learns what real images look like, those gradients carry more and more information about real images – second-hand, but real.

### 7.2 The simple Fashion MNIST GAN (Task 5)
Coding size 30. Generator `Dense(100, selu) → Dense(150, selu) → Dense(784, sigmoid) → Reshape([28, 28])`; discriminator `Flatten → Dense(150, selu) → Dense(100, selu) → Dense(1, sigmoid)`; `gan = Sequential([generator, discriminator])`. The discriminator is compiled with BCE/RMSprop, then made non-trainable, then `gan` is compiled.

**The `trainable` note.** In the TF 2.0 code of the source, Keras read the `trainable` flag **only when a model is compiled**: `discriminator.trainable = False` before compiling `gan` froze it *inside* `gan`, while `discriminator.train_on_batch()` (compiled earlier) still trained it. Keras 3 reads the flag **when the step runs**, so after `discriminator.trainable = False` the call `discriminator.train_on_batch()` silently trains nothing. The loop below therefore switches the flag explicitly – and an audit checks, step by step, that each phase changes only the network it should.

```python
CODINGS = 30


class MinibatchStd(L.Layer):
    """appends one extra feature: the standard deviation across the batch, averaged over the chosen axes (a diversity statistic for the discriminator)"""
    def __init__(self, axes=(0,), **kwargs):
        super().__init__(**kwargs); self.axes = list(axes)

    def call(self, x):
        v = tf.reduce_mean(tf.math.reduce_std(x, axis=self.axes))
        return tf.concat([x, tf.fill(tf.concat([tf.shape(x)[:-1], [1]], 0), v)], axis=-1)

    def compute_output_shape(self, s):
        return tuple(s[:-1]) + (s[-1] + 1,)

    def get_config(self):
        return {**super().get_config(), "axes": self.axes}


def build_simple_gan(use_mbd=False, codings=CODINGS):
    g = keras.Sequential([keras.Input(shape=(codings,)), L.Dense(100, activation="selu"), L.Dense(150, activation="selu"), L.Dense(28 * 28, activation="sigmoid"), L.Reshape([28, 28])], name="generator")
    d = keras.Sequential([keras.Input(shape=(28, 28)), L.Flatten(), L.Dense(150, activation="selu"), L.Dense(100, activation="selu")] + ([MinibatchStd()] if use_mbd else []) + [L.Dense(1, activation="sigmoid")], name="discriminator")
    return g, d


g0, d0 = build_simple_gan()
audit("generator maps a 30-number coding to a 28×28 image in [0, 1]; discriminator maps an image to one probability", g0.output_shape == (None, 28, 28) and d0.output_shape == (None, 1) and float(g0(tf.random.normal([4, 30])).numpy().min()) >= 0)
print(f"generator {g0.count_params():,} parameters · discriminator {d0.count_params():,} parameters")


def weights_of(m):
    return [w.numpy().copy() for w in m.weights]


def changed(before, m):
    return any(not np.allclose(a, b.numpy()) for a, b in zip(before, m.weights))


def sample_grid(g, z, scale01=True):
    imgs = g(z, training=False).numpy()
    return imgs if scale01 else (np.squeeze(imgs) + 1) / 2
```

### 7.3 Version 1 – the book's loop with `train_on_batch`
`tf.data` pipeline with `shuffle(1000)`, `batch(32, drop_remainder=True)`, `prefetch(1)`. For each batch: noise → fakes; `X_fake_and_real` (fakes first); labels `[0…0, 1…1]`; `discriminator.train_on_batch`; then new noise, all labels 1, `gan.train_on_batch`. Because this loop is slow on a CPU (two Python-level Keras calls per step) it runs for only a few epochs here; the faster version of Section 7.4 does the long run.

```python
def train_gan_tob(g, d, X, batch_size=32, n_epochs=3, codings=CODINGS, snapshot_z=None):
    d.compile(loss="binary_crossentropy", optimizer="rmsprop")
    d.trainable = False
    gan = keras.Sequential([g, d]); gan.compile(loss="binary_crossentropy", optimizer="rmsprop")
    ds = tf.data.Dataset.from_tensor_slices(X).shuffle(1000).batch(batch_size, drop_remainder=True).prefetch(1)
    snaps, checks, losses = {}, [], []
    for epoch in range(1, n_epochs + 1):
        t0 = time.time()
        for step, X_batch in enumerate(ds):
            noise = tf.random.normal(shape=[batch_size, codings])
            generated_images = g(noise, training=True)
            X_fake_and_real = tf.concat([generated_images, tf.cast(X_batch, tf.float32)], axis=0)
            y1 = tf.constant([[0.]] * batch_size + [[1.]] * batch_size)
            wd0, wg0 = weights_of(d), weights_of(g)
            d.trainable = True
            ld = d.train_on_batch(X_fake_and_real, y1)                       # phase 1: only the discriminator learns
            d.trainable = False
            c1 = (changed(wd0, d), changed(wg0, g)) if step < 3 else None
            noise = tf.random.normal(shape=[batch_size, codings])
            y2 = tf.constant([[1.]] * batch_size)
            wd1, wg1 = weights_of(d), weights_of(g)
            lg = gan.train_on_batch(noise, y2)                                # phase 2: only the generator learns
            if step < 3:
                checks.append((epoch, step, c1, (changed(wd1, d), changed(wg1, g))))
            if step % 50 == 0:
                losses.append((epoch, step, float(ld), float(lg)))
        snaps[epoch] = sample_grid(g, snapshot_z)
        print(f"epoch {epoch}/{n_epochs}: {time.time() - t0:.0f} s · last discriminator loss {float(ld):.3f} · generator loss {float(lg):.3f}")
    return snaps, checks, losses


set_seed(11); fresh()
g1, d1 = build_simple_gan()
z_fixed = tf.random.normal([16, CODINGS], seed=1)
snaps1, checks1, losses1 = train_gan_tob(g1, d1, X_train, batch_size=32, n_epochs=SZ["gan_tob_epochs"], snapshot_z=z_fixed)
audit("phase 1 changes only the discriminator (checked on the first 3 steps of every epoch)", all(c[2] == (True, False) for c in checks1), checks1[0])
audit("phase 2 changes only the generator (checked on the first 3 steps of every epoch)", all(c[3] == (False, True) for c in checks1), checks1[0])
show_images(np.concatenate(list(snaps1.values())), ncols=8, size=1.0, suptitle=f"train_on_batch GAN: the same 16 codings after epoch {', '.join(str(e) for e in snaps1)} (rows)")
RESULTS["gan_tob"] = dict(losses=losses1)
```

### 7.4 Version 2 – the same GAN with `tf.GradientTape` and two optimizers (Task 7)
Instead of toggling `trainable`, each phase computes gradients **only for the variables it owns**: `tape.gradient(loss, discriminator.trainable_variables)` for the discriminator phase, `…generator.trainable_variables` for the generator phase, with two separate optimizers. Nothing needs freezing. The loop logs, every epoch, the discriminator loss, the generator loss and the discriminator's accuracy on real and on fake images, and (every epoch) a diversity check: the generated images are labelled by the helper classifier of Phase 6 (accuracy > 0.85 on real data), and we count how many classes appear and how evenly.

The same loop also supports **experience replay** (Section 7.7) and a **minibatch-diversity** layer in the discriminator.

```python
class ReplayBuffer:
    """ring buffer of generated images from past iterations; the oldest are dropped when it is full"""
    def __init__(self, capacity, shape=(28, 28), seed=0):
        self.buf = np.zeros((capacity,) + tuple(shape), "float32"); self.cap, self.n, self.ptr = capacity, 0, 0
        self.rng = np.random.default_rng(seed)

    def push(self, batch):
        for im in batch:
            self.buf[self.ptr] = im; self.ptr = (self.ptr + 1) % self.cap; self.n = min(self.n + 1, self.cap)

    def sample(self, k):
        return self.buf[self.rng.integers(0, self.n, k)]


def gan_diversity(g, n=2000, codings=CODINGS, seed=0, scale01=True, cond=None):
    z = tf.random.normal([n, codings], seed=seed)
    fakes = g(z, training=False).numpy() if cond is None else g([z, cond], training=False).numpy()
    fakes = np.squeeze(fakes) if scale01 else (np.squeeze(fakes) + 1) / 2
    p = fashion_clf.predict(fakes, batch_size=500, verbose=0)
    share = np.bincount(p.argmax(1), minlength=10) / n
    ent = float(-(share[share > 0] * np.log(share[share > 0])).sum())
    return dict(entropy=ent, classes=int((share > 0.02).sum()), conf=float(p.max(1).mean()), share=share)


def train_gan_tape(g, d, X, epochs, batch=64, replay=0, seed=0, lr_g=1e-3, lr_d=1e-3, codings=CODINGS, snap_every=None, z_fixed=None, scale01=True, verbose=True):
    opt_g, opt_d = keras.optimizers.RMSprop(learning_rate=lr_g), keras.optimizers.RMSprop(learning_rate=lr_d)
    bce = keras.losses.BinaryCrossentropy()

    @tf.function
    def make_fakes(z):
        return g(z, training=True)

    @tf.function
    def d_step(real, fake):
        with tf.GradientTape() as tape:
            pr, pf = d(real, training=True), d(fake, training=True)
            loss = 0.5 * (bce(tf.ones_like(pr), pr) + bce(tf.zeros_like(pf), pf))
        opt_d.apply_gradients(zip(tape.gradient(loss, d.trainable_variables), d.trainable_variables))
        return loss, tf.reduce_mean(tf.cast(pr > 0.5, tf.float32)), tf.reduce_mean(tf.cast(pf < 0.5, tf.float32))

    @tf.function
    def g_step(z):
        with tf.GradientTape() as tape:
            pf = d(g(z, training=True), training=True)
            loss = bce(tf.ones_like(pf), pf)
        opt_g.apply_gradients(zip(tape.gradient(loss, g.trainable_variables), g.trainable_variables))
        return loss

    buf = ReplayBuffer(replay * batch, seed=seed) if replay else None
    ds = tf.data.Dataset.from_tensor_slices(X).shuffle(1000, seed=seed).batch(batch, drop_remainder=True).prefetch(1)
    hist = {k: [] for k in ("d_loss", "g_loss", "acc_real", "acc_fake", "entropy", "classes", "conf", "sec")}; snaps = {}
    for epoch in range(1, epochs + 1):
        t0 = time.time(); acc = []
        for xb in ds:
            fake = make_fakes(tf.random.normal([batch, codings]))
            if buf is not None:
                buf.push(fake.numpy()); fake_for_d = tf.constant(buf.sample(batch))
            else:
                fake_for_d = fake
            dl, ar, af = d_step(xb, fake_for_d)
            gl = g_step(tf.random.normal([batch, codings]))
            acc.append((float(dl), float(gl), float(ar), float(af)))
        a = np.mean(acc, axis=0); dv = gan_diversity(g, codings=codings, seed=epoch, scale01=scale01)
        for k, v in zip(("d_loss", "g_loss", "acc_real", "acc_fake"), a):
            hist[k].append(float(v))
        hist["entropy"].append(dv["entropy"]); hist["classes"].append(dv["classes"]); hist["conf"].append(dv["conf"]); hist["sec"].append(time.time() - t0)
        if snap_every and (epoch % snap_every == 0 or epoch == 1 or epoch == epochs) and z_fixed is not None:
            snaps[epoch] = sample_grid(g, z_fixed, scale01)
        if verbose and (epoch % max(1, epochs // 6) == 0 or epoch == 1):
            print(f"epoch {epoch:3d}/{epochs}: D loss {a[0]:.3f} · G loss {a[1]:.3f} · D acc real {a[2]:.2f} / fake {a[3]:.2f} · classes among fakes {dv['classes']} (entropy {dv['entropy']:.2f}) · {hist['sec'][-1]:.1f} s")
    return hist, snaps


set_seed(11); fresh()
g2, d2 = build_simple_gan()
z_fixed16 = tf.random.normal([16, CODINGS], seed=2)
hist_main, snaps_main = train_gan_tape(g2, d2, X_train, SZ["gan_epochs"], batch=SZ["gan_batch"], seed=11, snap_every=max(1, SZ["gan_epochs"] // 6), z_fixed=z_fixed16)
RESULTS["gan_main"] = {k: v for k, v in hist_main.items()}
rows_ = np.concatenate([snaps_main[e] for e in snaps_main])
show_images(rows_, ncols=8, size=1.0, suptitle="GradientTape GAN, the same 16 codings after epoch " + ", ".join(str(e) for e in snaps_main) + " (two rows per snapshot)")
audit("the discriminator and generator stay separate: trainable flags were never toggled (both True) and each optimizer only owned its own variables", d2.trainable and g2.trainable)
```

### 7.5 Why GANs are hard to train (Task 8)
In my own words:
* **Nash equilibrium.** A situation in which no player can do better by changing *its own* strategy while the other keeps its strategy (the classic example: everyone driving on the left, or everyone on the right – if everybody else does it, you are best off doing it too). A GAN has exactly **one** such equilibrium: a generator whose images are perfectly realistic, and a discriminator that can only guess, 50 / 50. Nothing guarantees that training ever gets there: each network improves against the *current* opponent, which keeps changing.
* **Mode collapse.** The generator's outputs become less and less diverse. The story from the source: the generator gets better at shoes than at anything else, so it makes only shoes; the discriminator learns to reject shoes, so the generator moves to shirts; the discriminator then rejects shirts, and the generator cycles through a few classes without ever being good at all of them.
* **Oscillation / instability.** Both players keep adjusting to each other, so the parameters can oscillate, and training can suddenly diverge for no obvious reason.
* **Sensitivity to hyperparameters.** Especially to the *relative* learning rates and strengths of the two networks; a discriminator that is too strong gives the generator no usable gradient, one that is too weak gives it no useful signal either.
* **Cost functions.** Many papers propose new losses; a 2018 Google study questions how much they matter. The field is still active and the dynamics are not fully understood.

### 7.7 Experience replay and minibatch diversity (Tasks 10–11)
**Experience replay.** Keep a **buffer** of generated images from past iterations (when it is full, the oldest are dropped). Train the discriminator on real images plus fake images **drawn from the buffer**, not only on the current generator's output. The discriminator then keeps seeing old kinds of fakes, so the generator cannot escape by cycling back to something the discriminator has just forgotten. (Here: a buffer of 50 batches.)

**Minibatch discrimination (simplified).** Give the discriminator a statistic computed across the whole batch – the standard deviation of its features across the batch, averaged to one number – as an extra input feature. A generator that makes near-identical images produces a batch with a tiny statistic, which the discriminator can learn to reject; this pushes the generator towards diversity. (The progressive-GAN version of this layer is tested in Phase 8.)

Each variant is trained for the same number of epochs with 2 seeds; all samples and curves are kept.

```python
def run_variant(name, seed, replay=0, mbd=False, epochs=SZ["gan_var_epochs"]):
    set_seed(seed); fresh()
    g, d = build_simple_gan(use_mbd=mbd)
    h, sn = train_gan_tape(g, d, X_train, epochs, batch=SZ["gan_batch"], replay=replay, seed=seed, snap_every=epochs, z_fixed=z_fixed16, verbose=False)
    dv = gan_diversity(g, seed=99)
    return dict(variant=name, seed=seed, classes=dv["classes"], entropy=dv["entropy"], conf=dv["conf"], acc_real=float(np.mean(h["acc_real"][-3:])), acc_fake=float(np.mean(h["acc_fake"][-3:])),
                g_swing=float(np.std(h["g_loss"][-5:])), hist=h, samples=sn[epochs], share=dv["share"])


var_runs = []
for seed in SZ["seeds"][:pick(2, 1)]:
    for name, kw in (("baseline", {}), ("experience replay", dict(replay=50)), ("minibatch diversity", dict(mbd=True))):
        var_runs.append(run_variant(name, seed, **kw))
        print(f"done: {name}, seed {seed}: {var_runs[-1]['classes']} classes, entropy {var_runs[-1]['entropy']:.2f}")
vt = pd.DataFrame([{k: v for k, v in r.items() if k not in ("hist", "samples", "share")} for r in var_runs])
agg = vt.groupby("variant", sort=False).agg(**{"classes found (of 10)": ("classes", lambda v: pm(v, 1)), "class entropy (max 2.30)": ("entropy", lambda v: pm(v, 2)), "classifier confidence": ("conf", lambda v: pm(v, 2)),
                                                "D accuracy real": ("acc_real", "mean"), "D accuracy fake": ("acc_fake", "mean"), "G-loss swing (std, last 5 epochs)": ("g_swing", "mean")})
display(agg.round(3)); display(vt.drop(columns=[]).round(3))
RESULTS["gan_variants"] = vt.to_dict("list")
show_images(np.concatenate([r["samples"][:10] for r in var_runs]), ncols=10, size=1.0, suptitle="final samples (same 10 codings): " + " · ".join(f"{r['variant']} s{r['seed']}" for r in var_runs))
fig, ax = plt.subplots(1, 2, figsize=(10, 3))
for r in var_runs:
    ax[0].plot(r["hist"]["entropy"], label=f"{r['variant']} s{r['seed']}", ls="-" if r["seed"] == SZ["seeds"][0] else "--", color={"baseline": "C0", "experience replay": "C1", "minibatch diversity": "C2"}[r["variant"]])
    ax[1].plot(r["hist"]["acc_fake"], ls="-" if r["seed"] == SZ["seeds"][0] else "--", color={"baseline": "C0", "experience replay": "C1", "minibatch diversity": "C2"}[r["variant"]])
ax[0].set_title("entropy of the class shares among fakes", fontsize=9); ax[0].legend(fontsize=6); ax[1].set_title("discriminator accuracy on fakes", fontsize=9); ax[1].axhline(0.5, ls=":", color="k")
for a in ax:
    a.set_xlabel("epoch")
plt.tight_layout(); plt.show()
for nm in ("experience replay", "minibatch diversity"):
    v_, gap_, noise_ = noise_verdict(vt[vt.variant == "baseline"].entropy.values, vt[vt.variant == nm].entropy.values)
    print(f"class entropy, {nm} minus baseline: {gap_:+.3f} – {v_}")
rb = ReplayBuffer(10)
rb.push(np.ones((5, 28, 28))); rb.push(2 * np.ones((5, 28, 28))); rb.push(3 * np.ones((4, 28, 28)))
vals_, cnts_ = np.unique(rb.buf[:, 0, 0], return_counts=True)
audit("the replay buffer never holds more than its capacity and overwrites the oldest images first (5×1, 5×2, 4×3 pushed into 10 slots → one 1, five 2s, four 3s)", rb.n == 10 and dict(zip(vals_.tolist(), cnts_.tolist())) == {1.0: 1, 2.0: 5, 3.0: 4})
same_ = MinibatchStd()(tf.ones([8, 5])).numpy(); div_ = MinibatchStd()(tf.random.normal([8, 5], seed=0)).numpy()
audit("the minibatch-diversity layer adds exactly one feature: 0 for a batch of identical rows, clearly larger for a diverse batch", same_.shape == (8, 6) and float(same_[0, -1]) < 1e-6 and float(div_[0, -1]) > 0.1, f"{float(same_[0, -1]):.4f} vs {float(div_[0, -1]):.3f}")
```

### 8.1 The five DCGAN guidelines and the small DCGAN (Tasks 1–3)
Radford et al. (2015) found a recipe that makes convolutional GANs train reliably:
1. **Replace pooling layers** with strided convolutions (discriminator) and transposed convolutions (generator).
2. Use **Batch Normalization** in both networks, except in the generator's output layer and the discriminator's input layer.
3. **Remove fully connected hidden layers** for deeper architectures.
4. Use **ReLU** in the generator for all layers except the output, which uses **tanh**.
5. Use **leaky ReLU** in the discriminator for all layers.

The small DCGAN of the source for Fashion MNIST: coding size 100; generator `Dense(7·7·128) → Reshape([7, 7, 128]) → BatchNorm → Conv2DTranspose(64, 5, strides 2, same, selu) → BatchNorm → Conv2DTranspose(1, 5, strides 2, same, tanh)`; discriminator `Conv2D(64, 5, strides 2, same, LeakyReLU(0.2)) → Dropout(0.4) → Conv2D(128, 5, strides 2, same, LeakyReLU(0.2)) → Dropout(0.4) → Flatten → Dense(1, sigmoid)`. The images are rescaled to **[−1, 1]** (`X · 2 − 1`) to match the generator's tanh output.

**Two deviations from the guidelines in the source**, and why: the discriminator uses **Dropout instead of Batch Normalization** and the generator uses **SELU instead of ReLU** – training was unstable otherwise. Section 8.2 checks how much the discriminator's normalisation matters.

```python
DC_CODINGS = 100
X_dc = (X_train.reshape(-1, 28, 28, 1) * 2.0 - 1.0).astype("float32")
audit("the DCGAN training images are rescaled to [−1, 1] to match tanh", X_dc.min() >= -1.0 and X_dc.max() <= 1.0 and abs(X_dc.min() + 1) < 1e-6 and abs(X_dc.max() - 1) < 1e-6, f"[{X_dc.min():.1f}, {X_dc.max():.1f}]")


def build_dcgan(bn_d=False, codings=DC_CODINGS, cond=False):
    zin = keras.Input(shape=(codings,))
    if cond:
        yin = keras.Input(shape=(10,)); x = L.Concatenate()([zin, yin])
    else:
        x = zin
    x = L.Dense(7 * 7 * 128)(x); x = L.Reshape([7, 7, 128])(x); x = L.BatchNormalization()(x)
    x = L.Conv2DTranspose(64, 5, strides=2, padding="same", activation="selu")(x); x = L.BatchNormalization()(x)
    x = L.Conv2DTranspose(1, 5, strides=2, padding="same", activation="tanh")(x)
    g = keras.Model([zin, yin] if cond else zin, x, name="dc_generator" + ("_cond" if cond else ""))
    iin = keras.Input(shape=(28, 28, 1))
    if cond:
        yin_d = keras.Input(shape=(10,)); maps = L.Lambda(lambda y: tf.tile(tf.reshape(y, [-1, 1, 1, 10]), [1, 28, 28, 1]), output_shape=(28, 28, 10))(yin_d)
        h = L.Concatenate()([iin, maps])
    else:
        h = iin
    h = L.Conv2D(64, 5, strides=2, padding="same")(h)
    if bn_d:
        h = L.BatchNormalization()(h)
    h = L.LeakyReLU(0.2)(h); h = L.Dropout(0.4)(h)
    h = L.Conv2D(128, 5, strides=2, padding="same")(h)
    if bn_d:
        h = L.BatchNormalization()(h)
    h = L.LeakyReLU(0.2)(h); h = L.Dropout(0.4)(h)
    d = keras.Model([iin, yin_d] if cond else iin, L.Dense(1, activation="sigmoid")(L.Flatten()(h)), name="dc_discriminator" + ("_cond" if cond else ""))
    return g, d


g_a, d_a = build_dcgan()
shapes_g = [l.output.shape[1:] for l in g_a.layers if l.__class__.__name__ in ("Reshape", "Conv2DTranspose")]
shapes_d = [l.output.shape[1:] for l in d_a.layers if l.__class__.__name__ == "Conv2D"]
print("generator:", [tuple(s) for s in shapes_g], "· discriminator conv outputs:", [tuple(s) for s in shapes_d])
audit("shape audit generator: 7×7×128 → 14×14×64 → 28×28×1", [tuple(s) for s in shapes_g] == [(7, 7, 128), (14, 14, 64), (28, 28, 1)])
audit("shape audit discriminator: 28×28×1 → 14×14×64 → 7×7×128 → 1 probability", [tuple(s) for s in shapes_d] == [(14, 14, 64), (7, 7, 128)] and d_a.output_shape == (None, 1))
print(f"DCGAN generator {g_a.count_params():,} parameters · discriminator {d_a.count_params():,} parameters")
n_bn_g = sum(isinstance(l, L.BatchNormalization) for l in g_a.layers); n_bn_d = sum(isinstance(l, L.BatchNormalization) for l in d_a.layers)
audit("as in the source: Batch Norm in the generator (2 layers), none in the discriminator (Dropout instead), tanh output", n_bn_g == 2 and n_bn_d == 0 and g_a.layers[-1].activation.__name__ == "tanh")
```

```python
def gan_eval_tanh(g, n=2000, seed=0, codings=DC_CODINGS, labels=None):
    z = tf.random.normal([n, codings], seed=seed)
    out = g(z, training=False) if labels is None else g([z, tf.one_hot(labels, 10)], training=False)
    fakes = (np.squeeze(out.numpy()) + 1) / 2
    p = fashion_clf.predict(fakes, batch_size=500, verbose=0); share = np.bincount(p.argmax(1), minlength=10) / n
    ent = float(-(share[share > 0] * np.log(share[share > 0])).sum())
    return dict(entropy=ent, classes=int((share > 0.02).sum()), conf=float(p.max(1).mean()), share=share, pred=p.argmax(1), fakes=fakes)


def train_dcgan(g, d, X, y=None, steps=1000, batch=64, seed=0, lr_g=2e-4, lr_d=2e-4, replay=0, z_fixed=None, y_fixed=None, snap_at=(), n_eval=6, cond=False):
    opt_g = keras.optimizers.Adam(learning_rate=lr_g, beta_1=0.5); opt_d = keras.optimizers.Adam(learning_rate=lr_d, beta_1=0.5)
    bce = keras.losses.BinaryCrossentropy()
    call_g = (lambda z, y_, tr: g([z, y_], training=tr)) if cond else (lambda z, y_, tr: g(z, training=tr))
    call_d = (lambda im, y_, tr: d([im, y_], training=tr)) if cond else (lambda im, y_, tr: d(im, training=tr))

    @tf.function
    def make_fakes(z, y_):
        return call_g(z, y_, True)

    @tf.function
    def d_step(real, fake, y_real, y_fake):
        with tf.GradientTape() as tape:
            pr, pf = call_d(real, y_real, True), call_d(fake, y_fake, True)
            loss = 0.5 * (bce(tf.ones_like(pr), pr) + bce(tf.zeros_like(pf), pf))
        opt_d.apply_gradients(zip(tape.gradient(loss, d.trainable_variables), d.trainable_variables))
        return loss, tf.reduce_mean(tf.cast(pr > 0.5, tf.float32)), tf.reduce_mean(tf.cast(pf < 0.5, tf.float32))

    @tf.function
    def g_step(z, y_):
        with tf.GradientTape() as tape:
            pf = call_d(call_g(z, y_, True), y_, True)
            loss = bce(tf.ones_like(pf), pf)
        opt_g.apply_gradients(zip(tape.gradient(loss, g.trainable_variables), g.trainable_variables))
        return loss

    buf = ReplayBuffer(replay * batch, shape=(28, 28, 1), seed=seed) if replay else None
    ybuf = np.zeros((replay * batch, 10), "float32") if (replay and cond) else None
    data = tf.data.Dataset.from_tensor_slices((X, tf.one_hot(y, 10) if y is not None else tf.zeros([len(X), 10]))).shuffle(2000, seed=seed).repeat().batch(batch, drop_remainder=True).prefetch(1)
    hist = {k: [] for k in ("step", "d_loss", "g_loss", "acc_real", "acc_fake")}; evals = {"step": [], "entropy": [], "classes": [], "conf": []}; snaps = {}; acc = []
    log_every = max(1, min(50, steps // 8))
    eval_steps = set(np.linspace(steps // n_eval, steps, n_eval).astype(int).tolist()); t0 = time.time()
    for step, (xb, yb) in enumerate(data.take(steps), 1):
        z = tf.random.normal([batch, DC_CODINGS]); fake = make_fakes(z, yb)
        if buf is not None:
            buf.push(fake.numpy())
            if ybuf is not None:
                for i, l_ in enumerate(yb.numpy()):
                    ybuf[(buf.ptr - len(yb) + i) % buf.cap] = l_
            sel = buf.rng.integers(0, buf.n, batch)
            fake_d = tf.constant(buf.buf[sel]); y_fake = tf.constant(ybuf[sel]) if ybuf is not None else yb
        else:
            fake_d, y_fake = fake, yb
        dl, ar, af = d_step(xb, fake_d, yb, y_fake)
        gl = g_step(tf.random.normal([batch, DC_CODINGS]), yb)
        acc.append((float(dl), float(gl), float(ar), float(af)))
        if step % log_every == 0:
            a = np.mean(acc, axis=0); acc = []
            for k, v in zip(("d_loss", "g_loss", "acc_real", "acc_fake"), a):
                hist[k].append(float(v))
            hist["step"].append(step)
        if step in eval_steps:
            ev = gan_eval_tanh(g, seed=step, labels=np.random.default_rng(step).integers(0, 10, 2000) if cond else None)
            for k in ("entropy", "classes", "conf"):
                evals[k].append(ev[k])
            evals["step"].append(step)
        if step in snap_at and z_fixed is not None:
            o = g(z_fixed if not cond else [z_fixed, tf.one_hot(y_fixed, 10)], training=False).numpy()
            snaps[step] = (np.squeeze(o) + 1) / 2
    return hist, evals, snaps, time.time() - t0
```

### 8.2 Sensitivity study (Task 4)
GANs are touchy. Five short runs (`dc_var_steps` steps each), changing one thing at a time from the baseline: **(a)** a different random seed, **(b)** the generator's learning rate 2× the discriminator's, **(c)** the opposite, **(d)** Batch Normalization in the discriminator. Every sample grid is shown; settings that fail are named from the numbers (few classes, a discriminator that wins completely or loses completely).

```python
def run_dc_variant(name, seed=11, lr_g=2e-4, lr_d=2e-4, bn_d=False, replay=0, steps=SZ["dc_var_steps"]):
    set_seed(seed); fresh()
    g, d = build_dcgan(bn_d=bn_d)
    h, ev, sn, sec = train_dcgan(g, d, X_dc, steps=steps, seed=seed, lr_g=lr_g, lr_d=lr_d, replay=replay, z_fixed=z_dc, snap_at=(steps,))
    dv = gan_eval_tanh(g, seed=99)
    return dict(variant=name, seed=seed, classes=dv["classes"], entropy=dv["entropy"], conf=dv["conf"], acc_real=float(np.mean(h["acc_real"][-4:])), acc_fake=float(np.mean(h["acc_fake"][-4:])), d_loss=float(np.mean(h["d_loss"][-4:])), g_loss=float(np.mean(h["g_loss"][-4:])), sec=sec, samples=sn[steps], hist=h, gen=g)


sens = [run_dc_variant("baseline (seed 11)"), run_dc_variant("different seed (22)", seed=22), run_dc_variant("G lr = 2 × D lr", lr_g=4e-4, lr_d=2e-4),
        run_dc_variant("D lr = 2 × G lr", lr_g=2e-4, lr_d=4e-4), run_dc_variant("BatchNorm in the discriminator", bn_d=True)]
st = pd.DataFrame([{k: v for k, v in r.items() if k not in ("samples", "hist", "gen")} for r in sens]).set_index("variant")
display(st.round(3))
fig, axes = plt.subplots(len(sens), 1, figsize=(10, 1.45 * len(sens)))
for ax, r in zip(axes, sens):
    ax.imshow(np.concatenate(list(r["samples"][:12]), axis=1), cmap="binary", vmin=0, vmax=1); ax.set_xticks([]); ax.set_yticks([]); ax.set_ylabel(r["variant"], rotation=0, ha="right", va="center", fontsize=7)
plt.tight_layout(); plt.show()
RESULTS["dc_sensitivity"] = st.reset_index().to_dict("list")
base_ent = sens[0]["entropy"]
for r in sens[1:]:
    flag = []
    if r["classes"] <= max(2, sens[0]["classes"] - 3):
        flag.append("loses classes")
    if r["acc_fake"] > 0.95 and r["acc_real"] > 0.95:
        flag.append("discriminator wins completely")
    if r["acc_fake"] < 0.1:
        flag.append("generator wins completely")
    print(f"{r['variant']:34s}: {r['classes']} classes, entropy {r['entropy']:.2f} (baseline {base_ent:.2f}), D acc {r['acc_real']:.2f}/{r['acc_fake']:.2f} → " + (", ".join(flag) if flag else "no obvious failure by these measures"))
print("A single seed per setting: differences of this size can come from the seed alone (compare the 'different seed' row with the baseline).")
```

### 8.3 Latent arithmetic (Task 5)
Training in the brief is 50 epochs; here the DCGAN gets `dc_steps` steps (Section 8.1b), so only classes that the generator actually produces can be used: *sneaker* and *sandal* when both are generated (the brief's pair), otherwise the two most frequent classes.

The DCGAN paper averaged the codings of three "man with glasses" images, three "man without glasses" and three "woman without glasses", and **`men with glasses − men without glasses + women without glasses`** produced a **woman with glasses**; adding a little noise around the result gave many neighbouring examples. The same idea on Fashion MNIST: label 3 000 generated images with the helper classifier, take two classes that the generator produces, and compute `mean(class A codings) − mean(class B codings) + another class-B coding`. If the codings are organised linearly, the result should look like class A.

```python
set_seed(5)
z_all = tf.random.normal([3000, DC_CODINGS], seed=21)
fk = ((np.squeeze(g_dc(z_all, training=False).numpy()) + 1) / 2)
pr_all = fashion_clf.predict(fk, batch_size=500, verbose=0); lab = pr_all.argmax(1); conf = pr_all.max(1)
counts = np.bincount(lab, minlength=10)
order = np.argsort(-counts)
print("classes among 3 000 generated images (as labelled by the helper classifier):", {FASHION_CLASSES[i]: int(counts[i]) for i in order if counts[i] > 0})
cA, cB = int(order[0]), int(order[1])
if counts[7] >= 3 and counts[5] >= 4:
    cA, cB = 7, 5                  # as in the brief: sneaker − sandal + another sandal
pick_ = lambda c, k: np.flatnonzero(lab == c)[np.argsort(-conf[lab == c])][:k]      # the k most confidently recognised members of class c
iA, iB = pick_(cA, 3), pick_(cB, 4)
if counts[cB] < 4:
    print("fewer than two classes were generated, so the arithmetic cannot be demonstrated for this generator.")
    RESULTS["latent_arith"] = dict(A=None, B=None, share_A=float("nan"), share_ctrl=float("nan"))
else:
    zn = z_all.numpy()
    v_arith = zn[iA].mean(0) - zn[iB[:3]].mean(0) + zn[iB[3]]
    rng = np.random.default_rng(0)
    nbrs = v_arith[None] + 0.25 * rng.standard_normal((200, DC_CODINGS)).astype("float32")
    res = (np.squeeze(g_dc(nbrs, training=False).numpy()) + 1) / 2
    res_lab = fashion_clf.predict(res, verbose=0).argmax(1)
    ctrl = (np.squeeze(g_dc(zn[iB[3]][None] + 0.25 * rng.standard_normal((200, DC_CODINGS)).astype("float32"), training=False).numpy()) + 1) / 2
    ctrl_lab = fashion_clf.predict(ctrl, verbose=0).argmax(1)
    share_A, share_ctrl = float((res_lab == cA).mean()), float((ctrl_lab == cA).mean())
    print(f"A = {FASHION_CLASSES[cA]}, B = {FASHION_CLASSES[cB]}:  mean(A) − mean(B) + one more B  →  {100 * share_A:.0f} % of 200 noisy neighbours are classified as {FASHION_CLASSES[cA]}   (control: the plain B coding with the same noise: {100 * share_ctrl:.0f} %)")
    fig, axes = plt.subplots(1, 4, figsize=(12.5, 2.2), gridspec_kw=dict(width_ratios=[3, 1, 1, 6]))
    axes[0].imshow(np.concatenate(list(fk[iA]) + list(fk[iB[:3]]), axis=1), cmap="binary"); axes[0].set_title(f"3 × {FASHION_CLASSES[cA]} (A) and 3 × {FASHION_CLASSES[cB]} (B)", fontsize=8)
    axes[1].imshow(fk[iB[3]], cmap="binary"); axes[1].set_title("another B", fontsize=8)
    axes[2].imshow((np.squeeze(g_dc(v_arith[None], training=False).numpy()) + 1) / 2, cmap="binary"); axes[2].set_title("A − B + B'", fontsize=8)
    axes[3].imshow(np.concatenate(list(res[:10]), axis=1), cmap="binary"); axes[3].set_title("noisy neighbours of the result", fontsize=8)
    for a in axes:
        a.axis("off")
    plt.tight_layout(); plt.show()
    RESULTS["latent_arith"] = dict(A=FASHION_CLASSES[cA], B=FASHION_CLASSES[cB], share_A=share_A, share_ctrl=share_ctrl)
print("Honest reading: Fashion MNIST has no 'attributes' as clean as glasses, and a DCGAN that has seen only a few epochs has a rough latent space; the control row shows what the plain B coding with the same noise gives.")
```

### 8.5 Conditional GAN (Task 6)
Add the **class of each image as an extra input to both networks**, so that you can ask for a specific class. The generator gets the one-hot class concatenated to its coding; the discriminator gets the one-hot class turned into 10 constant feature maps that are stacked onto the image as extra channels. The discriminator now judges "is this a *real* image *of this class*?", so the generator must produce the requested class to fool it.

```python
set_seed(11); fresh()
g_c, d_c = build_dcgan(cond=True)
y_dc = y_train.astype("int64")
audit("the conditional generator takes [coding, one-hot class] and the discriminator takes [image, one-hot class]", len(g_c.inputs) == 2 and len(d_c.inputs) == 2 and g_c.inputs[1].shape[-1] == 10)
z_row = tf.random.normal([8, DC_CODINGS], seed=7)
y_grid = np.repeat(np.arange(10), 8)
h_c, ev_c, snaps_c, sec_c = train_dcgan(g_c, d_c, X_dc, y=y_dc, steps=SZ["cgan_steps"], seed=11, z_fixed=tf.tile(z_row, [10, 1]), y_fixed=y_grid, snap_at=(SZ["cgan_steps"],), cond=True)
print(f"{SZ['cgan_steps']} steps in {sec_c / 60:.1f} min")
grid_c = snaps_c[SZ["cgan_steps"]].reshape(10, 8, 28, 28)
fig, ax = plt.subplots(figsize=(7.8, 9.0)); ax.imshow(np.concatenate([np.concatenate(list(grid_c[r]), axis=1) for r in range(10)], axis=0), cmap="binary"); ax.set_yticks(np.arange(10) * 28 + 14); ax.set_yticklabels(FASHION_CLASSES); ax.set_xticks([]); ax.set_title("conditional DCGAN: one row per requested class (same 8 codings in every row)", fontsize=9); ax.grid(False); plt.show()
labels_req = np.random.default_rng(0).integers(0, 10, 3000)
ev_ctrl = gan_eval_tanh(g_c, n=3000, seed=5, labels=labels_req)
ctrl_acc = float((ev_ctrl["pred"] == labels_req).mean())
per_class = [float((ev_ctrl["pred"][labels_req == c] == c).mean()) for c in range(10)]
print(f"class control: the classifier recognises the requested class in {100 * ctrl_acc:.1f} % of 3 000 generated images (chance 10 %)")
display(pd.Series(per_class, index=FASHION_CLASSES, name="share recognised as the requested class").round(2).to_frame().T)
audit_trained("the conditional GAN obeys the requested class far above chance (> 50 % recognised)", lambda: ctrl_acc > 0.5, f"{100 * ctrl_acc:.1f} %")
RESULTS["cgan"] = dict(control=ctrl_acc, per_class=per_class, entropy=ev_ctrl["entropy"], steps=SZ["cgan_steps"], sec=sec_c)
```

### 8.6 Progressive growing of GANs (Task 7)
**Idea (Karras et al., 2018).** Training a GAN directly on large images is slow and unstable. Instead start with tiny images (here 7×7) so that the generator and discriminator first learn the *coarse structure*, then **add one layer at a time** to double the resolution (7 → 14 → 28), so the later layers only refine details. Each new layer is **faded in**: with a weight `α` that grows from 0 to 1 the output (generator) or input (discriminator) is a blend of the *old path* (the old image upsampled / the new image downsampled) and the *new path*. At `α = 0` the network behaves exactly like the previous stage, so nothing that was learned is disturbed; at `α = 1` the new layer carries everything. After the fade-in comes a stable stage at full `α`.

**Where the layers go.** New layers are added at the *end of the generator* (the high-resolution side) and the *beginning of the discriminator*; the earlier layers **stay trainable** and keep adapting. That is the difference from **greedy layer-wise training of stacked autoencoders** (Phase 3), where each layer is trained once and then frozen while the next is trained on its output: progressive growing trains the whole, growing network jointly, and what grows is the *resolution*, not the depth of one fixed-size representation.

Below: a small but real progressive GAN on Fashion MNIST. Its fade-in is **unit-tested** (at `α = 0` the grown generator equals the old one upsampled; at `α = 1` it equals the new path alone; the same for the discriminator), then it is trained through the schedule *7 stable → 14 fade-in → 14 stable → 28 fade-in → 28 stable*, with the same 16 codings shown after every stage.

```python
SZ["prog_steps"] = pick(250, 6)
RES = [7, 14, 28]
act_ = lambda t: tf.nn.leaky_relu(t, 0.2)


class ProgGAN:
    """tiny progressive GAN; the layers of all resolutions exist from the start, the schedule decides which are used"""
    def __init__(self, zdim=64, ch=32):
        self.zdim, self.alpha = zdim, tf.Variable(1.0, trainable=False)
        self.zd = L.Dense(7 * 7 * ch); self.gb = {r: L.Conv2D(ch, 3, padding="same") for r in RES}
        self.to_rgb = {r: L.Conv2D(1, 1) for r in RES}; self.from_rgb = {r: L.Conv2D(ch, 1) for r in RES}
        self.db = {r: L.Conv2D(ch, 3, padding="same") for r in RES}; self.d_out = L.Dense(1, activation="sigmoid")
        self.up, self.pool, self.flat = L.UpSampling2D(2), L.AveragePooling2D(2), L.Flatten()
        self.ch = ch

    def G(self, z, res, fade=False):
        x = tf.reshape(act_(self.zd(z)), [-1, 7, 7, self.ch]); x = act_(self.gb[7](x)); prev = 7
        for r in RES[1:RES.index(res) + 1]:
            if r == res and fade:
                old = self.up(self.to_rgb[prev](x))
            x = act_(self.gb[r](self.up(x)))
            if r == res and fade:
                return tf.tanh((1 - self.alpha) * old + self.alpha * self.to_rgb[r](x))
            prev = r
        return tf.tanh(self.to_rgb[res](x))

    def D(self, x, res, fade=False):
        idx = RES.index(res); h = act_(self.from_rgb[res](x))
        for k in range(idx, -1, -1):
            h = act_(self.db[RES[k]](h))
            if k > 0:
                h = self.pool(h)
                if k == idx and fade:
                    h = (1 - self.alpha) * act_(self.from_rgb[RES[k - 1]](self.pool(x))) + self.alpha * h
        return self.d_out(self.flat(h))

    def layers(self):
        return [self.zd, self.d_out] + list(self.gb.values()) + list(self.to_rgb.values()) + list(self.from_rgb.values()) + list(self.db.values())

    def gvars(self):
        return self.zd.trainable_variables + [v for d_ in (self.gb, self.to_rgb) for l in d_.values() for v in l.trainable_variables]

    def dvars(self):
        return self.d_out.trainable_variables + [v for d_ in (self.from_rgb, self.db) for l in d_.values() for v in l.trainable_variables]


set_seed(11); fresh()
pg = ProgGAN()
_ = pg.G(tf.zeros([2, 64]), 28, True); _ = pg.D(tf.zeros([2, 28, 28, 1]), 28, True)       # build every layer once
zt = tf.random.normal([4, 64], seed=1); xt = tf.random.uniform([4, 28, 28, 1], -1, 1, seed=2)
pg.alpha.assign(0.0)
g_old = pg.G(zt, 7).numpy(); g_f0 = pg.G(zt, 14, True).numpy()
audit("fade-in, generator: at α = 0 the grown generator (14×14) equals the old 7×7 generator, upsampled", np.allclose(g_f0, np.repeat(np.repeat(g_old, 2, 1), 2, 2), atol=1e-5), f"max difference {np.abs(g_f0 - np.repeat(np.repeat(g_old, 2, 1), 2, 2)).max():.1e}")
pg.alpha.assign(1.0)
audit("fade-in, generator: at α = 1 the output is the new path alone (same as the finished 14×14 generator)", np.allclose(pg.G(zt, 14, True).numpy(), pg.G(zt, 14, False).numpy(), atol=1e-6))
pg.alpha.assign(0.0); x14 = xt[:, ::2, ::2, :]
audit("fade-in, discriminator: at α = 0 the grown discriminator on a 14×14 image equals the old one on the image downsampled to 7×7", np.allclose(pg.D(x14, 14, True).numpy(), pg.D(L.AveragePooling2D(2)(x14), 7).numpy(), atol=1e-5))
pg.alpha.assign(1.0)
audit("fade-in, discriminator: at α = 1 the new path alone is used", np.allclose(pg.D(x14, 14, True).numpy(), pg.D(x14, 14, False).numpy(), atol=1e-6))
audit("the schedule changes the resolution of the generator output: 7, 14, 28", [pg.G(zt, r).shape[1] for r in RES] == RES)
```

```python
def train_prog(pg, X, P, batch=64, seed=11):
    """schedule: 7 stable, 14 fade-in, 14 stable, 28 fade-in, 28 stable – P steps each; returns sample grids after each stage"""
    opt_g = keras.optimizers.Adam(2e-3, beta_1=0.5); opt_d = keras.optimizers.Adam(2e-3, beta_1=0.5); bce = keras.losses.BinaryCrossentropy()
    Xr = {28: X, 14: tf.nn.avg_pool2d(X, 2, 2, "VALID").numpy(), 7: tf.nn.avg_pool2d(X, 4, 4, "VALID").numpy()}
    opt_d.build(pg.dvars()); opt_g.build(pg.gvars())          # Keras 3: an optimizer only knows the variables it was built with
    pairs = lambda gr, vs: [(g_, v) for g_, v in zip(gr, vs) if g_ is not None]

    @tf.function
    def step(real, res, fade):
        z = tf.random.normal([batch, pg.zdim])
        with tf.GradientTape() as tp:
            fake = pg.G(z, res, fade); pr, pf = pg.D(real, res, fade), pg.D(fake, res, fade)
            dl = 0.5 * (bce(tf.ones_like(pr), pr) + bce(tf.zeros_like(pf), pf))
        dv = pg.dvars(); opt_d.apply_gradients(pairs(tp.gradient(dl, dv), dv))
        with tf.GradientTape() as tp:
            pf = pg.D(pg.G(tf.random.normal([batch, pg.zdim]), res, fade), res, fade); gl = bce(tf.ones_like(pf), pf)
        gv = pg.gvars(); opt_g.apply_gradients(pairs(tp.gradient(gl, gv), gv))
        return dl, gl

    stages = [(7, False), (14, True), (14, False), (28, True), (28, False)]; rng = np.random.default_rng(seed); out = []; zf = tf.random.normal([16, pg.zdim], seed=3)
    for si, (res, fade) in enumerate(stages):
        for s in range(P):
            pg.alpha.assign((s + 1) / P if fade else 1.0)
            idx = rng.integers(0, len(X), batch); dl, gl = step(tf.constant(Xr[res][idx]), res, fade)
        pg.alpha.assign(1.0)
        out.append(dict(stage=f"{res}×{res} {'fade-in' if fade else 'stable'}", res=res, d=float(dl), g=float(gl), imgs=(np.squeeze(pg.G(zf, res, False).numpy()) + 1) / 2))
    return out


t0 = time.time(); prog_out = train_prog(pg, X_dc, SZ["prog_steps"]); sec_prog = time.time() - t0
print(f"progressive schedule: 5 stages × {SZ['prog_steps']} steps (batch 64) in {sec_prog / 60:.1f} min")
fig, axes = plt.subplots(1, 5, figsize=(14, 2.9))
for ax, o in zip(axes, prog_out):
    r = o["res"]; ax.imshow(np.concatenate([np.concatenate(list(o["imgs"][4 * i:4 * i + 4]), axis=1) for i in range(4)], axis=0), cmap="binary", vmin=0, vmax=1); ax.set_title(f"after {o['stage']}\n(D {o['d']:.2f} · G {o['g']:.2f})", fontsize=8); ax.axis("off")
plt.tight_layout(); plt.show()
fin = prog_out[-1]["imgs"]; pcl = fashion_clf.predict(fin, verbose=0)
RESULTS["progressive"] = dict(steps=SZ["prog_steps"], conf=float(pcl.max(1).mean()), classes=int(len(set(pcl.argmax(1)))))
print(f"final 28×28 images: classifier confidence {pcl.max(1).mean():.2f}, {len(set(pcl.argmax(1)))} different classes among 16 codings")
print("Honest note: with a few hundred steps per stage the pictures are rough blobs. What is shown is the mechanism – the schedule, the fade-in and its exact α = 0 / α = 1 behaviour – not a competitive generator.")
```

### 8.7 The three special layers of progressive GANs (Task 8) – with unit tests
**(a) Minibatch standard deviation.** An extra feature map that holds one number – how much the examples in the batch differ from each other. If the generator collapses to one picture this number is near 0 and the discriminator can recognise the fake. The brief's version computes `tf.reduce_mean(tf.math.reduce_std(x, axis=[0, -1]))` and appends it as one more channel. (The layer is the same `MinibatchStd` as in Phase 7, here with `axes = (0, −1)` for 4-D feature maps.)

**(b) Equalized learning rate.** Weights are stored as plain `N(0, 1)` numbers and multiplied by the He constant `√(2 / fan_in)` **at run time**. (The brief says "divide by `√(2 / n_inputs)`"; the paper's formula `ŵ = w / c` uses a constant `c = √(fan_in / 2)`, so the effective weights get the standard deviation `√(2 / fan_in)` of He initialisation. Dividing by `√(2 / fan_in)` literally would be wrong – it is tested below – so the layer multiplies, and the change is logged.) Then every weight lives on the same scale, so one learning rate (and Adam-style optimizers that normalise gradients) updates *all* layers at the same relative speed – whereas with ordinary He initialisation, a layer with a large fan-in has tiny stored weights and a fixed-size update changes it much more.

**(c) Pixelwise normalization.** After each generator convolution every pixel's feature vector is rescaled to unit average magnitude: `x / √(mean(x², over channels) + ε)`. It stops the signal magnitudes of the two networks from spiralling out of control.

```python
class EqualizedDense(L.Layer):
    def __init__(self, units, **kw):
        super().__init__(**kw); self.units = units

    def build(self, s):
        self.w = self.add_weight(shape=(s[-1], self.units), initializer=keras.initializers.RandomNormal(0, 1), name="w")
        self.b = self.add_weight(shape=(self.units,), initializer="zeros", name="b"); self.c = math.sqrt(2.0 / s[-1])

    def call(self, x):
        return tf.matmul(x, self.w * self.c) + self.b


class EqualizedConv2D(L.Layer):
    def __init__(self, filters, k=3, **kw):
        super().__init__(**kw); self.filters, self.k = filters, k

    def build(self, s):
        self.w = self.add_weight(shape=(self.k, self.k, s[-1], self.filters), initializer=keras.initializers.RandomNormal(0, 1), name="w")
        self.b = self.add_weight(shape=(self.filters,), initializer="zeros", name="b"); self.c = math.sqrt(2.0 / (self.k * self.k * s[-1]))

    def call(self, x):
        return tf.nn.conv2d(x, self.w * self.c, 1, "SAME") + self.b


class PixelNorm(L.Layer):
    def __init__(self, eps=1e-8, **kw):
        super().__init__(**kw); self.eps = eps

    def call(self, x):
        return x * tf.math.rsqrt(tf.reduce_mean(tf.square(x), axis=-1, keepdims=True) + self.eps)


rs = np.random.default_rng(0)
# --- minibatch std
xs = tf.constant(rs.normal(size=(16, 5, 5, 8)).astype("float32")); mb = MinibatchStd(axes=(0, -1))(xs).numpy()
expected = float(np.mean(np.std(xs.numpy(), axis=(0, 3))))
audit("MinibatchStd: output has one extra channel, the original channels are unchanged", mb.shape == (16, 5, 5, 9) and np.allclose(mb[..., :8], xs.numpy()))
audit("MinibatchStd: the extra channel equals mean(std over axes [0, −1]) computed with numpy and is the same at every position", np.allclose(mb[..., 8], expected, atol=1e-5) , f"{mb[0, 0, 0, 8]:.4f} vs {expected:.4f}")
audit("MinibatchStd: a constant batch gives 0", float(MinibatchStd(axes=(0, -1))(tf.ones([16, 5, 5, 8])).numpy()[..., 8].max()) == 0.0)
gray = lambda t: tf.tile(tf.reduce_mean(t, axis=-1, keepdims=True), [1, 1, 1, 8])        # images whose 8 channels are equal
same = gray(tf.tile(xs[:1], [16, 1, 1, 1])); diverse = gray(xs)
v_same, v_div = float(MinibatchStd(axes=(0, -1))(same).numpy()[0, 0, 0, 8]), float(MinibatchStd(axes=(0, -1))(diverse).numpy()[0, 0, 0, 8])
print(f"added channel for 16 identical images: {v_same:.4f} · for 16 different images: {v_div:.4f}")
audit("MinibatchStd (brief's axes [0, −1]): the added channel is ≈ 0 for a batch of identical images and larger for diverse images", v_same < 1e-6 and v_div > 0.05, f"{v_same:.4f} vs {v_div:.4f}")
same_ch = tf.tile(xs[:1], [16, 1, 1, 1])
v_both, v_batch = float(MinibatchStd(axes=(0, -1))(same_ch).numpy()[0, 0, 0, 8]), float(MinibatchStd(axes=(0,))(same_ch).numpy()[0, 0, 0, 8])
print(f"if the identical images have different values in their different channels: axes [0, −1] gives {v_both:.3f} (it also counts the spread between channels), the batch axis only gives {v_batch:.3f}")
audit("the batch-axis-only statistic is exactly 0 for 16 identical images whatever their channels contain", v_batch == 0.0)
print("→ why it helps: the discriminator sees this number for the whole batch. A collapsed generator produces a batch with a value near 0, which never happens for real batches, so the discriminator can tell the batch is fake and the generator is pushed to produce varied examples. Single images can no longer fool it one by one.")
# --- equalized lr
d_eq, d_he = EqualizedDense(64), L.Dense(64, kernel_initializer="he_normal")
xin = tf.constant(rs.normal(size=(4000, 256)).astype("float32")); y_eq, y_he = d_eq(xin).numpy(), d_he(xin).numpy()
audit("equalized lr: the stored weights are N(0, 1) (std ≈ 1)", abs(float(d_eq.w.numpy().std()) - 1) < 0.05, f"{d_eq.w.numpy().std():.3f}")
audit("equalized lr: the layer output has the same scale as a He-initialised layer (variance ≈ 2 for unit input)", abs(y_eq.var() - y_he.var()) / y_he.var() < 0.1, f"{y_eq.var():.2f} vs {y_he.var():.2f}")
ratio = {}
for fan in (10, 1000):
    e, h = EqualizedDense(8), L.Dense(8, kernel_initializer="he_normal"); e(tf.zeros([1, fan])); h(tf.zeros([1, fan]))
    step_ = 0.01      # one sign-like (Adam-style) update of the stored weights
    rel_e = step_ * e.c / (np.std(e.w.numpy()) * e.c); rel_h = step_ / np.std(h.kernel.numpy())
    ratio[fan] = (rel_e, rel_h)
print("relative change of the effective weights after one update of size 0.01 on the stored weights:", {f"fan-in {k}": f"equalized {a:.3f} · He-init {b:.3f}" for k, (a, b) in ratio.items()})
audit("equalized lr: the same update changes layers of different fan-in equally (within 10 %), whereas with He initialisation the large layer changes much more (> 5×)", abs(ratio[10][0] - ratio[1000][0]) / ratio[10][0] < 0.1 and ratio[1000][1] / ratio[10][1] > 5)
wrong_var = float((d_eq.w.numpy() / d_eq.c).var()); right_var = float((d_eq.w.numpy() * d_eq.c).var())
audit("dividing by √(2 / fan_in) literally would give effective weights with a huge variance; multiplying gives the He variance 2 / fan_in", right_var < 0.01 and wrong_var > 100 * right_var, f"multiply {right_var:.4f} · divide {wrong_var:.1f}")
log_change("equalized lr: 'divide by sqrt(2 / n_inputs)' (brief)", "multiply by sqrt(2 / n_inputs) at run time", "equivalent to the paper's w / c with c = sqrt(n_inputs / 2); the literal division gives a wrong scale (tested)")
ec = EqualizedConv2D(16, 3); out = ec(tf.constant(rs.normal(size=(8, 12, 12, 8)).astype("float32")))
audit("equalized conv: output shape (8, 12, 12, 16), stored weights std ≈ 1, effective weights std ≈ √(2 / (3·3·8))", out.shape == (8, 12, 12, 16) and abs(ec.w.numpy().std() - 1) < 0.1 and abs(ec.w.numpy().std() * ec.c - math.sqrt(2 / 72)) < 0.02)
# --- pixelnorm
raw_pn = (5 * rs.normal(size=(6, 7, 7, 32))).astype("float32"); pn = PixelNorm()(tf.constant(raw_pn)).numpy()
audit("pixelwise norm: the mean square over channels is 1 at every pixel", np.allclose((pn ** 2).mean(-1), 1.0, atol=1e-4), f"{(pn ** 2).mean(-1).min():.4f}…{(pn ** 2).mean(-1).max():.4f}")
cos = float(raw_pn[0, 0, 0] @ pn[0, 0, 0] / (np.linalg.norm(raw_pn[0, 0, 0]) * np.linalg.norm(pn[0, 0, 0])))
audit("pixelwise norm: finite on an all-zero input, and the direction of each feature vector is kept (cosine 1)", np.isfinite(PixelNorm()(tf.zeros([2, 3, 3, 4])).numpy()).all() and cos > 0.9999)
```

### 8.8 How to evaluate a GAN (Task 9)
A GAN has no loss that tells you how good the pictures are – the discriminator loss only says how well *this* discriminator is doing against *this* generator. So evaluation is hard and is done in several ways, each with a blind spot:

| Method | What it measures | Blind spot |
|---|---|---|
| Looking at samples | realism, artefacts | subjective; cannot see mode dropping if you only view a few; cherry-picking |
| Inception Score (IS) | an image classifier is confident (quality) and the classes are spread out (diversity) | no comparison with the real data; ignores diversity *within* a class; depends on the ImageNet classifier |
| Fréchet Inception Distance (FID) | distance between the Gaussian fitted to real-image features and the one fitted to fake-image features (lower is better) | needs thousands of samples, biased with small samples, depends on the feature network, and only captures mean and covariance |
| Precision / recall of distributions | precision: fakes that look real; recall: real data that the generator can reach | needs a feature space and choices of k |
| Human evaluation | the final judge | slow, costly, noisy |
| Nearest-neighbour check | is the generator just copying training images? | pixel distance is a poor similarity measure |
| Task-based | does synthetic data help train a classifier? | measures usefulness, not realism |

**Optional, and not in the source:** a *feature-space Fréchet distance* on a small scale. The Inception network is not available offline here, so the features come from our own helper Fashion MNIST classifier (the 1600 numbers before its output layer). This is **not comparable with published FID values**, but it is useful as a relative measure between our own generators. It is sanity-checked first.

```python
feat_net = keras.Model(fashion_clf.inputs, fashion_clf.layers[-3].output)       # Flatten output of the helper classifier


def feats(imgs):
    return feat_net.predict(np.squeeze(imgs).reshape(-1, 28, 28), batch_size=500, verbose=0).astype("float64")


def frechet(f1, f2):
    m1, m2 = f1.mean(0), f2.mean(0); c1, c2 = np.cov(f1, rowvar=False), np.cov(f2, rowvar=False)
    ev, U = np.linalg.eigh(c1); S = (U * np.sqrt(np.clip(ev, 0, None))) @ U.T          # symmetric square root of c1
    tr_sqrt = np.sqrt(np.clip(np.linalg.eigvalsh(S @ c2 @ S), 0, None)).sum()          # trace of sqrt(c1 c2)
    return float(((m1 - m2) ** 2).sum() + np.trace(c1) + np.trace(c2) - 2 * tr_sqrt)


n_f = 2000
fr_real_a, fr_real_b = feats(X_valid[: n_f // 2 if len(X_valid) >= n_f else len(X_valid) // 2]), feats(X_test[:n_f // 2])
fd_same, fd_noise = frechet(fr_real_a, fr_real_b), frechet(fr_real_a, feats(np.random.default_rng(0).uniform(size=(len(fr_real_b), 28, 28))))
print(f"sanity: real vs real = {fd_same:.1f} · real vs random noise = {fd_noise:.1f}")
audit("the feature-space Fréchet distance is small for two samples of real images and large for noise (> 10×)", fd_noise > 10 * fd_same, f"{fd_same:.1f} vs {fd_noise:.1f}")
rows = {}
fr_real = feats(X_test[:n_f])
rows["real test images (second half)"] = frechet(fr_real_a, feats(X_test[n_f // 2:n_f]))
rows["VAE samples (Phase 6)"] = frechet(fr_real, feats(vdec(tf.random.normal([n_f, 10])).numpy()))
rows["dense GAN (Phase 7)"] = frechet(fr_real, feats(g2(tf.random.normal([n_f, CODINGS]), training=False).numpy()))
rows["DCGAN (8.1)"] = frechet(fr_real, feats(gan_eval_tanh(g_dc, n=n_f, seed=7)["fakes"]))
rows["conditional DCGAN (8.5, random classes)"] = frechet(fr_real, feats(gan_eval_tanh(g_c, n=n_f, seed=7, labels=np.random.default_rng(1).integers(0, 10, n_f))["fakes"]))
fd_tab = pd.Series(rows, name="feature-space Fréchet distance (lower = closer to real)").round(1).to_frame(); display(fd_tab)
RESULTS["frechet"] = rows
print("Read with care: a single run per model, our own classifier's features, and generators trained for very different numbers of steps – this ranks these particular runs, it is not a verdict on the model families.")
```

### 8.9 StyleGAN – the generator, annotated (Tasks 10–12)
StyleGAN (Karras et al., 2018) keeps the discriminator and the loss of a progressive GAN and redesigns the **generator** (the figure below is drawn from the description in the brief):

* **Mapping network** – an 8-layer MLP turns the coding `z` into an intermediate vector `w`. (`z` has to follow a fixed Gaussian; `w` is free to be shaped so that factors of variation become easy to separate.)
* **Affine transformations "A"** – one `Dense` layer without activation per synthesis level turns `w` into a *style vector* (a scale and a bias for every feature map). Early levels control high-level traits (pose, adult vs. child, the overall shape of the garment), late levels control fine texture (hair colour, fabric pattern).
* **Synthesis network** – starts from a **learned constant** (not from the coding), then repeats *upsampling → convolution → noise → AdaIN*.
* **Noise "B"** – at each level one Gaussian feature map is drawn, broadcast over all channels, multiplied by **learned per-channel factors** and added *before the activation*.
* **AdaIN** (adaptive instance normalisation) – each feature map is standardised on its own (mean 0, std 1 over its pixels), then scaled and shifted by the style's scale and bias for that map.

```python
fig, ax = plt.subplots(figsize=(11.5, 4.9)); ax.set_xlim(0, 22); ax.set_ylim(0, 10); ax.axis("off")
box = lambda x, y, w, h, t, c="#dfe9f5", fs=8: (ax.add_patch(plt.Rectangle((x, y), w, h, fc=c, ec="#445", lw=1)), ax.text(x + w / 2, y + h / 2, t, ha="center", va="center", fontsize=fs))
arr = lambda x0, y0, x1, y1: ax.annotate("", xy=(x1, y1), xytext=(x0, y0), arrowprops=dict(arrowstyle="->", color="#334"))
box(0.2, 4.2, 1.6, 1.6, "z\n(coding)", "#fff2cc"); arr(1.8, 5, 2.6, 5)
box(2.6, 3.2, 3.2, 3.6, "mapping network\n8 × Dense\n(MLP)", "#e2f0d9"); arr(5.8, 5, 6.6, 5); box(6.6, 4.2, 1.4, 1.6, "w", "#fff2cc")
for i, (yy, name) in enumerate([(8.0, "A₁"), (5.0, "A₂"), (2.0, "A₃")]):
    arr(8.0, 5, 9.0, yy + 0.5); box(9.0, yy, 1.5, 1.0, f"{name}\nDense", "#fce4d6", 7)
ax.text(9.75, 9.35, "style vectors\n(scale, bias per map)", ha="center", fontsize=7)
box(12.0, 0.2, 2.0, 1.0, "learned\nconstant", "#fff2cc", 7)
for x0, lab in [(14.8, "7×7 level"), (17.0, "14×14 level"), (19.2, "28×28 level")]:
    box(x0, 1.0, 1.8, 1.2, "up-sample\nconv", "#dfe9f5", 6.5); box(x0, 2.6, 1.8, 0.9, "+ noise (B)", "#ededed", 6.5); box(x0, 4.0, 1.8, 0.9, "AdaIN", "#fce4d6", 6.5)
    ax.text(x0 + 0.9, 5.4, lab, ha="center", fontsize=7, weight="bold")
for yy_src, x_dst, yy_dst in [(8.5, 14.8, 4.45), (5.5, 17.0, 4.45), (2.5, 19.2, 4.45)]:
    ax.plot([10.5, x_dst - 0.3, x_dst - 0.3], [yy_src, yy_src, yy_dst], color="#b55", lw=0.8)
arr(14.0, 0.7, 14.8, 1.4); ax.text(21.3, 0.8, "→ image\n28×28", fontsize=7, ha="center")
ax.set_title("StyleGAN-style generator: z → mapping network → w → A (styles) → AdaIN at every level;  constant input;  noise B at every level", fontsize=9)
plt.tight_layout(); plt.show()
```

### 8.10 Why separate noise inputs matter (Task 11)
Some parts of an image are truly random: freckles, individual hair strands, the exact folds of a fabric. If that randomness had to come from the coding `z`, then (1) the coding would spend part of its limited capacity on things that do not matter for what the picture *is*, (2) the noise would have to be carried deep through the whole network up to the layer where it is used, and (3) if the generator instead made its own pseudo-random numbers from the activations, it would waste weights on that job and could produce visible repeating patterns (artefacts). Giving the network **extra noise inputs at every level** avoids all three: `z`/`w` describe *what* is in the picture, the noise decides the unimportant *details*. The experiment in 8.13 measures this on the toy model: changing only the noise should change the picture much less than changing the coding.

### 8.11 Mixing regularization – style mixing (Task 12)
During training, for some percentage of the images two codings `z₁`, `z₂` are sent through the mapping network to give `w₁`, `w₂`. The first levels of the synthesis network use `w₁` and the rest use `w₂`, with a **random cutoff level**. The network can therefore no longer assume that the styles of neighbouring levels are correlated; each style vector has to be useful on its own and is pushed to influence only a limited set of traits (**locality**). The same trick at test time is *style mixing*: take the coarse traits from one image and the fine traits from another.

### 8.12 The AdaIN layer and the toy style-based generator (Task 13)
`AdaIN(x, w)`: standardise every feature map of `x` independently, then `y = (1 + y_s) · x̂ + y_b`, where `(y_s, y_b)` come from a small Dense layer applied to the style `w` (the offset `1 +` makes a zero-initialised style start at plain instance normalisation). The toy generator: mapping MLP (4 layers here, not 8 – the toy is small) → a learned 7×7 constant → three levels (7×7, 14×14, 28×28), each *(transposed) convolution → learned-scale noise → LeakyReLU → AdaIN*, one style per level, a 1×1 output convolution with tanh. It is trained against the discriminator of Section 8.1 for `style_steps` steps with **mixing regularization** (50 % of the images use two codings and a random cutoff). *A faithful StyleGAN is out of scope; this shows the mechanism.*

```python
class AdaIN(L.Layer):
    def __init__(self, eps=1e-5, **kw):
        super().__init__(**kw); self.eps = eps

    def build(self, shapes):
        self.c = shapes[0][-1]; self.affine = L.Dense(2 * self.c); self.affine.build(shapes[1])

    @staticmethod
    def modulate(x, ys, yb, eps=1e-5):
        mu = tf.reduce_mean(x, axis=[1, 2], keepdims=True); var = tf.math.reduce_variance(x, axis=[1, 2], keepdims=True)
        return (1.0 + ys[:, None, None, :]) * (x - mu) * tf.math.rsqrt(var + eps) + yb[:, None, None, :]

    def call(self, inputs):
        x, w = inputs; ys, yb = tf.split(self.affine(w), 2, axis=-1)
        return AdaIN.modulate(x, ys, yb, self.eps)


class NoiseB(L.Layer):
    """adds one Gaussian map (shared by all channels) multiplied by a learned factor per channel (init 0)"""
    def build(self, s):
        self.scale = self.add_weight(shape=(s[-1],), initializer="zeros", name="noise_scale")

    def call(self, x, noise=None):
        if noise is None:
            noise = tf.random.normal(tf.concat([tf.shape(x)[:3], [1]], 0))
        return x + noise * self.scale


rs = np.random.default_rng(1)
xa = tf.constant((3.0 * rs.normal(size=(5, 9, 9, 6)) + 2.0).astype("float32"))
ys_t, yb_t = tf.constant(rs.uniform(-0.5, 1.0, size=(5, 6)).astype("float32")), tf.constant(rs.normal(size=(5, 6)).astype("float32"))
out_a = AdaIN.modulate(xa, ys_t, yb_t).numpy()
audit("AdaIN: every feature map has mean = style bias and std = 1 + style scale", np.allclose(out_a.mean((1, 2)), yb_t.numpy(), atol=1e-4) and np.allclose(out_a.std((1, 2)), 1 + ys_t.numpy(), atol=1e-3), f"max deviation of mean {np.abs(out_a.mean((1, 2)) - yb_t.numpy()).max():.1e}, of std {np.abs(out_a.std((1, 2)) - 1 - ys_t.numpy()).max():.1e}")
ad = AdaIN(); ad.build([xa.shape, (None, 8)]); ad.affine.kernel.assign(tf.zeros_like(ad.affine.kernel))
out_0 = ad([xa, tf.constant(rs.normal(size=(5, 8)).astype("float32"))]).numpy()
audit("AdaIN layer: a zero affine map gives plain instance normalisation (mean 0, std 1 per map), whatever the style vector is", np.allclose(out_0.mean((1, 2)), 0, atol=1e-4) and np.allclose(out_0.std((1, 2)), 1, atol=1e-3))
ad.affine.kernel.assign(tf.constant(rs.normal(size=ad.affine.kernel.shape).astype("float32")))
audit("AdaIN layer: different style vectors give different outputs (the style really controls the maps)", not np.allclose(ad([xa, tf.ones([5, 8])]).numpy(), ad([xa, -tf.ones([5, 8])]).numpy()))
nb_ = NoiseB(); xz = tf.zeros([32, 16, 16, 4]); nb_.build(xz.shape)
audit("noise layer: with the initial scale 0 it adds nothing", np.allclose(nb_(xz).numpy(), 0))
nb_.scale.assign([0.0, 1.0, 2.0, 3.0]); on = nb_(xz).numpy()
audit("noise layer: the added noise has std = learned scale per channel, is identical across channels up to that factor, and differs from call to call", np.allclose(on.std((0, 1, 2)), [0, 1, 2, 3], atol=0.1) and np.allclose(on[..., 2], 2 * on[..., 1]) and not np.allclose(on, nb_(xz).numpy()))

LEVELS = 3


class ToyStyleGen:
    def __init__(self, zdim=64, wdim=64, ch=48):
        self.zdim, self.ch = zdim, ch
        self.mapping = [L.Dense(wdim) for _ in range(4)]
        self.const = tf.Variable(tf.random.normal([1, 7, 7, ch], stddev=0.5), name="constant")
        self.c7 = L.Conv2D(ch, 3, padding="same"); self.t14 = L.Conv2DTranspose(ch, 4, strides=2, padding="same"); self.t28 = L.Conv2DTranspose(ch // 2, 4, strides=2, padding="same")
        self.noise = [NoiseB() for _ in range(LEVELS)]; self.adain = [AdaIN() for _ in range(LEVELS)]; self.out = L.Conv2D(1, 1)

    def w(self, z):
        h = z
        for l in self.mapping:
            h = tf.nn.leaky_relu(l(h), 0.2)
        return h

    def synth(self, ws, noises=None):
        """ws: list of LEVELS style vectors; noises: optional list of LEVELS noise maps"""
        n = (lambda i, x: None) if noises is None else (lambda i, x: noises[i])
        x = tf.tile(self.const, [tf.shape(ws[0])[0], 1, 1, 1])
        x = self.c7(x);   x = tf.nn.leaky_relu(self.noise[0](x, n(0, x)), 0.2); x = self.adain[0]([x, ws[0]])
        x = self.t14(x);  x = tf.nn.leaky_relu(self.noise[1](x, n(1, x)), 0.2); x = self.adain[1]([x, ws[1]])
        x = self.t28(x);  x = tf.nn.leaky_relu(self.noise[2](x, n(2, x)), 0.2); x = self.adain[2]([x, ws[2]])
        return tf.tanh(self.out(x))

    def __call__(self, z, noises=None):
        w = self.w(z); return self.synth([w] * LEVELS, noises)

    def mixed(self, z1, z2, cutoff, noises=None):
        """levels < cutoff use z1, the others use z2"""
        w1, w2 = self.w(z1), self.w(z2); return self.synth([w1 if l < cutoff else w2 for l in range(LEVELS)], noises)

    def variables(self):
        return [self.const] + [v for l in self.mapping + [self.c7, self.t14, self.t28, self.out] + self.noise + self.adain for v in l.trainable_variables]


set_seed(11); fresh()
sg = ToyStyleGen(); _ = sg(tf.random.normal([2, 64]))
n_sg = sum(int(np.prod(v.shape)) for v in sg.variables())
print(f"toy style generator: {n_sg:,} trainable parameters · output {tuple(sg(tf.random.normal([2, 64])).shape)}")
audit("toy style generator: coding of 64 numbers → image 28×28×1 in [−1, 1]", tuple(sg(tf.random.normal([2, 64])).shape) == (2, 28, 28, 1))
z1_, z2_ = tf.random.normal([3, 64], seed=1), tf.random.normal([3, 64], seed=2)
fixed_noise = [tf.random.normal([3, s, s, 1], seed=5 + s) for s in (7, 14, 28)]
audit("mixing with cutoff 0 reproduces the second image exactly and cutoff 3 the first (same noise)", np.allclose(sg.mixed(z1_, z2_, 0, fixed_noise).numpy(), sg(z2_, fixed_noise).numpy(), atol=1e-6) and np.allclose(sg.mixed(z1_, z2_, 3, fixed_noise).numpy(), sg(z1_, fixed_noise).numpy(), atol=1e-6))
audit("at initialisation the learned noise scales are 0, so the noise cannot change the picture yet", np.allclose(sg(z1_).numpy(), sg(z1_).numpy(), atol=1e-6))
```

```python
def train_style(sg, X, steps, batch=64, seed=11, mix_prob=0.5, lr=2e-4):
    g_opt = keras.optimizers.Adam(lr, beta_1=0.5); d_opt = keras.optimizers.Adam(lr, beta_1=0.5); bce = keras.losses.BinaryCrossentropy()
    _, d = build_dcgan(); gv = sg.variables()

    @tf.function
    def gen(z1, z2, use_mix, cutoff):
        w1, w2 = sg.w(z1), sg.w(z2)
        ws = []
        for l in range(LEVELS):
            second = tf.logical_and(use_mix, l >= cutoff)[:, None]
            ws.append(tf.where(second, w2, w1))
        return sg.synth(ws)

    @tf.function
    def step(real):
        B = tf.shape(real)[0]
        z1, z2 = tf.random.normal([B, sg.zdim]), tf.random.normal([B, sg.zdim])
        use_mix = tf.random.uniform([B]) < mix_prob; cutoff = tf.random.uniform([B], 1, LEVELS, dtype=tf.int32)
        with tf.GradientTape() as tp:
            fake = gen(z1, z2, use_mix, cutoff); pr, pf = d(real, training=True), d(fake, training=True)
            dl = 0.5 * (bce(tf.ones_like(pr), pr) + bce(tf.zeros_like(pf), pf))
        d_opt.apply_gradients(zip(tp.gradient(dl, d.trainable_variables), d.trainable_variables))
        with tf.GradientTape() as tp:
            fake = gen(tf.random.normal([B, sg.zdim]), tf.random.normal([B, sg.zdim]), use_mix, cutoff); pf = d(fake, training=True); gl = bce(tf.ones_like(pf), pf)
        g_opt.apply_gradients(zip(tp.gradient(gl, gv), gv))
        return dl, gl

    ds = tf.data.Dataset.from_tensor_slices(X).shuffle(2000, seed=seed).repeat().batch(batch, drop_remainder=True).prefetch(1); hist = []; t0 = time.time()
    for i, xb in enumerate(ds.take(steps), 1):
        dl, gl = step(xb)
        if i % max(1, steps // 10) == 0:
            hist.append((i, float(dl), float(gl)))
    return hist, time.time() - t0


hist_sg, sec_sg = train_style(sg, X_dc, SZ["style_steps"])
print(f"{SZ['style_steps']} steps in {sec_sg / 60:.1f} min · last logged losses: D {hist_sg[-1][1]:.2f}, G {hist_sg[-1][2]:.2f}")
sc = [float(s_.numpy().mean()) for s_ in [n_.scale for n_ in sg.noise]]
print("mean learned noise scale per level (init 0):", [round(v, 4) for v in sc])
audit("training moved the noise scales away from their initial 0 (the network uses its noise inputs)", any(abs(v) > 1e-4 for v in sc), f"{[round(v, 4) for v in sc]}")
zs = tf.random.normal([16, 64], seed=9); smp = (np.squeeze(sg(zs).numpy()) + 1) / 2
show_images(smp, ncols=8, size=1.0, suptitle="toy style-based generator, 16 random codings")
```

### 8.13 Style mixing, and what noise does (Tasks 11–13)
Top row: six pictures from codings `B₁…B₆`; left column: six pictures from codings `A₁…A₆`. Each inner cell takes the **first level (7×7, coarse traits) from the row's A** and the **other levels (14×14 and 28×28) from the column's B**. The second grid moves the cutoff: the first *two* levels come from A. All pictures of one grid use the same noise maps, so only the styles differ. Numbers: the correlation of the mixed picture with pure A and with pure B – the more levels come from A, the more it should resemble A.

```python
def grid_mix(cutoff, za, zb, noises):
    rows = []
    for i in range(len(za)):
        rows.append(np.concatenate([(np.squeeze(sg.mixed(tf.repeat(za[i:i + 1], len(zb), 0), zb, cutoff, [tf.repeat(n_[:1], len(zb), 0) for n_ in noises]).numpy()) + 1) / 2], axis=0))
    return np.array(rows)          # (rows, cols, 28, 28)


za, zb = tf.random.normal([6, 64], seed=31), tf.random.normal([6, 64], seed=32)
nz1 = [tf.random.normal([1, s, s, 1], seed=40 + s) for s in (7, 14, 28)]
pureA = (np.squeeze(sg(za, [tf.repeat(n_, 6, 0) for n_ in nz1]).numpy()) + 1) / 2; pureB = (np.squeeze(sg(zb, [tf.repeat(n_, 6, 0) for n_ in nz1]).numpy()) + 1) / 2
fig, axes = plt.subplots(1, 2, figsize=(13, 6.2))
for ax, cutoff, ttl in zip(axes, (1, 2), ("first level (7×7) from A, the rest from B", "first two levels (7×7, 14×14) from A, the last from B")):
    G_ = grid_mix(cutoff, za, zb, nz1)
    full = np.ones((7 * 28, 7 * 28)); full[28:, 0:28] = np.concatenate(list(pureA), axis=0); full[0:28, 28:] = np.concatenate(list(pureB), axis=1)
    for i in range(6):
        for j in range(6):
            full[28 * (i + 1):28 * (i + 2), 28 * (j + 1):28 * (j + 2)] = G_[i, j]
    ax.imshow(full, cmap="binary", vmin=0, vmax=1); ax.set_title(f"cutoff {cutoff}: {ttl}", fontsize=8); ax.axis("off")
plt.tight_layout(); plt.show()
corr = lambda a, b: float(np.mean([np.corrcoef(x.ravel(), y.ravel())[0, 1] for x, y in zip(a, b)]))
rowsm = {}
big_nz = [tf.repeat(n_, 36, 0) for n_ in nz1]
zA36, zB36 = tf.repeat(za, 6, 0), tf.tile(zb, [6, 1]); A36, B36 = np.repeat(pureA, 6, 0), np.tile(pureB, (6, 1, 1))
for cutoff in range(LEVELS + 1):
    m = (np.squeeze(sg.mixed(zA36, zB36, cutoff, big_nz).numpy()) + 1) / 2
    rowsm[cutoff] = dict(levels_from_A=cutoff, corr_with_A=corr(m, A36), corr_with_B=corr(m, B36))
mt = pd.DataFrame(rowsm).T.set_index("levels_from_A").round(3); mt.index.name = "levels taken from A (rest from B)"; display(mt)
RESULTS["style_mixing"] = mt.reset_index().to_dict("list")
audit_trained("more levels from A → the picture resembles A more and B less (correlations rise/fall from 0 to 3 levels)", lambda: mt["corr_with_A"].iloc[-1] > mt["corr_with_A"].iloc[0] + 0.2 and mt["corr_with_B"].iloc[0] > mt["corr_with_B"].iloc[-1] + 0.2, f"corr with A {mt['corr_with_A'].iloc[0]:.2f} → {mt['corr_with_A'].iloc[-1]:.2f}")
print("Reading the table: 0 levels from A = pure B, 3 levels from A = pure A (both are exact copies, so those rows show the correlation of the same pictures with each other). In between, the share of A shows how much of the picture each level controls.")
# noise vs coding
wfix = sg.w(tf.random.normal([1, 64], seed=77)); rng_n = np.random.default_rng(0)
imgs_noise = np.stack([np.squeeze(sg.synth([tf.repeat(wfix, 1, 0)] * LEVELS, [tf.random.normal([1, s, s, 1], seed=int(rng_n.integers(1e6))) for s in (7, 14, 28)]).numpy()) for _ in range(24)])
imgs_z = np.squeeze(sg(tf.random.normal([24, 64], seed=78), [tf.repeat(n_, 24, 0) for n_ in nz1]).numpy())
sd_noise, sd_z = float(imgs_noise.std(0).mean()), float(imgs_z.std(0).mean())
print(f"pixel standard deviation over 24 pictures: changing ONLY the noise {sd_noise:.4f} · changing ONLY the coding {sd_z:.4f}  (ratio {sd_noise / max(sd_z, 1e-9):.2f})")
show_images(np.concatenate([(imgs_noise[:8] + 1) / 2, (imgs_z[:8] + 1) / 2]), ncols=8, size=1.0, suptitle="top: same coding, different noise · bottom: different codings, same noise")
RESULTS["style_noise"] = dict(sd_noise=sd_noise, sd_z=sd_z)
audit_trained("noise only changes small details: the picture varies much less with the noise than with the coding (ratio < 0.5)", lambda: sd_noise < 0.5 * sd_z, f"{sd_noise:.4f} vs {sd_z:.4f}")
sp = fashion_clf.predict((np.squeeze(sg(tf.random.normal([2000, 64], seed=5)).numpy()) + 1) / 2, batch_size=500, verbose=0)
RESULTS["style_gan"] = dict(steps=SZ["style_steps"], conf=float(sp.max(1).mean()), classes=int((np.bincount(sp.argmax(1), minlength=10) / 2000 > 0.02).sum()))
print(f"toy style generator, 2 000 random pictures: classifier confidence {sp.max(1).mean():.2f}, {RESULTS['style_gan']['classes']} classes with more than 2 % share")
```

**Note on the failed proof above (kept on purpose).** The toy style generator, after 1 500 steps, produces pictures that are very similar to each other (any two pictures built from different codings already correlate at 0.95, and the helper classifier finds only 2 classes with more than 2 % share), and its learned noise scales stayed near 0.001. With so little variety between pictures, mixing styles cannot produce a visible trend: the correlation with A goes from 0.946 (nothing from A) to 0.947 and 0.950 (one and two levels from A) and only jumps to 1.0 when everything comes from A. The mechanism is tested and exact (cutoff 0 and 3 reproduce the two pure pictures, AdaIN and noise layers pass their unit tests), but the locality effect of mixing regularization is **not** demonstrated by this run; a much longer-trained model would be needed. The +0.2 threshold in the check was unreachable for a nearly collapsed toy.

## Capstone A · Autoencoder suite on Fashion MNIST (Task 14)
Seven autoencoders on the same data, the same optimizer (Adam 1e-3), the same batch (64) and the same number of epochs (`cap_epochs`, on `cap_train` training images; the validation images are never used for training), **3 seeds each**:

| | model | coding | notes |
|---|---|---|---|
| (a) | **linear AE** (≙ PCA) | 30 | `Dense(30)` / `Dense(784)`, no activations, MSE loss; PCA(30) is added as the analytic reference |
| (b) | **stacked** | 30 | 784-100-30-100-784, SELU / sigmoid, BCE |
| (c) | **stacked + tied** | 30 | the decoder uses the transposed encoder kernels |
| (d) | **convolutional** | 30 | the three conv/pool blocks of Phase 4, then `Flatten → Dense(30)` (the bare conv bottleneck has 576 numbers) |
| (e) | **denoising** | 30 | dropout 0.3 on the input during training only |
| (f) | **sparse (KL)** | **300** | 300 sigmoid units, target 0.1, weight 0.05 – an over-complete coding, so "30" is not possible |
| (g) | **VAE** | 30 | the coding is the mean μ; reconstruction = decoder(μ) |

Metrics: **held-out pixel-MSE** (mean over the validation pixels, from clean inputs), **parameters**, **accuracy of a linear classifier on the codings** (logistic regression, standardised codings, trained on 5 000 training images, scored on the validation images), and a **t-SNE plot** of each coding (seed 11 model). Seeds: mean ± sample standard deviation; differences that fit inside that spread are not called findings.

```python
Xcap = X_train[:SZ["cap_train"]]; ycap = y_train[:SZ["cap_train"]]
yv_ = y_valid


def build_linear():
    enc = keras.Sequential([keras.Input(shape=(28, 28)), L.Flatten(), L.Dense(30)], name="lin_encoder")
    dec = keras.Sequential([keras.Input(shape=(30,)), L.Dense(784), L.Reshape([28, 28])], name="lin_decoder")
    return enc, dec, keras.Sequential([enc, dec], name="linear")


def build_conv30():
    enc = keras.Sequential([keras.Input(shape=(28, 28)), L.Reshape([28, 28, 1]),
                            L.Conv2D(16, 3, padding="same", activation="selu"), L.MaxPool2D(2), L.Conv2D(32, 3, padding="same", activation="selu"), L.MaxPool2D(2),
                            L.Conv2D(64, 3, padding="same", activation="selu"), L.MaxPool2D(2), L.Flatten(), L.Dense(30, activation="selu")], name="conv30_encoder")
    dec = keras.Sequential([keras.Input(shape=(30,)), L.Dense(576, activation="selu"), L.Reshape([3, 3, 64]),
                            L.Conv2DTranspose(32, 3, strides=2, padding="valid", activation="selu"), L.Conv2DTranspose(16, 3, strides=2, padding="same", activation="selu"),
                            L.Conv2DTranspose(1, 3, strides=2, padding="same", activation="sigmoid"), L.Reshape([28, 28])], name="conv30_decoder")
    return enc, dec, keras.Sequential([enc, dec], name="conv30")


def build_dae30():
    layers_in = [keras.Input(shape=(28, 28)), L.Flatten(), L.Dropout(0.3)]
    enc = keras.Sequential(layers_in + [L.Dense(100, activation="selu"), L.Dense(30, activation="selu")], name="dae30_encoder")
    dec = keras.Sequential([keras.Input(shape=(30,)), L.Dense(100, activation="selu"), L.Dense(784, activation="sigmoid"), L.Reshape([28, 28])], name="dae30_decoder")
    return enc, dec, keras.Sequential([enc, dec], name="dae30")


CAP_MODELS = [("(a) linear AE", build_linear, "mse", dict()), ("(b) stacked", build_stacked, "binary_crossentropy", dict()), ("(c) stacked + tied", build_tied, "binary_crossentropy", dict()),
              ("(d) convolutional", build_conv30, "binary_crossentropy", dict()), ("(e) denoising", build_dae30, "binary_crossentropy", dict()),
              ("(f) sparse (KL)", build_sparse, "binary_crossentropy", dict(kind="kl", weight=0.05)), ("(g) VAE", None, None, None)]


def lin_acc2(A_tr, y_tr, A_te, y_te):
    mu, sd = A_tr.mean(0), A_tr.std(0) + 1e-6
    return float(LogisticRegression(max_iter=300).fit((A_tr - mu) / sd, y_tr).score((A_te - mu) / sd, y_te))


n_lin = min(len(Xcap), 5000)
cap_rows, cap_codes = [], {}
for s in RUN_SEEDS:
    for name, builder, loss, kw in CAP_MODELS:
        if builder is not None:
            enc, dec, ae, h, sec = train_ae(builder, loss=loss, optimizer=keras.optimizers.Adam(learning_rate=1e-3), seed=s, epochs=SZ["cap_epochs"], batch_size=64, X=Xcap, Xv=X_valid, **kw)
            code_tr, code_te = enc.predict(Xcap[:n_lin], batch_size=1000, verbose=0), enc.predict(X_valid, batch_size=1000, verbose=0); rec = ae.predict(X_valid, batch_size=1000, verbose=0); n_par = ae.count_params()
        else:
            ve, de, v, h, mon_, sec = train_vae(s, SZ["cap_epochs"], codings_size=30, X=Xcap)
            code_tr, code_te = ve.predict(Xcap[:n_lin], batch_size=1000, verbose=0)[0], ve.predict(X_valid, batch_size=1000, verbose=0)[0]; rec = de.predict(code_te, batch_size=1000, verbose=0); n_par = ve.count_params() + de.count_params()
        cap_rows.append(dict(model=name, seed=s, mse=float(np.mean((rec - X_valid) ** 2)), acc=lin_acc2(code_tr, ycap[:n_lin], code_te, yv_), params=n_par, coding=code_te.shape[1], sec=sec))
        if s == RUN_SEEDS[0]:
            cap_codes[name] = code_te
    print(f"seed {s} done")
cap = pd.DataFrame(cap_rows)
pca_ref = PCA(n_components=30).fit(Xcap.reshape(len(Xcap), -1))
pca_mse = float(np.mean((pca_ref.inverse_transform(pca_ref.transform(X_valid.reshape(len(X_valid), -1))) - X_valid.reshape(len(X_valid), -1)) ** 2))
pca_acc = lin_acc2(pca_ref.transform(Xcap[:n_lin].reshape(n_lin, -1)), ycap[:n_lin], pca_ref.transform(X_valid.reshape(len(X_valid), -1)), yv_)
raw_acc = lin_acc2(Xcap[:n_lin].reshape(n_lin, -1), ycap[:n_lin], X_valid.reshape(len(X_valid), -1), yv_)
tab = cap.groupby("model", sort=False).agg(coding=("coding", "first"), parameters=("params", "first"), val_pixel_MSE=("mse", lambda v: pm(v, 5)), linear_clf_accuracy=("acc", lambda v: pm(v, 3)), sec_per_epoch=("sec", lambda v: round(float(np.mean(v)), 1)))
tab.loc["reference: PCA(30)"] = [30, 784 * 30 * 2 + 784 + 30, f"{pca_mse:.5f} (analytic)", f"{pca_acc:.3f}", 0.0]
tab.loc["reference: raw 784 pixels"] = [784, 0, "0 (copy)", f"{raw_acc:.3f}", 0.0]
display(tab)
RESULTS["capstone_a"] = dict(table=tab.reset_index().astype(str).to_dict("list"), pca_mse=pca_mse, pca_acc=pca_acc, raw_acc=raw_acc, mse={k: g["mse"].tolist() for k, g in cap.groupby("model", sort=False)}, acc={k: g["acc"].tolist() for k, g in cap.groupby("model", sort=False)})
audit("the linear AE and PCA(30) reconstruct about equally well (Phase 1's result at coding size 30: within 25 %)", abs(cap[cap.model == "(a) linear AE"].mse.mean() - pca_mse) / pca_mse < 0.25 or QUICK, f"{cap[cap.model == '(a) linear AE'].mse.mean():.5f} vs PCA {pca_mse:.5f}")
audit("the tied autoencoder has fewer parameters than the untied stacked one", int(cap[cap.model == "(c) stacked + tied"].params.iloc[0]) < int(cap[cap.model == "(b) stacked"].params.iloc[0]))
audit("every model's codings carry class information: linear-classifier accuracy above chance (0.10) by at least 0.2", bool((cap.groupby("model").acc.mean() > 0.30).all()) or QUICK, f"lowest {cap.groupby('model').acc.mean().min():.3f}")
fig, ax = plt.subplots(1, 2, figsize=(11, 3.0))
order_ = list(tab.index[:7]); mse_m = [cap[cap.model == m].mse.mean() for m in order_]; mse_s = [cap[cap.model == m].mse.std(ddof=1) if len(RUN_SEEDS) > 1 else 0 for m in order_]
acc_m = [cap[cap.model == m].acc.mean() for m in order_]; acc_s = [cap[cap.model == m].acc.std(ddof=1) if len(RUN_SEEDS) > 1 else 0 for m in order_]
ax[0].bar(range(7), mse_m, yerr=mse_s, color="C0", capsize=3); ax[0].axhline(pca_mse, ls=":", color="k"); ax[0].set_xticks(range(7)); ax[0].set_xticklabels([m[:3] for m in order_]); ax[0].set_title("held-out pixel-MSE (mean ± std over seeds; dotted = PCA 30)", fontsize=8)
ax[1].bar(range(7), acc_m, yerr=acc_s, color="C1", capsize=3); ax[1].axhline(raw_acc, ls=":", color="k"); ax[1].set_xticks(range(7)); ax[1].set_xticklabels([m[:3] for m in order_]); ax[1].set_ylim(0.5, 0.9); ax[1].set_title("linear classifier on the coding (dotted = raw pixels)", fontsize=8)
plt.tight_layout(); plt.show()
```

## Capstone B · Pretraining with a denoising autoencoder (Task 15, Exercise 9)
MNIST (60 000 training images, 10 000 test images). A **deep denoising autoencoder** `784 → 256 → 128 → 64 → 128 → 256 → 784` (SELU, sigmoid output, BCE) is trained on **all 60 000 training images without labels**; its input is corrupted by Gaussian noise (σ = 0.3) *inside the autoencoder only* – the encoder that is reused later contains no noise layer, so it behaves identically in training and in the classifier. Then: (1) check the reconstructions, (2) visualise the images that most activate coding neurons, (3) build `encoder + Dense(10, softmax)` and train it on **only 500 labelled images** (50 per class) – randomly initialised, pretrained and frozen, pretrained and fine-tuned – with **3 seeds**, scored on the 10 000 test images.

```python
def build_deep_dae():
    enc = keras.Sequential([keras.Input(shape=(28, 28)), L.Flatten(), L.Dense(256, activation="selu"), L.Dense(128, activation="selu"), L.Dense(64, activation="selu")], name="capb_encoder")
    dec = keras.Sequential([keras.Input(shape=(64,)), L.Dense(128, activation="selu"), L.Dense(256, activation="selu"), L.Dense(784, activation="sigmoid"), L.Reshape([28, 28])], name="capb_decoder")
    ae = keras.Sequential([keras.Input(shape=(28, 28)), L.GaussianNoise(0.3), enc, dec], name="capb_dae")
    return enc, dec, ae


capb_rows, capb_first = [], None
for s in RUN_SEEDS:
    set_seed(s); fresh()
    enc, dec, ae = build_deep_dae()
    ae.compile(loss="binary_crossentropy", optimizer=keras.optimizers.Adam(learning_rate=1e-3))
    h_, sec_ = fit_quiet(ae, M_train_all, M_train_all, epochs=SZ["pre_ae_epochs"], batch_size=128)
    if capb_first is None:
        capb_first = (enc, dec, ae, h_)
    idx = balanced_subset(m_y_all, SZ["pre_labels"] // 10, np.random.default_rng(s)); Xl, yl = M_train_all[idx], m_y_all[idx]
    for name, (pretrained, trainable) in {"(a) no pretraining": (False, True), "(b) pretrained, frozen": (True, False), "(c) pretrained, fine-tuned": (True, True)}.items():
        set_seed(s)
        e = clone_encoder(enc) if pretrained else keras.models.clone_model(enc)
        clf = make_classifier(e, trainable); before = [w.copy() for w in e.get_weights()]
        fit_quiet(clf, Xl, yl, epochs=SZ["pre_clf_epochs"], batch_size=32)
        capb_rows.append(dict(seed=s, config=name, acc=float(clf.evaluate(M_test, m_y_test, verbose=0)[1]), enc_changed=max(float(np.abs(a - b).max()) for a, b in zip(before, e.get_weights()))))
    print(f"seed {s}: pretraining {sec_ * SZ['pre_ae_epochs'] / 60:.1f} min, loss {h_['loss'][0]:.4f} → {h_['loss'][-1]:.4f}")
capb = pd.DataFrame(capb_rows)
enc_b, dec_b, ae_b, h_b = capb_first
```

```python
rng = np.random.default_rng(0); clean = M_test[:8]; noisy = np.clip(clean + rng.normal(0, 0.3, clean.shape), 0, 1).astype("float32")
rec_n = dec_b.predict(enc_b.predict(noisy, verbose=0), verbose=0)
fig, axes = plt.subplots(3, 8, figsize=(9.2, 3.8))
for r, (t, arr) in enumerate((("clean test image", clean), ("noisy input (σ = 0.3, clipped)", noisy), ("reconstruction from the noisy input", rec_n))):
    for c in range(8):
        axes[r, c].imshow(arr[c], cmap="binary", vmin=0, vmax=1); axes[r, c].axis("off")
    axes[r, 0].set_title(t, fontsize=7, loc="left")
plt.tight_layout(); plt.show()
e_in, e_out = float(np.mean((noisy - clean) ** 2)), float(np.mean((rec_n - clean) ** 2))
print(f"on these 8 test images: noisy input vs clean {e_in:.4f} · reconstruction vs clean {e_out:.4f}")
audit_trained("the denoising autoencoder removes noise: the reconstruction is closer to the clean image than the noisy input is", lambda: e_out < e_in, f"{e_out:.4f} vs {e_in:.4f}")
```

## Capstone C · VAE, DCGAN, DCGAN + experience replay, conditional DCGAN (Task 16, Exercises 10 and 11)
Everything below uses the models trained earlier in the notebook (VAE of Phase 6; DCGAN of 8.1; the short DCGAN runs of 8.2 and 8.4 with and without experience replay; conditional DCGAN of 8.5). Metrics are computed on 2 000 generated images judged by the same helper classifier (class coverage, class entropy, confidence), the feature-space Fréchet distance of Section 8.8, and a **sharpness** score – the variance of the Laplacian of the image, averaged over images (blurry images score low; real images are the reference).

```python
def sharp(imgs):
    x = np.asarray(imgs).reshape(-1, 28, 28); lap = 4 * x[:, 1:-1, 1:-1] - x[:, :-2, 1:-1] - x[:, 2:, 1:-1] - x[:, 1:-1, :-2] - x[:, 1:-1, 2:]
    return float(lap.reshape(len(x), -1).var(axis=1).mean())


def samples_of(name, n=2000, seed=3):
    zz = tf.random.normal([n, 10 if name == "vae" else DC_CODINGS], seed=seed)
    if name == "vae":
        return vdec(zz).numpy()
    if name == "cgan":
        return gan_eval_tanh(g_c, n=n, seed=seed, labels=np.random.default_rng(seed).integers(0, 10, n))["fakes"]
    gen = {"dcgan": g_dc, "dc_short": sens[0]["gen"], "dc_replay": rep[0]["gen"]}[name]
    return (np.squeeze(gen(zz, training=False).numpy()) + 1) / 2


names_c = {"vae": f"VAE ({SZ['vae_epochs']} epochs)", "dcgan": f"DCGAN ({SZ['dc_steps']} steps)", "dc_short": f"DCGAN short ({SZ['dc_var_steps']} steps)", "dc_replay": f"DCGAN short + replay ({SZ['dc_var_steps']} steps)", "cgan": f"conditional DCGAN ({SZ['cgan_steps']} steps)"}
secs_c = {"vae": sec_v * SZ["vae_epochs"], "dcgan": sec_dc, "dc_short": sens[0]["sec"], "dc_replay": rep[0]["sec"], "cgan": sec_c}
rows_c, imgs_c = [], {}
fr_ref = feats(X_test[:2000])
for k, nm in names_c.items():
    im = samples_of(k); imgs_c[k] = im
    p = fashion_clf.predict(im, batch_size=500, verbose=0); share = np.bincount(p.argmax(1), minlength=10) / len(im)
    rows_c.append({"model": nm, "classes covered": int((share > 0.02).sum()), "class entropy (max 2.30)": float(-(share[share > 0] * np.log(share[share > 0])).sum()), "classifier confidence": float(p.max(1).mean()),
                   "Fréchet (feature space)": frechet(fr_ref, feats(im)), "sharpness": sharp(im), "training minutes": secs_c[k] / 60})
rows_c.append({"model": "real test images", "classes covered": 10, "class entropy (max 2.30)": 2.30, "classifier confidence": float(fashion_clf.predict(X_test[:2000], verbose=0).max(1).mean()), "Fréchet (feature space)": 0.0, "sharpness": sharp(X_test[:2000]), "training minutes": 0.0})
ct = pd.DataFrame(rows_c).set_index("model"); display(ct.round(3))
RESULTS["capstone_c"] = ct.reset_index().to_dict("list")
fig, axes = plt.subplots(6, 1, figsize=(11, 8.2))
for ax, (lab, im) in zip(axes, [("real", X_test[:14])] + [(names_c[k], imgs_c[k][:14]) for k in names_c]):
    ax.imshow(np.concatenate(list(np.squeeze(im)), axis=1), cmap="binary", vmin=0, vmax=1); ax.set_xticks([]); ax.set_yticks([]); ax.set_ylabel(lab, rotation=0, ha="right", va="center", fontsize=7)
plt.suptitle("side by side: real images, VAE, DCGAN, short DCGAN, short DCGAN + replay, conditional DCGAN (random classes)", fontsize=9); plt.tight_layout(); plt.show()
audit_trained("the VAE's samples are blurrier than real images (lower Laplacian variance), as expected for a decoder that outputs averages", lambda: ct.loc[names_c["vae"], "sharpness"] < ct.loc["real test images", "sharpness"], f"{ct.loc[names_c['vae'], 'sharpness']:.4f} vs {ct.loc['real test images', 'sharpness']:.4f}")
```

**Semantic interpolation in the DCGAN's coding space** (the same test as for the VAE in Phase 6): 40 pairs of codings; the straight line between the two codings is decoded and compared with a plain pixel blend of the two end images. A path through the generator's space should give images that still look like clothing at every step; a pixel blend gives two ghost images laid over each other.

```python
rng = np.random.default_rng(4); zs0, zs1 = tf.random.normal([40, DC_CODINGS], seed=61), tf.random.normal([40, DC_CODINGS], seed=62)
ts = np.linspace(0, 1, 10).astype("float32")
path = np.stack([(np.squeeze(g_dc(((1 - t) * zs0 + t * zs1), training=False).numpy()) + 1) / 2 for t in ts], axis=1)            # (40, 10, 28, 28)
blend = np.stack([(1 - t) * path[:, 0] + t * path[:, -1] for t in ts], axis=1)
conf_s = fashion_clf.predict(path[:, 1:-1].reshape(-1, 28, 28), batch_size=500, verbose=0).max(1).mean(); conf_p = fashion_clf.predict(blend[:, 1:-1].reshape(-1, 28, 28), batch_size=500, verbose=0).max(1).mean()
print(f"classifier confidence on the 8 inner steps: coding-space path {conf_s:.3f} · pixel blend {conf_p:.3f}")
show_images(np.concatenate([path[0], blend[0], path[1], blend[1]]), ncols=10, size=1.0, suptitle="rows 1 and 3: DCGAN coding-space interpolation · rows 2 and 4: pixel blending of the same end points")
RESULTS["dcgan_interp"] = dict(semantic=float(conf_s), pixel=float(conf_p))
audit_trained("a path through the DCGAN's coding space gives images that look more like clothing than a pixel blend (higher classifier confidence)", lambda: conf_s > conf_p, f"{conf_s:.3f} vs {conf_p:.3f}")
```

## The one-page guide · which unsupervised / generative model when (Task 17)
In my own words, with the evidence from this notebook in brackets.

**Dimensionality reduction – start with PCA, move to a (stacked, tied or convolutional) autoencoder when the data is curved.** A linear autoencoder spans the same subspace as PCA (Phase 1, Exercise 12), so it adds nothing; non-linear layers help when the data lies on a curved surface or when images have local structure (Capstone A).

**Visualisation – a t-SNE (or UMAP) of the autoencoder's coding.** Compress with an autoencoder to ~30 numbers first (it removes noise and makes t-SNE faster), then embed in 2-D; remember that the plot's clusters are not distances (Phase 2, Capstone A).

**Pretraining with few labels – (denoising) autoencoder as the encoder, then fine-tune.** Useful when unlabeled data is plentiful and labels are scarce, exactly the 500-label case of Phase 3 and Capstone B. Freeze the encoder when there are very few labels, fine-tune when there are more; always compare with a randomly initialised network over several seeds, because the gain can be small.

**Denoising / repair – denoising autoencoder.** Train with the corruption you expect (Gaussian noise, dropped pixels); it learns what is plausible, but its output is a plausible *guess*, not the missing data (Phase 5).

**Anomaly-style tasks – reconstruction error.** Train on normal data only; items that reconstruct badly are unusual. Works best when the anomaly differs in structure from the normal class and the model is not powerful enough to reconstruct everything (Exercise 13).

**Diverse but blurry generation – VAE.** Stable single-loss training, covers all the data, smooth latent space to interpolate in and to measure a likelihood bound on; the price is blurry samples (Phase 6, Capstone C).

**Sharp generation – GAN (DCGAN for images).** Sharper pictures because the discriminator punishes blur, but training is touchy: seed, learning rates and normalisation matter, and mode collapse has to be watched with class histograms and sample grids (Phases 7–8). Tricks that sometimes help: experience replay, minibatch diversity features.

**Controllable generation – conditional GAN (or a conditional VAE).** Feed the class (or another label) to both networks to choose what is generated (8.5). For traits that are not labelled, style-based generators separate coarse and fine styles and allow mixing (8.9–8.13).

**High-resolution synthesis – progressive growing and style-based GANs.** Grow the networks resolution by resolution with a fade-in, stabilise with minibatch standard deviation, equalized learning rate and pixelwise normalization, and let a mapping network plus per-level styles and noise do the rest (8.6–8.13). Out of reach on a CPU at full size; the toy versions here show the mechanisms.

**Rule of thumb.** Need a *representation*: autoencoder family. Need *samples that cover everything*: VAE. Need *the sharpest samples* and can afford tuning: GAN. Need *a say in what comes out*: conditional model. Need *size*: progressive / style-based.

### Exercise 12 · a linear autoencoder spans the PCA subspace; a non-linear one beats it on curved data
**(a) Same subspace, a different data set.** The 8×8 digits (64 numbers per image, 1 797 images), coding size 5. A linear autoencoder (`Dense(5)` → `Dense(64)`, MSE, Adam) is compared with PCA(5): the **principal angles** between the column space of the decoder's weight matrix and the span of the top-5 principal components (0° = identical subspace), and the reconstruction errors. (Phase 1 did the same on a 3-D plane; the individual axes of the AE are *not* the principal components – only the subspace is the same.)

```python
dg = load_digits(); Xd = (dg.data / 16.0).astype("float32"); Xd = Xd - Xd.mean(0)
from scipy.linalg import subspace_angles
rows12 = []
for s in RUN_SEEDS:
    set_seed(s); fresh()
    lin = keras.Sequential([keras.Input(shape=(64,)), L.Dense(5, use_bias=False), L.Dense(64, use_bias=False)]); lin.compile(loss="mse", optimizer=keras.optimizers.Adam(learning_rate=5e-3))
    fit_quiet(lin, Xd, Xd, epochs=pick(600, 20), batch_size=256)
    Wd = lin.layers[-1].get_weights()[0].T                # (64, 5): columns span the decoder's image space
    pc = PCA(n_components=5).fit(Xd); ang = np.degrees(subspace_angles(Wd, pc.components_.T))
    mse_lin = float(np.mean((lin.predict(Xd, verbose=0) - Xd) ** 2)); mse_pca = float(np.mean((pc.inverse_transform(pc.transform(Xd)) - Xd) ** 2))
    rows12.append(dict(seed=s, max_angle_deg=float(ang.max()), mse_linear_ae=mse_lin, mse_pca=mse_pca))
t12a = pd.DataFrame(rows12).set_index("seed"); display(t12a.round(5))
audit_trained("linear AE ≙ PCA on the digits: the largest principal angle between the decoder's subspace and PCA's is below 10° and the errors differ by less than 5 %", lambda: t12a.max_angle_deg.max() < 10 and (abs(t12a.mse_linear_ae - t12a.mse_pca) / t12a.mse_pca).max() < 0.05, f"max angle {t12a.max_angle_deg.max():.1f}°, error ratio {(t12a.mse_linear_ae / t12a.mse_pca).max():.3f}")
RESULTS["ex12a"] = t12a.reset_index().to_dict("list")
```

**Note on the failed proof above (kept on purpose).** The linear autoencoder gets within 1.7–4.3 % of PCA's error, but its decoder subspace differs from PCA's in at least one direction (largest principal angle 22°, 87° and 22° in the three seeds), so the check "below 10° and within 5 %" failed. Training longer does not change it (next cell: 3 000 epochs give the same numbers as 600). The 5th and 6th eigenvalues of the digits are close (ratio 1.18), and a subspace that swaps one of the last components for a neighbouring one costs very little error – so *the subspace is nearly, not exactly, PCA's*, unlike Phase 1 where the eigenvalues were well separated and the angle was below 2°. I did not find out whether this is an optimizer effect (Adam with minibatches stopping near a critical point) or a property of this data set; the supplementary cell only shows that it is not a matter of training length.

```python
# Supplementary check for the failed proof above (self-contained: needs only the bundled digits data)
import numpy as np, tensorflow as tf
from tensorflow import keras
from sklearn.datasets import load_digits
from sklearn.decomposition import PCA
from scipy.linalg import subspace_angles
Xd = (load_digits().data / 16.0).astype("float32"); Xd = Xd - Xd.mean(0)
pc8 = PCA(n_components=8).fit(Xd); ev = pc8.explained_variance_
print("PCA eigenvalues 1-8:", np.round(ev, 3), "| ratio of the 5th to the 6th:", round(float(ev[4] / ev[5]), 3))
pc5, pc6 = PCA(n_components=5).fit(Xd), PCA(n_components=6).fit(Xd)
mse_pca = float(np.mean((pc5.inverse_transform(pc5.transform(Xd)) - Xd) ** 2))
for epochs in (600, 3000):
    for seed in (11, 22, 33):
        keras.utils.set_random_seed(seed)
        lin = keras.Sequential([keras.Input(shape=(64,)), keras.layers.Dense(5, use_bias=False), keras.layers.Dense(64, use_bias=False)]); lin.compile(loss="mse", optimizer=keras.optimizers.Adam(learning_rate=5e-3))
        lin.fit(Xd, Xd, epochs=epochs, batch_size=256, verbose=0)
        Wd = lin.layers[-1].get_weights()[0].T
        a5 = np.degrees(subspace_angles(Wd, pc5.components_.T)); a6 = np.degrees(subspace_angles(Wd, pc6.components_.T))
        mse = float(np.mean((lin.predict(Xd, verbose=0) - Xd) ** 2))
        print(f"{epochs:5d} epochs, seed {seed}: error / PCA error {mse / mse_pca:.3f} | largest angle to PCA-5 span {a5.max():5.1f} deg | largest angle to the span of the top 6 components {a6.max():5.1f} deg")
```

```python
Xs, tcol = make_swiss_roll(n_samples=pick(4000, 800), noise=0.05, random_state=0); Xs = ((Xs - Xs.mean(0)) / Xs.std()).astype("float32")
ntr_ = int(0.75 * len(Xs)); Xs_tr, Xs_te, t_tr, t_te = Xs[:ntr_], Xs[ntr_:], tcol[:ntr_], tcol[ntr_:]
rows12b, keep = [], {}
for s in RUN_SEEDS:
    for kind in ("linear", "non-linear"):
        set_seed(s); fresh()
        if kind == "linear":
            m = keras.Sequential([keras.Input(shape=(3,)), L.Dense(2), L.Dense(3)])
        else:
            m = keras.Sequential([keras.Input(shape=(3,)), L.Dense(64, activation="elu"), L.Dense(64, activation="elu"), L.Dense(2), L.Dense(64, activation="elu"), L.Dense(64, activation="elu"), L.Dense(3)])
        m.compile(loss="mse", optimizer=keras.optimizers.Adam(learning_rate=3e-3)); fit_quiet(m, Xs_tr, Xs_tr, epochs=pick(300, 15), batch_size=128)
        rows12b.append(dict(model=kind, seed=s, test_mse=float(np.mean((m.predict(Xs_te, verbose=0) - Xs_te) ** 2))))
        if s == RUN_SEEDS[0]:
            keep[kind] = m
t12b = pd.DataFrame(rows12b); res12 = t12b.groupby("model", sort=False).test_mse.apply(lambda v: pm(v, 5)).to_frame("held-out MSE (mean ± std over seeds)"); display(res12)
pca2 = PCA(n_components=2).fit(Xs_tr); mse_pca2 = float(np.mean((pca2.inverse_transform(pca2.transform(Xs_te)) - Xs_te) ** 2)); print(f"PCA with 2 components (analytic): {mse_pca2:.5f}")
ml, mn = t12b[t12b.model == "linear"].test_mse.values, t12b[t12b.model == "non-linear"].test_mse.values
vrd, gap, _ = noise_verdict(ml, mn); print(f"non-linear minus linear: {gap:+.5f} – {vrd}")
audit_trained("a linear AE with a 2-D coding is as good as PCA(2) on the Swiss roll (within 10 %)", lambda: abs(ml.mean() - mse_pca2) / mse_pca2 < 0.10, f"{ml.mean():.5f} vs {mse_pca2:.5f}")
audit_trained("the non-linear AE reconstructs the Swiss roll with less than half of the linear AE's error", lambda: mn.mean() < 0.5 * ml.mean(), f"{mn.mean():.5f} vs {ml.mean():.5f}")
RESULTS["ex12b"] = dict(linear=list(ml), nonlinear=list(mn), pca2=mse_pca2)
fig = plt.figure(figsize=(14, 3.6))
panels = [("original (colour = position along the roll)", Xs_te), ("linear AE / PCA reconstruction", keep["linear"].predict(Xs_te, verbose=0)), ("non-linear AE reconstruction", keep["non-linear"].predict(Xs_te, verbose=0))]
for i, (ttl, P) in enumerate(panels):
    ax = fig.add_subplot(1, 4, i + 1, projection="3d"); ax.scatter(P[:, 0], P[:, 1], P[:, 2], c=t_te, cmap="viridis", s=4); ax.set_title(ttl, fontsize=8); ax.view_init(elev=12, azim=-70); ax.set_xticks([]); ax.set_yticks([]); ax.set_zticks([])
codes_nl = keras.Sequential(keep["non-linear"].layers[:3]).predict(Xs_te, verbose=0) if len(keep["non-linear"].layers) >= 3 else None
ax = fig.add_subplot(1, 4, 4); ax.scatter(codes_nl[:, 0], codes_nl[:, 1], c=t_te, cmap="viridis", s=4); ax.set_title("non-linear AE: the 2-D coding", fontsize=8); ax.set_xticks([]); ax.set_yticks([])
plt.tight_layout(); plt.show()
print("Reading: a flat plane cannot follow the spiral, so the linear reconstruction folds all turns onto one sheet; the non-linear AE follows the roll, and its 2-D coding is an (approximately) unrolled sheet whose colour changes smoothly.")
EX[12] = ("done", "Exercise 12 section: digits subspace angles (a) and the Swiss roll (b); Phase 1 for the 3-D plane")
```

### Exercise 13 · reconstruction error as an anomaly score
Train **only on one class** of Fashion MNIST (the "normal" class, training images of that class), then score the 10 000 test images by their reconstruction error; the 1 000 test images of the normal class should score low, the 9 000 others high. ROC-AUC measures how well the score separates the two groups (0.5 = no better than chance, 1 = perfect). Models: the stacked AE (784-100-30), the VAE (10-D coding; score = reconstruction error of the decoded mean), and a PCA(30) fitted on the normal class as a simple reference. Normal classes: Trouser (a very distinct shape), Sneaker, Shirt (similar to many other classes); 3 seeds.

```python
def anomaly_run(cls, seed):
    Xn = X_all_fashion[y_all_fashion == cls][: pick(5000, 800)]
    enc, dec, ae, h, sec = train_ae(build_stacked, loss="binary_crossentropy", optimizer=keras.optimizers.Adam(learning_rate=1e-3), seed=seed, epochs=SZ["anomaly_epochs"], batch_size=64, X=Xn, Xv=X_valid)
    s_ae = np.mean((ae.predict(X_test, batch_size=1000, verbose=0) - X_test) ** 2, axis=(1, 2))
    ve, de, v, h2, mon_, sec2 = train_vae(seed, SZ["anomaly_epochs"], codings_size=10, X=Xn)
    mu = ve.predict(X_test, batch_size=1000, verbose=0)[0]; s_vae = np.mean((de.predict(mu, batch_size=1000, verbose=0) - X_test) ** 2, axis=(1, 2))
    pca = PCA(n_components=30).fit(Xn.reshape(len(Xn), -1)); Xf = X_test.reshape(len(X_test), -1); s_pca = np.mean((pca.inverse_transform(pca.transform(Xf)) - Xf) ** 2, axis=1)
    is_anom = (y_test != cls).astype(int)
    return {"AE": s_ae, "VAE": s_vae, "PCA(30)": s_pca}, {k: float(roc_auc_score(is_anom, v)) for k, v in {"AE": s_ae, "VAE": s_vae, "PCA(30)": s_pca}.items()}


rows13, scores13 = [], {}
for cls in (1, 7, 6):
    for s in RUN_SEEDS:
        sc, auc = anomaly_run(cls, s)
        rows13.append(dict(normal_class=FASHION_CLASSES[cls], seed=s, **auc))
        if s == RUN_SEEDS[0]:
            scores13[cls] = sc
t13 = pd.DataFrame(rows13)
agg13 = t13.groupby("normal_class", sort=False)[["AE", "VAE", "PCA(30)"]].agg(lambda v: pm(v, 3)); display(agg13)
RESULTS["ex13"] = {c: {m: list(g[m]) for m in ("AE", "VAE", "PCA(30)")} for c, g in t13.groupby("normal_class", sort=False)}
fig, axes = plt.subplots(1, 3, figsize=(13.5, 2.9))
for ax, cls in zip(axes, (1, 7, 6)):
    sc = scores13[cls]["AE"]; hi = np.percentile(sc, 99)
    ax.hist(sc[y_test == cls], bins=40, range=(0, hi), alpha=0.7, label=f"normal ({FASHION_CLASSES[cls]})", density=True); ax.hist(sc[y_test != cls], bins=40, range=(0, hi), alpha=0.6, label="all other classes", density=True)
    ax.set_title(f"AE reconstruction error, normal = {FASHION_CLASSES[cls]} (AUC {roc_auc_score((y_test != cls).astype(int), sc):.2f})", fontsize=8); ax.set_xlabel("per-image MSE"); ax.legend(fontsize=7)
plt.tight_layout(); plt.show()
a_tr = t13[t13.normal_class == "Trouser"]
audit_trained("anomaly detection works for a distinctive class: AE ROC-AUC with 'Trouser' as the normal class is above 0.8 in every seed", lambda: bool((a_tr["AE"] > 0.8).all()), f"{pm(a_tr['AE'], 3)}")
easy, hard = t13[t13.normal_class == "Trouser"]["AE"].mean(), t13[t13.normal_class == "Shirt"]["AE"].mean()
print(f"AE ROC-AUC: Trouser {easy:.3f} · Shirt {hard:.3f} – " + ("the shirt, which looks like many other classes, is harder." if easy > hard else "no clear easy/hard difference in this run."))
print("Why a reconstruction score can fail: an autoencoder that is too powerful (or trained on a class that is a 'superset' of the shapes of the others) reconstructs the anomalies well too; that is why the score for a visually simple class such as sneakers or shirts is not guaranteed to separate.")
EX[13] = ("done", "Exercise 13 section: AE, VAE and PCA reconstruction-error scores, ROC-AUC for three normal classes, 3 seeds")
```

### Short answers to the conceptual exercises
1. **Main tasks of autoencoders:** feature extraction / learning compact representations, **dimensionality reduction** and **visualisation** (code → t-SNE), **unsupervised pretraining** of other networks, **denoising** and repair, **anomaly detection** by reconstruction error, and – in the variational form – **generation** of new data; also data compression and recommendation-style embeddings (sections 2–6, Capstone A, Exercise 13).
2. **Lots of unlabeled data, a few thousand labels:** first train an autoencoder (a denoising or stacked one; a convolutional one for images) on *all* the data, labelled and unlabeled. Then take the **lower layers of the encoder**, add a small output layer for the task, and train it on the labelled instances – with the encoder **frozen** (very few labels) or **fine-tuned** at a small learning rate (more labels). Compare with a randomly initialised network over several seeds (Phase 3: 500 labels, Capstone B). An alternative of the same kind is greedy layer-wise pretraining.
3. *(answered in section 2.7.)* A perfect reconstruction does not make an autoencoder good: it can be the identity or memorise the training set. Evaluate on held-out data, by looking at reconstructions, with a downstream task on the codings, with codings visualisations and with corrupted inputs.
4. **Undercomplete:** the coding is *smaller* than the input, so the network must compress; the risk of an *excessively* undercomplete one is that it cannot reconstruct the input well (it loses information that matters). **Overcomplete:** the coding is *as large as or larger* than the input; the risk is that it simply copies the input (the identity) or memorises, learning no useful features – unless a constraint is added (noise, sparsity, a bottleneck penalty). See Phase 5.
5. *(answered in section 3.3.)* Tie by making the decoder layer use the **transpose of the encoder's kernel** (`DenseTranspose`), with separate biases. The point: about half the weights (163,814 → 82,414 here), faster training, less overfitting.
6. A **generative model** can produce new instances that look like they come from the training distribution (it models, or at least samples from, the data distribution). A generative autoencoder is the **variational autoencoder (VAE)**: probabilistic (its coding is a sample from a learned Gaussian), and its coding space is regularised to be smooth, so decoding random codings gives new images (also: adversarial autoencoders, which are not covered here).
7. A **GAN** is made of two networks that play a game: a **generator** that turns random noise into fake data and a **discriminator** that tells real from fake; the generator is trained to fool the discriminator, the discriminator to catch the generator, and at the (ideal) equilibrium the fakes are indistinguishable from the real data. They shine in **image generation / super-resolution / inpainting**, **style transfer and image-to-image translation**, **data augmentation**, turning sketches into photos, text-to-image, generating faces and **high-resolution synthesis** (ProGAN, StyleGAN), and in producing realistic training data when real data is scarce.
8. **Difficulties:** (i) the training is a **two-player game** – there is no single falling loss, and a Nash equilibrium is not guaranteed to be reached; (ii) it can **oscillate or diverge** and is very **sensitive to hyper-parameters** and to the seed (8.2); (iii) **mode collapse** – the generator keeps producing a few outputs (or forgets modes and cycles between them) because the discriminator only judges one batch at a time (Phase 7); (iv) the discriminator can **overpower** the generator (vanishing generator gradient) or the reverse; (v) **no reliable metric** of quality and diversity (8.8). Remedies: experience replay, minibatch discrimination, feature matching, DCGAN design rules, Wasserstein-style losses, progressive growing, equalised learning rate, pixelwise normalisation.

### Honest limits of this notebook
* **Sizes.** Everything ran on a 2-core CPU. Autoencoders train for 5–10 epochs rather than the longer runs of the book; the **DCGAN, the conditional DCGAN, the short sensitivity runs and the toy StyleGAN are scaled by *steps* (`dc_steps`, `cgan_steps`, `dc_var_steps`, `style_steps`), not by the brief's 50 epochs** – a 50-epoch DCGAN would take hours here. Their pictures are therefore early-training pictures; the sensitivity study and the replay comparison use single short runs (two seeds for replay), and gaps are only called findings when they exceed the seed spread.
* **Judging generators.** Quality and diversity are measured with a helper Fashion MNIST classifier (class coverage, class entropy, confidence), a Laplacian-variance sharpness score and a feature-space Fréchet distance computed with the same helper network's features. These are **not** the published Inception-based FID, and numbers are only comparable inside this notebook.
* **Substitutes and deviations.** Datasets come from GitHub mirrors (Fashion MNIST) and a pickled MNIST copy; the KL regulariser keeps the brief's weight although Keras divides activity penalties by the batch size; equalized learning rate multiplies by √(2/fan_in) (the brief says "divide"); the minibatch-std layer is used in the brief's axes `[0, −1]` form and the batch-axis-only form is shown next to it. The "toy" progressive GAN and style generator show mechanisms, not competitive image quality; a faithful StyleGAN is out of scope.
* **Seeds.** Repeated experiments use the seeds in `RUN_SEEDS` (three). Anything run once is labelled as one run.
* **Failures are kept.** Any proof that failed is listed in the first table of this section instead of being relaxed; where a failed statement reflects reality (for example a small GAN that does not cover all classes) the text next to it says so.
